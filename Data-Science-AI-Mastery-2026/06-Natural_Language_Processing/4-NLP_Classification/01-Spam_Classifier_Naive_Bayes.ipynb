#### Naive Bayes

1.  Naive Bayes is a supervised machine learning algorithm that is used for performing the text classification.

2.  It is based on bayes theorem.

#### Types of Naive Bayes

There are two types of Naive Bayes. They are as follows:-----

1.  Gaussian Naive Bayes --------->

    It is the kind of naive bayes that can work with the continuous values.

2.  Multinomial Naive Bayes --------->

    It is the kind of naive bayes that can work with the texts easily.

    Here the input in the Multinomial Naive Bayes is the text where the input text data is converted into the numerical form.


    It can work with the texts easily because of the following:----

        (a.)    it can perform the word count frequency

        (b.)    It can perform the text classification operations easily.

#### Step 3: Define input and output

```python
#### Define the independent features
X = df['text']
```

```python
#### define the dependent features
Y = df['label']
```

#### OBSERVATIONS:

#### Step 4: Convert text into numerical features using TfidfVectorizer

```python
from    sklearn.feature_extraction.text import  TfidfVectorizer

#### create an object for TfidfVectorizer

tfidf = TfidfVectorizer()

#### using the object of TfidfVectorizer, transform the input text

X_sparse = tfidf.fit_transform(X)
```

```python
X_sparse
```

#### OBSERVATIONS:

1.  Here the sparse matrix is created that only stores non zero values and their positions and thus saves a lot of memory and their computation from being wasted.

2.  The above sparse matrix formed has 50 text messages with 116 unique vocabularies and 218 non zero words are formed and rest of the words are zero. So this is why it is known as sparse matrix.

```python
#### convert the sparse matrix into numpy array

X_array = X_sparse.toarray()
```

```python
X_array
```

#### OBSERVATIONS:

1.  The sparse matrix is converted into the numpy array for better view and visibility.

#### Step 6: Divide the independent and dependent variables  into training and testing data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X_sparse, Y, test_size=0.2,random_state=42,stratify=Y)
```

#### Step 7: Initialize Naive Bayes model

```python
from sklearn.naive_bayes import MultinomialNB

#### Create the object for Multinomial Naive Bayes model
multiNB = MultinomialNB()
```

#### OBSERVATIONS:

1.  The object for Multinomial Naive Bayes algorithm has been created.

#### Step 8: Train the model

```python
#### Train the model
multiNB.fit(X_train, Y_train)
```

#### OBSERVATIONS:

1.  The Multinomial Naive Bayes model has been trained with the help of the training data.

#### Step 8: Make predictions on the test data

```python
#### make the predictions on the test data

Y_pred = multiNB.predict(X_test)
```

```python
Y_pred
```

#### OBSERVATIONS:

1.  The output has been predicted for the given test data.

#### Step 9: Evaluate the model

```python
from sklearn.metrics import accuracy_score

ac = accuracy_score(Y_test, Y_pred)*100.0

print("Accuracy of the model is:", ac)
```

#### OBSERVATIONS:

1.  The accuracy of the model using Multinomial Naive Bayes algorithm for the Ham/spam dataset is 90 % which is very high.

2.  So the model is working completely fine.

#### Step 10: Testing on New Data

```python
new_text = ["Free money waiting for you"]

#### perform vectorization on the test data

transformed = tfidf.transform(new_text)

print(transformed)

#### predict the data
predictions = multiNB.predict(transformed)

print(predictions)

if(predictions[0] == 1):
    print("Spam")
else:
    print("Ham")
```
