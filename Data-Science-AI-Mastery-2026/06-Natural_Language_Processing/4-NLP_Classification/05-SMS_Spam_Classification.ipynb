#### Step 1: Import all the necessary libraries

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

import  nltk

from    nltk.tokenize   import  word_tokenize
from    nltk.corpus     import  stopwords
from    nltk.tokenize   import  RegexpTokenizer

from    sklearn.model_selection         import  train_test_split
from    sklearn.preprocessing           import  StandardScaler
from    sklearn.feature_extraction.text import  TfidfVectorizer
from    sklearn.metrics                 import  accuracy_score
from    sklearn.naive_bayes             import  MultinomialNB
```

#### OBSERVATIONS:

1.  MultinomialNB ----------> It deals with the test so as to perform the text classification.

#### Step 2: Load the Dataset

```python
#### Load the dataset
df = pd.read_csv('spam.csv',encoding='latin-1')
```

```python
df
```

#### OBSERVATIONS:

1.  The above dataset has two columns. They are

    (a.)    text -------->  text data

    (b.)    label ------->  label information either ham / spam

#### Step 3: Perform Data Preprocessing

```python
#### Remove all the unnecessary columns from the dataset

df.drop(columns=['Unnamed: 2','Unnamed: 3','Unnamed: 4'],inplace=True)
```

```python
df
```

#### OBSERVATIONS:

1.  All the unnecessary columns have been removed from the dataset

```python
#### Rename the column to the proper names
df.rename(columns={"v1":"label", "v2":"text"},inplace=True)
```

```python
df
```

#### OBSERVATIONS:

1.  The 'v1' and 'v2' column has been renamed to their proper names 'label' and text' respectively.

```python
#### Put the columns in their proper position
df=df[['text','label']]
```

```python
df
```

#### OBSERVATIONS:

1.  All the columns are now in the proper position.

```python
#### Check for the missing values

df.isnull().sum()
```

#### OBSERVATIONS:

1.  There are no missing values in the dataset.

```python
#### Check for the class distribution
df['label'].value_counts()
```

#### OBSERVATIONS:

1.  The dataset has 4825 ham messages and 747 spam messages. So the dataset is good.

```python
#### Convert the labels to numbers
from sklearn.preprocessing import LabelEncoder

#### create an object for Label Encoder
label = LabelEncoder()

#### using the object of LabelEncoder, transform the data
df['label'] = label.fit_transform(df['label'])
```

```python
df
```

#### OBSERVATIONS:

1.  After performing label encoding, the labels of 'spam' message has been converted into 1 and the labels of 'ham' message has been converted into 0.

#### Step 4: Divide the dataset into independent and dependent variables

```python
#### independent variables
X = df['text']
```

```python
X
```

```python
#### dependent features
Y = df['label']
```

```python
Y
```

#### OBSERVATIONS:

1.  The dataset has been divided into the independent and dependent features.

#### Step 5: Split dataset into train and test

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

```python
X_train
```

```python
X_test
```

```python
print("Shape of the input training data is:", X_train.shape)
print("Shape of the input testing  data is:", X_test.shape)
```

```python
Y_train
```

```python
Y_test
```

```python
print("Shape of the output training data is:", Y_train.shape)
print("Shape of the output testing  data is:", Y_test.shape)
```

#### OBSERVATIONS:

1.  The dataset is divided into 

    (a.)    80 % training data ------------>  4457 records

    (b.)    20 % testing data  ------------>  1115 records

#### Step 6: Convert text into numerical features using Bag Of Words

```python
from sklearn.feature_extraction.text import TfidfVectorizer

#### create an object for TfidfVectorizer
tfidf = TfidfVectorizer()

#### using the object of TfidfVectorizer, transform the text
X_train_vectorized = tfidf.fit_transform(X_train)
```

```python
X_train_vectorized
```

```python
X_test_vectorized = tfidf.transform(X_test)
```

```python
X_test_vectorized
```

#### OBSERVATIONS:

1.  The input text has been converted into sparse matrix which is a numerical representation of the words.

#### Step 7: Initialize Naive Bayes model

```python
from sklearn.naive_bayes import MultinomialNB

#### Create an object for MultinomialNB
multi_NB = MultinomialNB()
```

#### OBSERVATIONS:

1.  The object for Multinomial Naive Bayes model has been created.

#### Step 8: Train the model

```python
#### using the object of MultinomialNB, train the model
multi_NB.fit(X_train_vectorized, Y_train)
```

#### OBSERVATIONS:

1.  The Multinomial Naive Bayes model has been trained with the help of training data.

#### Step 9: Make predictions

```python
#### Make predictions
Y_pred_multinb = multi_NB.predict(X_test_vectorized)
```

```python
Y_pred_multinb
```

#### OBSERVATIONS:

1.  The predictions have been made for the test data.

#### Step 10: Evaluate model

```python
from sklearn.metrics import accuracy_score

ac = accuracy_score(Y_test, Y_pred_multinb)*100.0

print("Accuracy of the model is:", ac)
```

#### OBSERVATIONS:

1.  The accuracy of the Multinomial Naive Bayes model using SMS Spam Classification dataset is 96.05 % which is very high.

2.  So the model is working fine.

#### Step 11: Testing on New Data

```python
#### test data
new_sms = [
    "FREE entry into our prize draw. WIN cash now. Call immediately."
]

#### transform the test data
transformed = tfidf.transform(new_sms)
print(transformed)

#### perform the predictions on the test data
predictions = multi_NB.predict(transformed)
print(predictions)

#### check the probabilities of each class
prob = multi_NB.predict_proba(transformed)
print(prob)

if(predictions[0] == 1):
    print("Spam")
else:
    print("Ham")
```
