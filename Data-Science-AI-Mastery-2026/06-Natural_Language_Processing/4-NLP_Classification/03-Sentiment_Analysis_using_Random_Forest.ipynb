#### Step 4: Perform the Text Preprocessing Function

```python
#### define the function
def clean_review(text):
    #### convert the text into lower case
    text = text.lower()
    #### convert the text into words by performing word tokenization
    words = word_tokenize(text)
    #### convert the words in list into the text form
    words = ' '.join(words)
    #### create an object for RegexpTokenizer
    reg = RegexpTokenizer(r'\w+')
    #### apply the object of RegexpTokenizer to remove all the punctuations and the special symbols from the text
    ans = reg.tokenize(words)
    #### convert the text in lists to words
    ans = ' '.join(ans)
    return(ans)
```

#### OBSERVATIONS:

1.  The text preprocessing function has been defined that will clean all the reviews from the dataset using clean_text unction.

#### Step 5: Apply Cleaning

```python
#### call the function
df['clean_review'] = df['review'].apply(clean_review)
```

#### Step 7: Convert Text to Numerical Features (TF-IDF)

```python
from   sklearn.feature_extraction.text  import TfidfVectorizer

#### create an object for TfidfVectorizer
tfidf = TfidfVectorizer()

#### using the object for TfidfVectorizer, transform the text
X_sparse_matrix = tfidf.fit_transform(df['clean_review'])
```

#### Step 8: Perform the Train-Test Split

```python
#### Keep the input data in the form of sparse matrix

X_sparse_matrix
```

#### OBSERVATIONS:

1.  For splitting the data into training and testing, the data for the input must be in sparse matrix. This is because

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X_sparse_matrix, Y, test_size=0.2,random_state=42,stratify=Y)
```

#### Step 9: Train Random Forest Classifier Model

```python
from sklearn.ensemble import RandomForestClassifier

#### create an object for Random Forest Classifier
rfc = RandomForestClassifier(n_estimators = 200,random_state=42)

#### using the object of Random Forest Classifier, train the model
rfc.fit(X_train, Y_train)
```

#### OBSERVATIONS:

1.  The object for Random Forest Classifier has been created with the following parameters :----

    (a.)    n_estimators = 200 ---------->   200 decision trees are used for the estimation of the model

    (b.)    random_state = 42 ----------->   ensures the reproducibility

2.  The model is trained with the help of the training data

#### Step 10: Perform the predictions of the model

```python
#### predict the model
Y_pred = rfc.predict(X_test)
```

```python
Y_pred
```

#### OBSERVATIONS:

1.  The output has been predicted for the test data.

#### Step 11: Evaluate the Model

```python
from sklearn.metrics import accuracy_score

ac = accuracy_score(Y_test, Y_pred)*100.0
print("Accuracy score of the model is:", ac)
```

#### OBSERVATIONS:

1.  The accuracy score of the Random Forest Classifier model using imdb movie review dataset is 78 % which is very high.

2.  So the model is working fine.
