#### Step 4: Perform the Text Preprocessing Function

```python
#### define the function

#### define the function
def clean_news(text):
    #### convert the text into lower case
    text = text.lower()
    #### convert the text into words by performing word tokenization
    words = word_tokenize(text)
    #### convert the words in list into the text form
    words = ' '.join(words)
    #### create an object for RegexpTokenizer
    reg = RegexpTokenizer(r'\w+')
    #### apply the object of RegexpTokenizer to remove all the punctuations and the special symbols from the text
    ans = reg.tokenize(words)
    #### convert the text in lists to words
    ans = ' '.join(ans)
    return(ans)
```

#### Step 5: Apply Cleaning

```python
#### define the function call
df['clean_news'] = df['news'].apply(clean_news)
```

```python
df['clean_news']
```

#### OBSERVATIONS:

1.  The text preprocessing function has been defined that will clean all the reviews from the dataset using clean_news function.

#### Step 6:  Perform the Data Preprocessing Function

```python
#### Map the False as 1 and True as 0
from    sklearn.preprocessing import LabelEncoder

#### create an object for LabelEncoder
label = LabelEncoder()

#### using the object of label encoder, transform the labels
df['label'] = label.fit_transform(df['label'])
```

```python
#### Map the 0(False) as 1 and 1(True) as 0 for identifying the real and the fake news
df['label'] = df['label'].map({0:1,1:0})
```

#### OBSERVATIONS:

1.  After applying the mapping operation, all the Fake news havve been labelled as 1 and all the real news have been labelled as 0.

#### Step 7: Convert Text to Numerical Features (TF-IDF)

#### Step 9: Train Logistic Regression Model

```python
from sklearn.linear_model import LogisticRegression

#### create an object for Logistic Regression
log = LogisticRegression(max_iter=1000)

#### using the object for Logistic Regression, train the model
log.fit(X_train, Y_train)
```

#### OBSERVATIONS:

#### Step 10: Train the Support Vector Classifier Model

```python
from sklearn.svm import SVC

#### Create an object for SVC
svc = SVC(kernel='linear')

#### using the object of SVC, train the model
svc.fit(X_train, Y_train)
```

#### OBSERVATIONS:

1.  The object of SVC model has been created with kernel as linear means it can handle linear data

#### Step 12: Evaluate the Model

```python
#### Evaluate the model for Logistic Regression

aclog = accuracy_score(Y_test, Y_pred_log)*100.0
print("Accuracy score for the Logistic Regression model is:", aclog)
```

```python
#### Evaluate the model for SVC

acsvc = accuracy_score(Y_test, Y_pred_svc)*100.0
print("Accuracy score for the Support Vector Classifier model is:", acsvc)
```
