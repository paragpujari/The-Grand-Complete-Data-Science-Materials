#### Step 3: Read the imdb Dataset

```python
#### Load the dataset
import pandas as pd

# Load full dataset (example path)
df = pd.read_csv("IMDB_Dataset.csv")

# Take a balanced sample
df_small = df.groupby("sentiment").sample(n=500, random_state=42)

# Reset index
df_small = df_small.reset_index(drop=True)

# Save reduced dataset
df_small.to_csv("IMDB_reduced.csv", index=False)

print(df_small.shape)
```

#### Step 4: Perform the Text Preprocessing Function

```python
#### define the function for text pre-processing
def clean_text(text):
    #### convert the text into lower case
    text = text.lower()
    #### perform the word tokenization on the text
    words = word_tokenize(text)
    #### convert the words in list into the text form
    words_text = ' '.join(words)
    #### perform the regular expression on the text
    reg = RegexpTokenizer(r'\w+')
    #### apply the object of regular expression tokenizer on the text
    ans = reg.tokenize(words_text)
    #### conveer the text in lists to words
    ans = ' '.join(ans)
    return(ans)
```

#### OBSERVATIONS:

1.  The text preprocessing function has been defined that will clean all the reviews from the dataset using clean_text unction.

#### Step 5: Apply Cleaning

```python
#### call the function for text cleaning
df_small['clean_review'] = df_small['review'].apply(clean_text)
```

```python
df_small['clean_review']
```

#### OBSERVATIONS:

1.  The clean_text function has been called that will clean the 'review' column and store the processed text in the 'clean_review' column of the dataset.

#### Step 6: Perform the Data Preprocessing Function

```python
#### Map the labels 'positive' ---------> 1 and 'negative' ------------> 0

df_small['sentiment'] = df_small['sentiment'].map({'positive':1, 'negative':0})
```

```python
df_small
```

#### OBSERVATIONS:

1.  The positive category of the sentiment column has been mapped to 1

2.  The negative category of the sentiment column has been mapped to 0.

```python
#### Remove the column 'review' from the dataset
df_small.drop(columns='review',inplace=True)
```

```python
df_small
```

```python
#### Reorder the position of 'clean_review' and 'sentiment' in the dataframe
df_small=df_small[['clean_review','sentiment']]
```

#### Step 7: Convert Text to Numerical Features (TF-IDF)

```python
#### Convert the review into Tfidf Vectorizer
from sklearn.feature_extraction.text import TfidfVectorizer

#### create an object for TfidfVectorizer
tfidf = TfidfVectorizer()

#### using the object of TfidfVectorizer, transform the text
X_sparse_matrix = tfidf.fit_transform(df_small['clean_review'])
```

```python
#### convert the sparse matrix into the numpy array
X_array = X_sparse_matrix.toarray()
```

#### Step 8: Perform the Train-Test Split

```python
#### Keep the input data in the form of sparse matrix

X_sparse_matrix
```

#### OBSERVATIONS:

1.  For splitting the data into training and testing, the data for the input must be in sparse matrix. This is because

    (a.)    converting it into the numpy array means converting the sparse matrix into dense matrix which makes the data more complex

    (b.)    It even causes more huge dense memory causing it for crashing.

```python
from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test = train_test_split(X_sparse_matrix,Y,test_size=0.2,random_state=42,stratify=Y)
```

#### Step 9: Train Logistic Regression Model

```python
from sklearn.linear_model import LogisticRegression

#### create an object for Logistic Regression
log = LogisticRegression(max_iter=1000)

### using the object of Logistic Regression, train the model
log.fit(X_train, Y_train)
```

#### OBSERVATIONS:

1.  The logistic regression has been trained with the help of the training data.

#### Step 10: Make Predictions for the model

```python
#### make the predictions
Y_pred = log.predict(X_test)
```

```python
Y_pred
```

#### OBSERVATIONS:

1.  The output has been predicted for the test data.

#### Step 11: Evaluate the Model

```python
from sklearn.metrics import accuracy_score

ac = accuracy_score(Y_test, Y_pred)*100.0

print("Accuracy score of the model is:", ac)
```

#### OBSERVATIONS:

1.  The accuracy score of the Logistic regression for imdb dataset is 81 % which is very high.

2.  So the model is working fine.

#### Step 12: Perform testing with the custom sentence

```python
#### define the custom sentence

new_text = ["I really love this amazing course"]

#### perform the vectorization of the text
transformeddata = tfidf.transform(new_text)
print(transformeddata)

#### perform the prediction on the transformed data
predictions = log.predict(transformeddata)

print(predictions)

if(predictions[0]==1):
    print("Positive")
else:
    print("Negative")
```
