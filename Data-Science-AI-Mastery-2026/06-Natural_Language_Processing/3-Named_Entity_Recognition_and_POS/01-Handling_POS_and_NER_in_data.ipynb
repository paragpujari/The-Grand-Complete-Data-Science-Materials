#### Parts Of Speech Tagging

1.  Parts of Speech Tagging (POS Tagging) refers to the process of assigning the parts of speech to each and every word in the given text.

#### Named Entity Recognition

1.  Named Entity Recognition helps in identifying and specifying all the important entities into the specific categories.


Ex:---  "Elon Musk founded Tesla in California in 2003."

### POS Tagging

1.  Elon Musk ------------> Noun

2.  founded   ------------> Verb

3.  Tesla     ------------> Noun

4.  California -----------> Noun

5.  2003       -----------> Year

Ex :---  'Elon Musk founded Tesla in California in 2003'

### Named Entity Recognition

1.  Elon Musk -------------->  Person

2.  Tesla     -------------->  Org

3.  California ------------->  LOC/GPE

4.  2003      -------------->  DATE

#### Step 3: Define the Sample Text

```python
# Sample text containing punctuation and special characters
text = "Hello!!! My name is Parag Pujari :) #DataScience @2026 $$$ Welcome to NLP--world!!!"

print(text)
```

#### OBSERVATIONS:

1.  Here the sample text has been defined.

#### Step 4: Tokenize the text

```python
from nltk.tokenize import word_tokenize

#### perform the word tokenization on the above text

words = word_tokenize(text)

print(words)
```

#### OBSERVATIONS:

1.  The entire text has been divided into the words.

#### Step 5: Define the english stop words and remove the stop words from the text

```python
#### define all the english stop words

from nltk.corpus import stopwords

english_stopwords = set(stopwords.words("english"))

print(english_stopwords)
```

```python
#### Remove the english stop words from the text

res = [x for x in words if (x not in english_stopwords)]
```

```python
res
```

```python
#### convert the filtered text in text form

res = ' '.join(res)
```

```python
res
```

#### Step 6: Use the Regular Expression Tokenization and remove all the punctuations

```python
from nltk.tokenize import RegexpTokenizer

#### create an object for Regular Expression Tokenizer

reg = RegexpTokenizer(r'\w+')

#### Apply the object on the text to remove all the punctuations and special symbols from the text

res = reg.tokenize(res)
```

```python
res
```

#### OBSERVATIONS:

1. After applying regular expression on the text, all the punctuations and the special symbols have been removed from the text.

#### Step 7: Perform Parts of Speech Tagging (POS) Tagging on the refined words in the text

```python
from nltk import pos_tag

#### Perform POS Tagging on each and every tokenized words in the list

ans = pos_tag(res)
```

```python
ans
```

#### Step 12: Perform the POS based Chunking on the text


It is a technique of grouping all the same POS Tags into one group

```python
from nltk import RegexpParser
```

```python
#### Define the chunking grammar

grammar = r"""
    NP: {<DT>?<JJ>*<NNP>+}   # Noun Phrase: optional determiner, adjectives, proper nouns
"""
```

```python
grammar
```

#### OBSERVATIONS:

1.  It defines the grammar structure with Noun Phrase followed determiner then Adjective with one or more occurances of Proper Noun.

```python
#### Create an object for RegexpParser and create the chunk parser

reg = RegexpParser(grammar)
```

```python
reg
```

#### OBSERVATIONS:

1.  Here the chunk parser is created where it accepts the input as the grammar and groups all the words belonging to one phrase.

```python
#### Apply the POS Chunking on the POS Tags

tree = reg.parse(postagdata)
```

```python
tree
```

#### OBSERVATIONS:

1.  POS Chunking creates a syntax tree where all words belonging to one phrase is in one group.
