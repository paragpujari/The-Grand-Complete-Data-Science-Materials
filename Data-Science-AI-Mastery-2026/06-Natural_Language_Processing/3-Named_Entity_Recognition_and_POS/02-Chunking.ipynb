#### Parts Of Speech Tagging

1.  Parts of Speech Tagging (POS Tagging) refers to the process of assigning the parts of speech to each and every word in the given text.

#### Named Entity Recognition

1.  Named Entity Recognition helps in identifying and specifying all the important entities into the specific categories.


Ex:---  "Elon Musk founded Tesla in California in 2003."

### POS Tagging

1.  Elon Musk ------------> Noun

2.  founded   ------------> Verb

3.  Tesla     ------------> Noun

4.  California -----------> Noun

5.  2003       -----------> Year

Ex :---  'Elon Musk founded Tesla in California in 2003'

### Named Entity Recognition

1.  Elon Musk -------------->  Person

2.  Tesla     -------------->  Org

3.  California ------------->  LOC/GPE

4.  2003      -------------->  DATE


####  Perform a simple end-to-end Python implementation of Handling punctuation & special characters using NLTK

##### Steps used in this Algorithm:---

1.  Import all the necessary libraries

2.  Download all the NLTK resources

3.  Define the Sample Text

4.  Tokenize the text

5.  Define the english stop words and remove the stop words from the text

6.  Use the Regular Expression Tokenization and remove all the punctuations

7.  Perform POS Tagging on the  words in the text

8.  Performing Named Entity Recognition 

9.  Performing Chunking 

#### Step 1: Import all the necessary libraries

```python
import numpy                as np
import pandas               as pd
import matplotlib.pyplot    as plt
import seaborn              as sns

import nltk

from   nltk.tokenize import word_tokenize, sent_tokenize
from   nltk.stem     import PorterStemmer, WordNetLemmatizer
from   nltk.corpus   import stopwords
from   nltk          import ne_chunk,pos_tag
```

#### Step 2: Download the necessary resources required for NLTK

```python
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker')
nltk.download('treebank')
```

#### Step 3: Define the Sample Text

```python
#### Define te sample text

text = "Hey!!! I am learning NLP with Python :) #MachineLearning @2026 $$$ This is an amazing--journey into AI world!!!"

print(text)
```

#### Step 4: Tokenize the text

```python
from nltk.tokenize import word_tokenize

#### Apply word tokenization on the above text

words = word_tokenize(text)
```

```python
words
```

#### OBSERVATIONS:

1.  The text has been broken into the words after applying the word tokenization on it.

#### Step 5: Define the english stop words and remove the stop words from the text

```python
#### define the english stop words

from nltk.corpus import stopwords
english_stopwords = set(stopwords.words("english"))
```

```python
english_stopwords
```

#### OBSERVATIONS:

1.  All the english stop words have been defined in the text.

```python
#### Remove the english stop words from the text

res = [x for x in words if(x not in english_stopwords)]
```

```python
res
```

```python
#### Convert the filtered text in text form

res = ' '.join(res)
```

```python
res
```

#### Step 6: Use the Regular Expression Tokenization and remove all the punctuations

```python
from nltk.tokenize import RegexpTokenizer

#### Create an object for Regular Expression Tokenizer

reg = RegexpTokenizer(r'\w+')

#### Apply the object of Regular Expression Tokenizer and remove all the punctuations from the text

ans = reg.tokenize(res)
```

```python
ans
```

#### OBSERVATIONS:

1.  All the punctutaions and the special symbols have been removed from the text.

#### Step 7: Perform POS Tagging on the  words in the text

```python
import spacy
```

#### OBSERVATIONS:

1. spacy is an advanced library used in NLP that performs

    (a.)    POS Tagging

    (b.)    Named Entity Recognition

    (c.)    Document parsing

```python
#### Load a pre trained englisg model

nlp = spacy.load("en_core_web_sm")
```

#### OBSERVATIONS:

1.  en_core_web_sm is a pretrained english model that can work on the small datasets but fails on the lareg datasets.

2.  It is loaded by spacy and is stored in nlp object.

3.  The main role of en_core_web_sm is that it knows already about :-----

    (a.)    grammatical rules

    (b.)    POS Tags

    (c.)    Document Parsing

```python
text = "Sundar Pichai joined Google in California in 2004."
```

```python
#### Apply the NLP on the text

obj = nlp(text)
```

```python
obj
```

#### OBSERVATIONS:

1.  The text is trained on the nlp library and performs well on

    (a.) POS Tagging
 
    (b.) Named Entity Recognition    

```python
#### Perform the POS Tagging on the text object

for x in obj:
    print(x.text,"------>",x.pos_,"--------->",x.tag_)
```

#### OBSERVATIONS:

1.  Here the POS Tag is obtained for each and every word in the sentence along with the detailed pos.

#### Step 8: Performing Named Entity Recognition on the text object

```python
#### Perform the NER on the text object

for x in obj.ents:
    print(x.text,"------->",x.label_)
```

#### OBSERVATIONS:

1.  It loops through all the named entities and assigns the categories to all the named entities in the text.

#### Step 9: Performing Chunking on the tagged words in the sentences 

#### (a.) (Named Entity Chunking)

```python
#### Perform POS Tags on the text

text = "Sundar Pichai joined Google in California in 2004."

#### Perform word tokenization on the above text
words = word_tokenize(text)

print(words)

#### Perform POS Tags on the tokenized words
ans = pos_tag(words)
```

```python
ans
```

#### OBSERVATIONS:

1.  The Parts Of Speech has been assigned to each and every word in the text.

```python
#### Apply Named Entity Chunking on the tokenized pos tags

ansnechunk = ne_chunk(ans)
```

```python
ansnechunk
```

#### OBSERVATIONS:

1.  It displays a parse tree where all the named entities belonging to one categories are placed together and assigned one label to them.
