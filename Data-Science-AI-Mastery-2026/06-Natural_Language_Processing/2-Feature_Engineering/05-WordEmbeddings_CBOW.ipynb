## **Word Embeddings**

1.  Word Embedding refers to the technique of converting each and every input word into the numerical dense dimensional vectors.

2.  The words with the similar meanings with have the same dense dimensional vectors.


## **Example:**

| Word  | Vector                 |
| ----- | ---------------------- |
| King  | [0.25, 0.81, 0.44,...] |
| Queen | [0.28, 0.79, 0.47,...] |
| Apple | [0.91, 0.12, 0.33,...] |


## **Importance of Word Embeddings:**

1.  It is used to find out the similarities between two words by computing the similarity scores in the form of dense dimensional numerical vectors.

2. It helps in finding out the meaning behind each and every word in the input text.

3.  It is used to perform the dimensionality reduction.

4.  It helps in improving the speed and the accuracy of the model.

5.  It helps in finding out the relation of one word with respect to the other based on the similarity scores.

**Example:**

1.  king - man + woman = queen

## **CBOW**

1.  CBOW stands for Continuous Bag Of Words.

2.  The main idea behind this is to predict the target word based on the surrounding context  input word.

Ex:   I love machine learning

Context window = 2, it means we need to predict the target word based on the 2 surrounding context input word before it and 2 surrounding context input word after it.

      I love _ learning

      It forms the context window of the size 2 and forms the context input words

      I,love,_,learning

      Based on the above context input words, it predicts the target word: machine

      I,love,machine,learning.

## **How CBOW Works Internally**

1.  All the context input words are converted into the dense dimensional vectors.

2.  Then the dense dimensional numerical vectors are all averaged.

3.  This provides the probability of the target word.

## **Step 6: Perform the normalization on the corpus text**

```python
#### Remove all the punctuations and the special symbols from the text

from nltk.tokenize import RegexpTokenizer

#### create an object for Reg exp tokenizer
reg = RegexpTokenizer(r'\w+')

#### apply the object of reegular expression on the text to remove all the punctuations and special symbols from the text
ans = reg.tokenize(res)
```

## **Step 7: Define the list of tokenized sentences**

```python
### define the list of tokenized sentences

corpus = [
    "Natural Language Processing and Machine Learning are closely related fields.",
    "Word embeddings are a type of word representation that allows words to be represented as vectors.",
    "Deep learning models are used for creating high-quality word embeddings.",
    "The meaning of words can be captured by training on large text datasets.",
    "Gensim is a popular library for training Word2Vec models in Python."
]
```

```python
#### perform the word tokenization on the corpus text

res = [word_tokenize(x.lower()) for x in corpus if(x != '')]

print(res)
```

## **OBSERVATIONS:**

1. After applying word tokenization on each sentences in the list, it has become the list of all the tokenized sentences that can be the input for the CBOW Word2Vec model.

## **Step 8: Train the Word2Vec model**

```python
## Train the Word2Vec model

cbow = Word2Vec(
    sentences   =  res             ,
    vector_size =   50             ,
    window      =    3             ,
    min_count   =    1             ,
    sg          =    0             ,
    epochs      =    100
)
```

```python
cbow
```

## **OBSERVATIONS:**

1.  The cbow Word2Vec model has been created with the help of the following parameters :-----

    (a.)  sentences = res -------->  list of all the tokenized sentences

    (b.)  vector    =  50 -------->  size of the dense dimensional vectore form of input word

    (c.)  window = 3 -------------->  size of context window. if window = 3 then 3 context window before the target and 3 after the target.

    (d.)  min_count = 1------------->  minimum number of words needed to form a vocabulary.

    (e.)  sg = 0 ------------------>  Type of model architecture where the input context word is predicted from the target word.

    (f.)  epochs = 100 ------------>  No of iterations needed to train the model.

## **Step 9: Get the list of all the vocabularies from the dictionaries**
