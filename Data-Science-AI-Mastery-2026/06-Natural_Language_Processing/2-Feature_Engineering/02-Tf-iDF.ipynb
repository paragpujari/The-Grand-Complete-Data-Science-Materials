#### Tf- iDF Vectorizer

    Tf-iDF Vectorizer stands for Term Frequency Inverse Document Frequency.

    It is an improvised version of Bag Of Words(BOW).

    It plays an important role in determining how important a particular word is in the document by not only determining the number of times a particular term appears in the document but also assigning the weights to each and every word in the document.

    So by this the most important word in the document will have the maximum weight and the least important word will have the minimum weight.

    The output of Tf-iDF matrix will be in numerical form of the weights.

    Tf ----->  Term Frequency ----------->  It determines how many times a particular term appears in the text document.

    iDF -----> Inverse Document Frequency -------> It determines how unique a particular term appears across the entire corpus.

#### Condition of Tf-iDF Vectorizer:---

1.  A Tf-iDF Vectorizer is said to be high if and only if

        (a.)    a particular term appears in a particular text document.

        (b.)    a psrticular terms appears uniquely accross the entire corpus.

#### Significance of Tf-iDF Vectorizer:-----

1.  It helps in determining the importance of each and every word in the document.

2.  It is used in search engines.

3.  It is used in the reduction of the noise.

4.  It helps in improving the speed and accuracy of the model.

1.  Import all the necessary libraries

2.  Import the object for IFiDFVectorizer

#### Step 1: Import all the necessary libraries

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

from    sklearn.feature_extraction.text import TfidfVectorizer
```

#### Step 2: Import the object of TfidfVectorizer

```python
#### Create an object for Tfidfvectorizer

tfidf = TfidfVectorizer()
```

#### Step 3: Define the Sample Text

```python
corpus = [
    "I love data science and machine learning",
    "Data science is fun and exciting",
    "I love deep learning and neural networks",
    "Machine learning is a part of data science"
]
```

```python
corpus
```

#### Step 4: Transform the text into Sparse Matrix

```python
#### Transform the text into sparse matrix

sparse_matrix = tfidf.fit_transform(corpus)
```

```python
sparse_matrix
```

#### OBSERVATIONS:

1.  The corpus text has been converted into the sparse matrix because to store all the non zero values and their positions and prevent the unnecessary memory wastage.

#### Step 5: Convert the sparse matrix into dense matrix

```python
#### Convert the sparse matrix into dense matrix

dense_matrix = sparse_matrix.toarray()
```

```python
dense_matrix
```

#### OBSERVATIONS:

1.  The sparse matrix has been converted into the dense matrix and each row in the dense matrix has been converted into the numerical vector form.

2.  The numerical vector form of the values are in the form between 0 to 1 in fractional form.

3.  The columns with the high numerical values specifies higher weights have been given to them, so this specifies the importance of each word in the document.

4.  The highest value specifies the most important word and the lowest value specifies the least important word in the document.

#### Step 6: Construct the DataFrame from the dense matrix

```python
df = pd.DataFrame(dense_matrix, columns = tfidf.get_feature_names_out())
```

```python
df
```

#### OBSERVATIONS:

1.  The above dataframe represents the tfidf matrix with the numerical values ranging between 0 to 1 in fractional form.

2.  The word with the highest value signifies the most imporatnt word.

3.  The word with the least   value signifies the least important word.

4.  This matrix helps in identifying the most important word from the document.
