#### Bi-Grams, Tri-Grams, N-Grams

1.  Bi-Grams ---------------------> It can be defined as a sequence of two consecutive words.

2.  Tri-Grams --------------------> It can be defined as the sequence of three consecutive words.

3.  N-Grams ----------------------> It can be defined as a sequence of N-Consecutive words.


#### Importance of Bi-Grams, Tri-Grams, N-Grams

1.  It is used for the auto completion of the sentences

2.  It is used for sentimental analysis

3.  It is used for sentence completion

4.  It is used for text analysis

5.  It is used for text similarity

6.  It is used for text prediction

7.  It is used for Machine Translation.

#### Step 1: Import all the necessary libraries

```python
from    nltk import word_tokenize, ngrams
```

#### OBSERVATIONS:

1.  word_tokenize ---------> divides the text into words.

2.  ngrams       ----------> A sequence of N Consecutive words are formed from the words in list

#### Step 2: Define the text

```python
#### define the text
text = "I love learning data science"
```

```python
text
```

#### Step 3: Apply Word Tokenization on the text

```python
#### apply word tokenization on the above text

words = word_tokenize(text)
```

```python
words
```

#### OBSERVATIONS:

1.  On applying word tokenization on the above text, the text has been divided into individual words.

#### Step 4: Apply Bi-Gram on the tokenized words

```python
from nltk import ngrams

bigram = list(ngrams(words,2))
```

```python
print(bigram)
```

#### OBSERVATIONS:

1.  On applying ngrams on the tokenized words, it creates a window of the size 2 and forms a sequence of two consecutive words.

#### Step 5: Apply Tri-Gram on the tokenized words

```python
trigram = list(ngrams(words,3))
```

```python
trigram
```

#### OBSERVATIONS:

1.  On applying ngrams on the tokenized words, it creates a window of the size 3 and forms a sequence of three consecutive words.

#### Step 6: Apply Quarter-Gram on the tokenized words

```python
quartergram = list(ngrams(words,4))
```

```python
quartergram
```

#### OBSERVATIONS:

1.  On applying ngrams on the tokenized words, it creates a window of the size 4 and forms a sequence of four consecutive words.

#### Step 6: Apply N-Gram on the tokenized words

```python
pentagram = list(ngrams(words,5))
```

```python
pentagram
```

#### OBSERVATIONS:

1.  On applying ngrams on the tokenized words, it creates a window of the size 5 and forms a sequence of five consecutive words.
