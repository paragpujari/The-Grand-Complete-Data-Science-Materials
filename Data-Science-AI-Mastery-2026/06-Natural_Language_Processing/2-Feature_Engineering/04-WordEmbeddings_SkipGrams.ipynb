#### **Word Embeddings**

1.  Word Embeddings is a text representation technique that is used to convert each and every word into the dense numerical vector.

2.  The words with the similar meanings will have the same dense valued representations.

## **Example:**

| Word  | Vector                 |
| ----- | ---------------------- |
| King  | [0.25, 0.81, 0.44,...] |
| Queen | [0.28, 0.79, 0.47,...] |
| Apple | [0.91, 0.12, 0.33,...] |


## **Importance of Word Embeddings:**

1.  It is used to find out the similarities between two words by computing the similarity scores in the form of dense dimensional numerical vectors.

2. It helps in finding out the meaning behind each and every word in the input text.

3.  It is used to perform the dimensionality reduction.

4.  It helps in improving the speed and the accuracy of the model.

5.  It helps in finding out the relation of one word with respect to the other based on the similarity scores.



**Example:**

1.  king - man + woman = queen

## **Skip Grams**

1.  The main idea behind the Skip-Grams is to predict the surrounding context input word based on the target word.

Ex:  I love machine learing.

  Let us consider, the target word = machine. Based on the target word, we need to predict the surrounding context input words.

  Let us consider the context window is of the size 2, it means we need to predict 2 surrounding context input word before the target word and 2 surrounding context input word after the target word.

  If the target = 'machine' ---------> 2 surrounding context input words predicted before the target are ('I','love') and after the target is ('learning').

## **How Skip Grams Works Internally**

1. The target word is converted into the dense dimensional numerical vector.

2. It is passed into the neural network model.

3. The target word predicts all the context input word independently.

4. Each context word predicts the probability score based on the softmax layer.

## **Step 3: Define the sample text**

```python
#### define the sample text

corpus = [
    "Natural Language Processing and Machine Learning are closely related fields.",
    "Word embeddings are a type of word representation that allows words to be represented as vectors.",
    "Deep learning models are used for creating high-quality word embeddings.",
    "The meaning of words can be captured by training on large text datasets.",
    "Gensim is a popular library for training Word2Vec models in Python."
]
```

## **Step 4: Perform the normalization of the text**

```python
#### convert the corpus to text form

corpus = " ".join(corpus)
```

```python
corpus
```

```python
from nltk.tokenize import RegexpTokenizer

#### Create an object for RegexpTokenizer

reg = RegexpTokenizer(r'\w+')

#### using the object of Regular expression tokenizer, remove all the punctutaions from the text
ans = reg.tokenize(corpus)

#### convert all the filtered text from lists to words
ans = " ".join(ans)
```

```python
ans
```

## **OBSERVATIONS:**

1.  Here the normalization of the corpus text has been performed in which all the punctutaions and the special symbols have been removed from the text.

##  **Step 5:  Perform the Word tokenization on the text**

```python
from nltk.tokenize import word_tokenize

#### perform the word tokenization on the filtered text
words = word_tokenize(ans)

#### convert all the words in the text to lower case
words = [x.lower() for x in words]
print(words)
```

## **Step 6: Define all the english stop words and remove them from the text**

```python
#### define all the english stopwords
from nltk.corpus import stopwords
english_stopwords = set(stopwords.words("english"))

#### remove all the stopwords from the text
res = [x for x in words if (x not in english_stopwords)]
```

```python
res
```

## **OBSERVATIONS:**

1.  All the english stop words have been removed from the input text.

## **Step 7: Define the list of tokenized sentences**

```python
#### Define the corpus text

corpus = [
    "Natural Language Processing and Machine Learning are closely related fields.",
    "Word embeddings are a type of word representation that allows words to be represented as vectors.",
    "Deep learning models are used for creating high-quality word embeddings.",
    "The meaning of words can be captured by training on large text datasets.",
    "Gensim is a popular library for training Word2Vec models in Python."
]

```

```python
##### To construct the list of all the tokenized words in the list

res = [word_tokenize(x.lower()) for x in corpus]

res
```

## **OBSERVATIONS:**

1.  After performing word tokenization, we have obtained the list of all the tokenized sentences which can be the input for Word2Vec model.

## **Step 8: Train the Word2Vec model**

```python
#### Train the Word2Vec model
skipgrams = Word2Vec(
    sentences     = res          ,
    vector_size   = 50           ,
    window        = 3            ,
    min_count     = 1            ,
    sg            = 1            ,
    epochs        = 100
)
```

```python
skipgrams
```

## **OBSERVATIONS:**

1.  The skip grams Word2Vec model has been created with the help of the following parameters :-----

    (a.)  sentences = res -------->  list of all the tokenized sentences

    (b.)  vector    =  50 -------->  size of the dense dimensional vectore form of input word

    (c.)  window = 3 -------------->  size of context window. if window = 3 then 3 context window before the target and 3 after the target.

    (d.)  min_count = 1------------->  minimum number of words needed to form a vocabulary.

    (e.)  sg = 1 ------------------>  Type of model architecture where the input context word is predicted from the target word.

    (f.)  epochs = 100 ------------>  No of iterations needed to train the model.
