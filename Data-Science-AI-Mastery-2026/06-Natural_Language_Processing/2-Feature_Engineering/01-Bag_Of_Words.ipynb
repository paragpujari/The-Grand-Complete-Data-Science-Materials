#### Bag Of Words(BOW) :-----

1.  Bag Of Words is a text representation technique that accepts the corpus text as an input and converts it into the numerical vectors so that it can easily be trained on the Machine Learning models.

#### Importance of Bag Of Words(BOW) :--------

1.  Bag Of Words(BOW) is used to convert the input corpus text into the numerical vectors.

2.  The output of the model is in the form of 0 or 1.

3.  It can easily be trained by the machine learning models.

4.  It is useful for feature extraction.

5.  It is applicable for smaller datasets.


####  Steps used in this Algorithm:---

1.  Import all the necessary libraries

2.  Import the object for CountVectorizer

3.  Define the Sample Text

4.  Transform the text into vector

5.  Convert vector into Numpy Array

6.  Get the columns from the vectors

7.  Get the DataFrame from the matrix

#### Step 1:  Import all the necessary libraries

```python
import  numpy   as  np
import  pandas  as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

from    sklearn.feature_extraction.text import  CountVectorizer
```

#### OBSERVATIONS:

1.  CountVectorizer --------->  

        (a.) It is a text representation technique that is used in case of small datasets.

        (b.) It determines the number of times a particular word appears in the document.

#### Step 2: Create an object for Count Vectorizer

```python
#### Create an object for Count Vectorizer
count = CountVectorizer()
```

#### Step 3: Define the sample text

```python
corpus = [
    "I love data science and machine learning",
    "Data science is amazing",
    "I love coding in Python",
    "Machine learning and deep learning are related fields"
]
```

```python
corpus
```

#### Step 4: Apply Count Vectorization on the corpus text and convert it into sparse matrix

```python
#### Transform the corpus into sparse matrix
sparse_matrix = count.fit_transform(corpus)
```

```python
sparse_matrix
```

#### OBSERVATIONS:

1.  The main reason for converting the corpus text into the sparse matrix is because:----

    (a.)    to store only the non zero values and their positions and save the memory.

#### Step 5: Convert the sparse matrix into numpy array

```python
#### Convert the sparse matrix into the dense matrix
dense_matrix = sparse_matrix.toarray()
```

```python
dense_matrix
```

#### OBSERVATIONS:

1.  The sparse matrix is converted into the dense matrix (list inside the list).

2.  Each row of the dense matrix represents the numerical vector comprising of zeros and ones.

3.  This is done for better view and visibility.

4.  This is how the corpus text is converted into the numerical vector.

#### Step 6: Get the columns from the vectors

```python
columns = count.get_feature_names_out()

print("Column obtained from the vector is:", columns)
```

#### Step 7: Construct the DataFrame from the matrix

```python
df = pd.DataFrame(dense_matrix, columns=count.get_feature_names_out())
```

```python
df
```

#### OBSERVATIONS:

1.  The above dataframe represents the sparse matrix where each row represents each sentence and the words represents the attributes of the dataframe.

2.  The number of times each word appears in the document is mentioned as its count else 0.

3.  So in this way, the entire corpus text is converted into the  matrix where each row represents the numericl vectors in 0 or 1 that can easily be trained by the machine learning model.
