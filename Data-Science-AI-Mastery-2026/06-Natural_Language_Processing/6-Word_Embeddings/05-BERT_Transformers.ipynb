1. BERT stands for Bidirectional Encoder Representations from Transformres. It refers to the technique of converting the text into the dense dimensional vectors known as Embeddings.

2.  These embeddings can capture the meaning and the context of words and sentences.

## **Step 1: Import all the necessary libraries**

```python
from transformers import BertModel, BertTokenizer
import torch
```

## **OBSERVATIONS:**

1.  transformers ---------->  

    (a.)  A transformer has a predefined BERT model.

    (b.)  It is a deep learning architecture used in NLP thst uses Self Attention Mechanisms to capture the relationship between the words in the sentences.

2.  BertTokenizer ---------->

    (a.)  It converts the sentences into the word tokens

    (b.)  converts the tokens into the token id

    (c.)  Adding special tokens like [CLS] and [SEP].

3.  BertModel --------------->

    (a.)  BERTModel loads a pretrained BERT neural network.

4.  torch ------------------->

    (a.)  It is a deep learning library used for performing te deep learning operations.

## **Step 2: Load the BERT Tokenizer**

```python
#### Load the BERT Tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
```

```python
tokenizer
```

## **OBSERVATIONS:**

1.  BertTokenizer -------->  

    It converts the sentences into words

    It converts the words into token id.

    It adds the special tokens like [CLS] and [SEP].

2.  Here the object of the Bert Tokenizer calls the function from_pretrained and loads the pre trained BERT Model.

3. The main objective of the above statement is to load a pre-trained BERT Tokenizer named 'bert-base-uncased' that converts the text into such a format that the BERT model can easily understand.

    (a.)  bert --------> converts the text into embeddings that capture the semantic meaning of the words in the sentences.

    (b.)  base -------->  defines the model size. BERT base has 3200 words/subwords

    (c.)  uncased ------> converts everything to lower case.

## **Step 3: Load the BERT Model**

```python
#### Load a pre trained model
model = BertModel.from_pretrained('bert-base-uncased')
```

```python
model
```

## **OBSERVATIONS**

1.  The main objective of the above statement is to load a pre-trained BERT Model named 'bert-base-uncased' that converts the tokenized text into the context based numerical representations called as embeddings.

## **Step 4: Define the Input Sentence:**

```python
#### Define the input text
text = "I love data science"
```

```python
text
```

## **Step 5: Tokenize the Sentence**

```python
#### perform the tokenization of the input text
ans = tokenizer(
    text,
    max_length = 10,
    padding = True,
    truncation = True,
    return_tensors = 'pt'
)
```

```python
ans
```

##**OBSERVATIONS**

1.  The main role of performing the tokenizer is to convert the input text into model based representation form in such a way that the BERT model can understand it easily.

2.  The tokenizer model has been trained with the following parameters :-----

    (a.)  text ---------->  it specifies the input text

    (b.)  max_length = 10 ------>  maximum length of the input text is 10

    (c.)  padding = True ------> makes sure that all the input sequences are of the equal length

    (d.)  truncation = True ------> if the length of the input sequence is greater than the max length then that portion of the length greater than the max length is cut off.

    (e.)  return_sensors = 'pt' ----->  Returns the output in the form of the PyTorch Sensors.

  3.  Here this model returns the output in form of

        {
            input_ids   ,

            attention_mask
        }

       (a.)  input_ids -------> It specifies the numerical representation of the words.

       (b.)   attention_mask ---------> It specifies which tokens are real and which are padded.

              (i.)  if Token is 1 ----------> real

              (ii.) if token is 0 ----------> padded.

## **Step 6: Generate Embeddings**

```python
### Generate the embeddings

with torch.no_grad():
  outputs = model(**ans)
```

```python
outputs
```

```python
outputs.last_hidden_state
```

## **OBSERVATIONS**

1.  It allows the BERT model to run in such a way that the outputs produced by the BERT model does not involve ant computation of the gradient

2.  It explains about the contextual embeddings for every token in the input sentence.

```python
### Output for the model
answers = outputs.last_hidden_state
```

## **OBSERVATIONS**

```python
token_embeddings = outputs.last_hidden_state
```
