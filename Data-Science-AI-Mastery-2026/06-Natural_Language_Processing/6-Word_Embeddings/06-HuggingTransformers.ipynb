### **Aim:---**

1.  The main objective of this problem statement is to perform the sentimental analysis for the Hugging Face.

## **Step 1: Import all the necessary libraries**

```python
from transformers import AutoTokenizer,AutoModelForSequenceClassification
import torch
```

## **OBSERVATIONS:**

1.  transformers ------>  It is a hugging face library used for handling the pre-trained NLP models.

2.  AutoTokenizer ------> It loads the correct tokenizer for the model.

3.  AutoModelForSequenceClassification ----> It loads the correct model for handling all the classification tasks.

4.  torch ----------> It is used to handle all kinds of the tensor operations.

## **Step 2: Choose the pre-trained model**

```python
### Define the model
model = "distilbert-base-uncased-finetuned-sst-2-english"
```

```python
model
```

## **OBSERVATIONS:**

1.  You are selecting a pretrained model from HuggingFace

    (a.)  distilbert → lightweight version of BERT (faster, smaller)

    (b.)  uncased → ignores uppercase/lowercase differences

    (c.)  finetuned-sst-2 → trained on SST-2 dataset (sentiment: positive/negative)

    (d.)  english → works on English text


2.  This model already knows the sentiment classification.

## **Step 3: Load tokenizer**

```python
#### Load the AutoTokenizer
tokenizer = AutoTokenizer.from_pretrained(model)
```

```python
tokenizer
```

## **OBSERVATIONS**

1.  The AutoTokenizer downloads the tokenizer associated with the model.

2.  The AutoTokenizer directly converts the input text into

    (a.)  input_ids -------->  numerical representation of the words

    (b.)  attention_mechanisms -------> which words matter

## **Step 4: Load the Model**

```python
#### Load the Model
models = AutoModelForSequenceClassification.from_pretrained(model)
```

```python
models
```

## **OBSERVATIONS:**

1.  This loads the pre-trained neural network model.

2.  This model is already trained for sentiment classification.

3.  The output layer provides the logits for Positive and Negative reviews.

## **Step 5: Define the Input Text**

```python
### Define the input text
text = "I love learning Data Science and NLP!"
```

```python
text
```

## **Step 6: Perform Tokenization on the input text**

```python
#### perform the Tokenization
inputs = tokenizer(
    text,
    padding=True,
    truncation=True,
    return_tensors="pt"
)
```

```python
inputs
```

##  **OBSERVATIONS:**

1.  The output formed after applying tokenizer on the input text is:--

    {
        'input_ids': [...],
        'attention_mask': [...]
     }


    (a.)  input_ids ------>  numerical representation of the words

    (b.)   attention_masks ------->   which words are important or not in form of padding.


## **Step 7: Perform Forward pass**

```python
#### perform the forward pass
with torch.no_grad():
  outputs = models(**inputs)
```

```python
outputs
```

## **OBSERVATIONS:**

1.  Here it allows the model to run without calculating the gradient descent for the model by disabling the gradient function.

2.  The tokenized inputs are passed to the models and then the logits or the raw predictions for the model.

## **Step 8: Extract logits**

```python
### Extract the logits
logits = outputs.logits
```

```python
logits
```

### **OBSERVATIONS:**

1.  logits are the raw scores before the probability conversion.

## **Step 9: Compute the Softmax probabilities for each labels**

```python
### Compute the probabilities for al classes
probabilities = torch.softmax(logits,dim=1)
```

```python
probabilities
```

## **OBSERVATIONS:**

1.  It computes the probability distributions for each raw score.

## **Step 10: Get predicted class**

```python
### Get the predicted class
predicted_class = torch.argmax(logits,dim=1).item()
```

```python
predicted_class
```

## **OBSERVATIONS:**

1. It shows the  predicted label index

## **Step 11: Map the labels**

```python
labels = ["NEGATIVE", "POSITIVE"]


print("Original Text is:", text)

#### get the predictions for the labels
print("Prediction is:", labels[predicted_class])

#### get the probability of the predicted class
print("Confidence:", probabilities[0][predicted_class].item())
```

## **OBSERVATIONS:**

1.  It extracts the probabilities for the predicted class and converts it into the number.
