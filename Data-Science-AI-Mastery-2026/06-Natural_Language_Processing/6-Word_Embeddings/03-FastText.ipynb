## **FastText**

1.  FastText is an improvised version of Word2Vec model.

2.  It consideres every word as atomic and breaks the words into the sub-words.

3.  It handles all the misspelled words and the morphological variations in the words.

## **Importance of FastText**

1.  It considers every word as atomic and breaks into the subwords.

2.  It handles all the misspelled words in the text.

3.  It handles the morphological variations of the words in the text.

3.  It finds out the vector representation of some unknown words not present in the vocabulary list based on the information of the subwords present in the vocabulary list.

## **Step 3: Prepare Your Text Data**

```python
#### Define the text data
corpus = [
    "Natural language processing is a fascinating field",
    "Machine learning enables computers to learn from data",
    "Deep learning is a subset of machine learning",
    "Word embeddings represent words as vectors",
    "FastText handles rare and unseen words effectively"
]
```

```python
corpus
```

## **OBSERVATIONS:**

1.  Here the corpus text data has been made.

## **Step 4: Perform the Tokenization on the corpus text**

```python
#### perform the word tokenization on the corpus text

tokenized_sentences = [word_tokenize(x.lower()) for x in corpus]
```

## **OBSERVATIONS:**

1.  Here Word tokenization has been applied on the corpus text in the list.

2.  The output formed is the list of all the tokenized sentences.

```python
### Another method for doing the word tokenization on the corpus text

corpus
```

```python
#### perform the word tokenization on the corpus text
words = word_tokenize(corpus)
```

```python
words
```

```python
#### Convert all the tokenized words in text form

words = ' '.join(words)
```

```python
words
```

## **OBSERVATIONS:**

1.  The corpus text has been tokenized into the words.

2.  The words in the list has been converted back into the text form.

## **Step 5: Perform the normalization on the text**

```python
from nltk.tokenize import RegexpTokenizer

#### Create an object for Regular Expression Tokenizer
reg = RegexpTokenizer(r'\w+')

#### Apply the object of Regular Expression and remove all the punctuations and special symbols from the text
words = reg.tokenize(words)
```

```python
words
```

## **OBSERVATIONS:**

1.  The object of Regular Expression Tokenizer has been created using \w+  where it extracts only more than one occurances of numbers and digits.

2.  Then the object of Regular expression has been applied on the text so as to remove all the punctuations and the special symbols from the text.

## **Step 6: Remove the stopwords from the text**

```python
#### define all the english stop words
from nltk.corpus import stopwords
english_stopwords = set(stopwords.words("english"))
```

```python
english_stopwords
```

```python
#### Remove all the english stop words from the text
ans = [x.lower() for x in words if(x not in english_stopwords)]
```

```python
ans
```

## **OBSERVATIONS:**

1. Here all the english stop words have been removed from the text.

## **Step 7: Train the FastText Model**

```python
### Train the Fast Text model
model = FastText(
    sentences         = tokenized_sentences         ,
    vector_size       =     50                      ,
    window            =      3                      ,
    min_count         =      1                      ,
    sg                =      1                      ,
    epochs            =      100
)
```

```python
model
```

##**OBSERVATIONS:**

1.  Here the FastText model has been trained with the following parameters:-------

    (a.)  sentences = tokenized_sentences . It accepts the input in form of tokenized words.

    (b.)  vector_size = 50 ----------->  Here the vector is a numerical dense dimensional vector of size 50.

    (c.)   window = 3 ---------------->  Here the size of the context input is 3

    (d.)  min_count = 1 -------------->  atleast 1 time, the word must appear in the dictionary.

    (e.)  sg = 1 --------------------->  The model architecture is Skip-Grams. Here based on the target word, surrounding context words are predicted.

    (f.)  epochs = 100 ---------------> Total 100 iterations are required to train the model.
