## **GloVe Embeddings:**

1.  GloVe stands for Global Vector For Word Representation.

2.  It helps in finding out the maximum frequently occuring words in the given context.

3.  The model for GloVe embeddings used is "glove-wiki-gigaward-100"

4.  Each and every word is converted into numerical dense dimensional vector of size 100 in glove-wiki-gigaward-100.

## **glove-wiki-gigaward-100**

1.  glove-wiki-gigaward-100 --------> It is a pre-trained model used for GloVe embeddings where each and every word is converted into dense dimensional vector of the size 100.

  (a.)  It helps in finding out the maximum frequently occuring words in the particular context.

## **Importance of GloVe**

1.  It helps in finding out the most frequently occuring words from the given text in the dense representation form.


| Word  | Vector Size |

### Dense Representation of the frequently occuring words in the same context

“cat” and “mat” co-occur in the same context (sentence).

“cat” and “dog” may co-occur often across many sentences.

“cat” and “table” may rarely co-occur.

So, word-word co-occurrence statistics capture how often each word appears with every other word.

We can store this information in a co-occurrence matrix X:

X[i][j] = number of times word j appears in the context of word i.

## **Step 8: Convert the sentece into GloVe Embeddings:**

```python
### perform the function definition
def sentence_vector(text):
  ### convert the text to lower case
  text = text.lower()
  ### perform the word tokenization on the text
  words = word_tokenize(text)
  #### create an empty list of word vectors
  word_vectors = []

  #### Check if the word exists in the vocabulary ot not then append in the word vector list
  for x in words:
    if(x in model):
      word_vectors.append(model[x])

  #### Check if the length of the word vectors is 0 then return null
  if(len(word_vectors)==0):
    return(np.zeros(100))

  #### perform the embeddings of the word vectors and get the average values aong the column wize manner
  return(np.mean(word_vectors,axis=0))

### define the input sentence
text = "I love machine learning"

### perform the function call
ans = sentence_vector(text)
print(ans)
```

### **OBSERVATIONS:**

1. We have taken the sentence as the input

2. Then perform the tokenization on the  input sentence

3. Checked if each word exists in the GloVe dictionary

4. If exists, then embedding is performed on it and we get the vector representation of it by taking the average of all the word vectors in column wise manner.

5. So the above is the vector representation of the sentence using GloVe embedding.
