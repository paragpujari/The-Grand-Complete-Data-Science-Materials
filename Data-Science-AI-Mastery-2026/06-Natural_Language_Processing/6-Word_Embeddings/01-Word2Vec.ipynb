## **Word2Vec**

1.  Word2Vec is a word embedding technique that is used to convert each and every input word into the dense dimensional numerical vector so that it can easily be trained by the machine learning model while preserving its semantic meaning.

2.  Words with the similar meaning will have the same numerical vector valued representations.

## **Problem Stetement:**

1.  The main aim of the problem statement is to find out the vector representation of each and every word and find out the similar words w.r.to the particular word and even compute the similarity ratio between the two words.

## **Step 3: Load the Dataset**

```python
### Load the dataset

df = pd.read_csv('/content/SMSSpamCollection',sep='\t',names=['label','message'])
```

```python
#### Interchange the positions of the message and the labels

df=df[['message','label']]
```

## **OBSERVATIONS:**

    (a.)  message -------->  It specifies the type of the email message.

## **Step 4: Perform Text Preprocessing**

```python
### perform word tokenization on all the sentences in the message column

res = [word_tokenize(x.lower()) for x in df['message']]
```

```python
res
```

## **OBSERVATIONS**

1.  The above list is a list of all the tokenized sentences obtained from the message column of the dataframe.

2.  This can be fed as an input into the Word2Vec model.

## **Step 5: Train the Word2Vec model**

```python
### Train the Word2Vec model

model = Word2Vec(
    sentences   =    res            ,
    vector_size =     50            ,
    window      =     5             ,
    min_count   =     1             ,
    sg          =     0             ,
    workers     =     1
)
```

## **OBSERVATIONS:**

1.  The model of Word2Vec has been created with the help of the following parameters:-----

    (a.)  sentences = list of the tokenized sentences

    (b.)  vector_size = size of the dense dimensional embedded vector

    (c.)  window = 5 ---->  5 context input neighbors need to be considered for determining the target

    (d.)  min_count = 1 ------>  Minimum number of frequency needed for the dictionary

    (e.)sg = 0 ------------> determines the type of the model architecture.

            (i.)  If sg = 0 ----------->  Model architecture is CBOW.


2.The CBOW model is trained.

```python
model
```

## **Step 9: Get the most similar words w.r.to a particular word**

```python
#### Get the most similar word w.r.to 'easy'
model.wv.most_similar('easy')
```

## **OBSERVATIONS:**

```python
#### Get the most similar word w.r.to 'free'
model.wv.most_similar('free')
```

## **OBSERVATIONS:**

1.  The above are the list of all the words that are similar to easy as their vector values are verymuch close to the vector representation value of 'free'.

## **Step 11: Get the Similarity between two words:**

```python
#### Get the similarity between two words
model.wv.similarity('free','win')
```

## **OBSERVATIONS:**

1.  Here the similarity score between the two words ('free','win') is 0.97 which states that these two words are more identical with respect to each other.
