#### Stemming

1.  Stemming is a technique used in text preprocessing that reduces the word to the root form which may or may not have a meaning.

#### Importance of Stemming:----

1.  Stemming reduces the word to the root form which may or may not have a meaning.

2.  It speeds up the NLP related tasks.

3.  It improves the accuracy and the performance of the model.

4.  It improves the searching tasks.


#### Types of Stemming :----

1.  PorterStemmer --------->

    (a.)    PorterStemmer is a widely used stemming algorithm that reduces the word into the root form by removing the suffixes from it.

#### Problem Stetement :-----

1.  The main objective of the problem statement is to reduce each and every processed word into the root form by using Stemming.

#### Step 1: Import all the necessary libraries

#### Step 3: Create a Sample Text

```python
#### define the sample text

text = """
Natural Language Processing allows computers to understand human language.
People are communicating using texts, emails, and social media every day.
We are building systems that can automatically analyze and respond intelligently.
"""
```

```python
text
```

#### Step 4: Perform the tokenization of the text

```python
#### perform sentence tokenization on the paragraph

from    nltk.tokenize   import  word_tokenize, sent_tokenize

#### convert the paragraphs into the sentences

sentences = sent_tokenize(text)
```

```python
sentences
```

#### OBSERVATIONS:

1.  After performing the sentence tokenization on the paragraph text, the paragraph text has been converted into the sentences.

```python
#### convert the text into words

from    nltk.tokenize import word_tokenize

words = word_tokenize(text)
```

```python
words
```

#### OBSERVATIONS:

1.  Here after applying word tokenization on the paragraph text, the paragraph text has been converted into the words.

#### Step 5: Load all the english stop words

```python
from nltk.corpus import stopwords

#### Load all the english stopwords
english_stopwords = set(stopwords.words("english"))
```

```python
english_stopwords
```

#### OBSERVATIONS:

1.  All the english stop words have been listed.

#### Step 6: Remove all the english stopwords from the text

```python
#### Remove all the english stop words from the text

res = [x for x in words if(x not in english_stopwords)]
```

```python
res
```

```python
#### convert the list into text
ans = ' '.join(res)
```

```python
ans
```

#### OBSERVATIONS:

1.  Here all the english stop words have been removed from the text.

#### Step 7: Remove all the punctuations and special symbols from the text

```python
from nltk.tokenize import RegexpTokenizer

#### create an object for RegexpTokenizer
reg = RegexpTokenizer(r'\w+')

#### Apply the object of RegexpTokenizer on the text to remove all the punctuations from the text
ansreg = reg.tokenize(ans)
```

```python
ansreg
```

#### OBSERVATIONS:

1.  After applying the regular expressions on the above text, all the punctuations and the special symbols have been removed rom the text.

#### Step 8: Apply Porter Stemmer on the above text

```python
from nltk.stem import PorterStemmer

#### create an object for Porter Stemmer

ps = PorterStemmer()

#### perform the stemming on the filtered words in the text
res = [ps.stem(x) for x in ansreg if(x != ' ')]
```

```python
res
```

#### OBSERVATIONS:

1.  Stemming has been applied on each and every word in the text.

2.  Some of the root words formed after stemming have a meaning and some do not have a meaning.
