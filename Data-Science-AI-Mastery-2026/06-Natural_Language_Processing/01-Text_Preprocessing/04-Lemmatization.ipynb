#### Step 3: Define the Sample Text

```python
# Define sample paragraph
text = """
Natural Language Processing allows computers to understand human language.
People are communicating using texts, emails, and social media every day.
We are building systems that can automatically analyze and respond intelligently.
"""
```

```python
text
```

#### OBSERVATIONS:

1.  A sample text paragraph has been defined .

#### Step 4: Perform word tokenization on the Sample Text

```python
from    nltk.tokenize import word_tokenize

#### perform word tokenization on the sample text

words = word_tokenize(text)
```

```python
words
```

#### OBSERVATIONS:

1.  After performing word tokenization on the above paragraph text, the text has been broken into the words.

#### Step 5: Define the english stop words and remove it

```python
#### define all the english stop words

from nltk.corpus import stopwords

english_stopwords = set(stopwords.words("english"))

print(english_stopwords)

#### filter out all the english stop words from the text
res = [x for x in words if(x not in english_stopwords)]

print(res)

#### convert the result in list into the text form

res = ' '.join(res)
```

#### Step 6: Define the Regular Expressions to remove all the punctautions from the text

```python
from nltk.tokenize import RegexpTokenizer

#### create an object for RegexpTokenizer

reg = RegexpTokenizer(r'\w+')

#### Apply the Regular Expression Tokenizer to remove all the punctuations and special symbols from the text

ans = reg.tokenize(res)
```

```python
ans
```

#### OBSERVATIONS:

1.  All the punctuations and the special symbols have been removed from the filtered text after performing Regular Expression Tokenizer.

#### Step 7: Peforming Lemmatization on the filtered words

```python
from nltk.stem import WordNetLemmatizer

#### Create an object for Word Net Lemmatize
lemma = WordNetLemmatizer()

#### Apply the Lemmatization on the filtered words

ansdata = [lemma.lemmatize(x.lower()) for x in ans]

print(ansdata)

#### Convert the lemmatized word in list into the text form

ansdata = ' '.join(ansdata)

print(ansdata)
```

#### OBSERVATIONS:

1.  The object of Word Net Lemmatizer has been created.

2.  Then each word has been looped in lists.

3.  The lemmatizer object is applied to each word in lower case.

4.  It converts each word to the root word having  valid meaning.
