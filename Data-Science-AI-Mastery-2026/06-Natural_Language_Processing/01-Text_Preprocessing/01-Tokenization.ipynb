#### Tokenization

1.  Tokenization refers to the process of dividing the text corpus into smaller units or tokens

#### Types of Tokenization :----

1.  Sentence Tokenization ============> It refers to the process of breaking the paragraphs text into the sentences.

Ex:--    Input :- "I love Data Science. It is amazing!"   ========> Paragraphs

         Output :- ["I love Data Science.","It is amazing!"]    ===============> Sentences in the form of lists

2.  Word Tokenization =================> It refers the process of breaking the sentences into the words.

Ex :---   Input :-- "I love Data Science." ===============> Sentence

          Output :-- ["I","love","Data","Science."]


#### Steps used in this Algorithm :------

1.  Import all the necessary libraries.

2.  Include all the necessary resources required for nltk.

3.  Define the corpus text.

4.  Perform the sentence tokenization on the corpus text to obtain the sentences.

5.  Perform the word tokenization on the sentences of the corpus text to get the word tokens.

6.  Perform Regular Exp Tokenization to remove all the unnecessary punctuations from the text.

#### Step 3: Define the corpus text

```python
text = """Tokenization is the first step in text processing. 
It breaks the text into smaller parts called tokens — words, phrases, or symbols.
This helps machines understand human language more effectively!
"""
```

```python
text
```

#### OBSERVATIONS:

1.  A corpus text can be considered as a text document having a collection of two or more sentences.

#### Step 4: Perform sentence tokenization on the above corpus text to obtain the sentences

```python
from    nltk.tokenize   import  sent_tokenize
```

```python
#### print the original text

print(text)
```

```python
#### Apply the sentence tokenization on the corpus text

sentences = sent_tokenize(text)
```

```python
sentences
```

#### OBSERVATIONS:

1.  The corpus text was initially a paragraph.

2.  It is converted into the sentences in list after applying sentence tokenization on it.

#### Step 5: Perform word tokenization on the sentences to get the words from the sentences

```python
from nltk.tokenize import word_tokenize
```

#### Method 1: Sentence ----------> Word Tokenization

```python
#### perform word tokenization on each and every sentences in the list
res =[word_tokenize(x) for x in sentences]
```

```python
res
```

#### OBSERVATIONS:

1.  We have a corpus text in paragraph.

2.  On applying the sentence tokenization on it, the paragraph is converted into the sentences in list.

3.  So we already have a list that stores all the sentences.

4.  Then word tokenization is performed on each and every sentences in the list, then the tokenized words are obtained.

5.  So the first list stores all the sentences and the second list stores all the words in the sentences.

6.  So this is the reason that list inside a list is ormed on performing the word tokenization on the sentences.

#### Method 2: Direct Word Tokenization on the Paragraph

```python
#### Apply word tokenization on the sentences

##### Print the original sentence
print(sentences)
```

```python
##### Convert the sentences into the paragraph
text = ' '.join(sentences)
print(text)
```

```python
#### perform word tokenization on the text
words = word_tokenize(text)
print(words)
```

#### OBSERVATIONS:

1.  We have the sentences in form of the list.

2.  Now the sentences in list is joined back into paragraph.

3.  Then the word tokenization on the entire paragraph to get the tokenized words from the paragraph.

#### Step 6: Perform Regular Expression Tokenization on the above text

```python
from nltk.tokenize import RegexpTokenizer

#### create an object for Regular expression Tokenization
reg = RegexpTokenizer(r'\w+')

#### Original text
print(text)

#### Performing the regularization on the above text
res = reg.tokenize(text)
print(res)
```

#### OBSERVATIONS:

1.  An object of Regular Expression Tokenization is created with the following parameters :----

    (a.)    \w ----------->    extracts the numbers and letters

    (b.)    +  ----------->     comprises of one or more occurances

2.  Here Regular expression tokenization is applied on the above text to remove all the punctuations and the special symbols from the text.
