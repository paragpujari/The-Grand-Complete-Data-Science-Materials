#### Stop Words :----

1.  Stop words are the words that do not add any meaning or value to the particular sentence or the text.

#### Why do we remove the stop words ?

    Stop words are removed so as 

    (i) to reduce the noise.

    (ii) to improve the accuracy of the model.

    (iii) to improve the performance of the model.

    (iv) to do better feature selection for the model.

#### Variants of Stop Words :-----

1.  english stop words =======> The words of the english dictionary that do not add any meaning to the text is known as english stop words.

#### Step 4: Perform the tokenization on the sample text

```python
#### define the original text

print('---------------------------Original Text---------------------------------------------')

print(text)

print('---------------------------Performing Tokenization on the Original Text---------------')

words = word_tokenize(text)

print(words)

print('--------------------------Performing Regular Expression Tokenization on the Original Text------')

#### Create an object for Reg exp tokenizer
reg = RegexpTokenizer(r'\w+')
#### Apply the object of regular expression tokenizer on the input text
ans = reg.tokenize(text)
print(ans)
```

    (a.)    An object for Regular Expression Tokenizer is created with

            \w ----------->  extracts numbers , letters and underscores

            +  ----------->  one or more occurances.

    (b.)    On applying the regular expression tokenizer on the text, it extracts all the letters and numbers from the text and removes all the punctuations, special symbols from the text.

#### Step 7: Write a reusuable function to remove all the stop words from the text

```python
#### define the function
def remove_stopwords(sample):
    #### perform the word tokenization on the sample text
    words = word_tokenize(sample)
    #### Join the tokenized words in list to text form
    texts = ' '.join(words)
    #### Apply Regular Expression Tokenizer on the text to remove all the punctuations and special symbols from the text
    reg = RegexpTokenizer(r'\w+')
    ans = reg.tokenize(texts)
    #### define all the english stop words
    english_stopwords = set(stopwords.words("english"))
    #### remove all the english stop words from the text
    res=[x for x in ans if(x.lower() not in english_stopwords)]
    #### join all the filtered words in the text form
    ans = ' '.join(res)
    return(ans)
```

```python
#### define the sample text
sample = 'This is a simple NLP example showing how to remove stopwords from text.'
#### call the function
ans = remove_stopwords(sample)
print(ans)
```
