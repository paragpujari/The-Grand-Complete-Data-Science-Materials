#### Regular Expressions

1.  Regular Expressions provides us the functions to search, replace, split and extract the patterns from the data.

#### Problem Stetement :-----

1.  The main objective of the problem statement is to preprocess the input text using the Regular Expressions.

#### Step 1: Import all the necessary libraries

```python
import re
```

#### Step 2: Define the Sample Text

```python
#### define the sample text

text = """
Hello everyone!

My email is parag123@gmail.com.
My phone number is 9876543210.

Visit https://www.google.com for more details.

The product costs ₹5000 and $100.

Data Science, Machine Learning, NLP!!!
"""
```

```python
text
```

#### OBSERVATIONS:

1.  Here the sample text is defined.

2.  This text contains:

    (a.)    Uppercase and lowercase letters

    (b.)    Email address

    (c.)    Phone number

    (d.)    URL

    (e.)    Special symbols

    (f.)    Numbers

We will clean and extract information using Regex.

#### Step 3: Find All the numbers

```python
#### Extract all the digits from the text

numbers = re.findall(r'\d+',text)
```

```python
print(numbers)
```

#### OBSERVATIONS:

1.  \d --------------->  It extracts all the digits from the text.

2.  +  --------------->  one or more occurances.

3.  It extracts the list of all the digits from the text.

#### Step 4: Find All the email

```python
#### Extract all the emails from the text

emails = re.findall(r'[\w\.-]+@[\w\.-]+',text)
```

#### OBSERVATIONS:

1.  [\w\.-]+ ----------------------->  It represents the emails that extracts the email from the text.

```python
emails
```

#### Step 5: Find All the phone numbers

```python
#### Extract the phone numbers from the text

phones = re.findall(r'\d{10}',text)
```

```python
phones
```

#### OBSERVATIONS:

1.  \d{10} ------------> It extracts the digits nearly 10 times.

#### Step 6: Extracts the URL

```python
#### Extracts the URL from the text

url = re.findall(r'https?://\S+',text)
```

```python
url
```

#### Step 7: Remove the Punctuations

```python
#### Remove all the punctuations from the text

clean_text = re.sub(r'[^\w\s]','',text)
```

```python
clean_text
```

#### Step 8: Remove the digits

```python
#### Remove the digits from the text

res = re.sub(r'\d+','',text)
```

```python
res
```

#### OBSERVATIONS:

1.  sub ------------>  It stands for substitute

    sub(pattern, replacement, text) = sub(r'\d+','',text)

    It finds out the occurances of all the digits and replaces it with '' in the text.

#### Step 9: Remove the URL

```python
#### Remove the URL from the text

res = re.sub(r'https?://\S+','',text)
```

```python
res
```

#### OBSERVATIONS:

1.  It substitutes all the URL with the blank spaces in the text.

#### Step 9: Remove all the email address from the text

```python
#### Remove all the email address from the text

res = re.sub(r'[\w\.-]+@[\w\.-]+','',text)
```

```python
res
```

#### OBSERVATIONS:

1.  It substitutes all the emails with the '' in the text.

#### Step 10: Remove the Extra Spaces from the text

```python
#### Remove the extra spaces from the text

res = re.sub(r'\s+',' ',text)
```

```python
res
```

#### OBSERVATIONS:

1.  It substitutes all the spaces with ' ' in the text.

#### Step 11: Extract all the words from the text

```python
#### Extract all the words from the text

words = re.findall(r'[A-Za-z]+',text)
```

```python
words
```

#### OBSERVATIONS:

1.  It extracts one or more than one occurances of [A-Z] or [a-z] from the input text.

#### Step 12: Extract all the capitalized words

```python
#### Extract all the capitalized words from the text

words = re.findall(r'\b[A-Z][a-z]*\b',text)
```

```python
words
```

#### OBSERVATIONS:

1.  Using \b[A-Z][a-z]*\b we have extracted all the capitalized words from the text.

#### Step 13: Split the text using Regex

```python
#### Splits the text

splits = re.split(r'\s+',text)
```

```python
splits
```

#### OBSERVATIONS:

1.  Here the entire text is splitted based on one or more spaces.
