#### Text Cleaning

1.  Text Cleaning refers to the process of cleaning the text from any punctuations, special symbols, stop words etc so as to improve the efficiency of the code.

#### Problem Statement :---

1.  The main aim of the problem statement is to clean the text so as to get the useful insights from the text.


#### Step 1: Import all the necessary libraries

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

import  nltk
from    nltk.tokenize       import word_tokenize, sent_tokenize,RegexpTokenizer
from    nltk.corpus         import stopwords
from    nltk.stem           import WordNetLemmatizer
```

#### Step 2: Download the additional resources required for NLTK

```python
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

#### OBSERVATIONS:

1.  stopwords ----------------->   It downloads the english stop words.

2.  wordnet   ----------------->   It downloads the wordnet needed for lemmatization.

3.  omw-1.4   ----------------->   It downloads the additional word net language mappings.

#### Step 3: Define the Sample Text

```python
#### Define the sample text

text = """
Data Science is one of the fastest-growing fields in the world.
The Data Scientists are working on Machine Learning, Deep Learning,
and Artificial Intelligence projects.
"""

print(text)
```

#### Step 4: Convert the text into lower case

---------------------------------------------Original Text------------------------------------------------------------------------

```python
text
```

```python
#### Convert the text into lower case

text = text.lower()
```

-------------------------------------------Lower Case Text-----------------------------------------------------------------------------

```python
text
```

#### OBSERVATIONS:

1.  Here the entire text is converted into the lower case.

#### Step 5: Perform Word Tokenization on the above text

```python
#### perform word tokenization on the above text

words = word_tokenize(text)
```

```python
words
```

#### OBSERVATIONS:

1.  The entire text has been divided into words.

#### Step 6: Define all the english stop words and remove them from the text

```python
#### define all the english stop words

english_stopwords = set(stopwords.words("english"))
```

```python
english_stopwords
```

#### OBSERVATIONS:

1.  All the english stop words have been defined.

```python
#### Remove all the english stop words from the text

res = [x for x in words if(x not in english_stopwords)]
```

```python
res
```

#### OBSERVATIONS:

1.  All the english stopwords have been removed from the text.

```python
#### convert the filtered word in list into the text form

res = ' '.join(res)
```

#### Step 7: Remove all the punctuations from the text

```python
from nltk.tokenize import RegexpTokenizer

#### create an object for RegexpTokenizer

reg = RegexpTokenizer(r'\w+')

#### using the object, remove all the punctuations and special symbols from the text

ans = reg.tokenize(res)
```

```python
ans
```

#### OBSERVATIONS:

1.  All the punctuations have been removed from the text.

#### Step 8: Perform Lemmatization on each and every filtered words in the list

```python
from nltk.stem import WordNetLemmatizer

#### create an object for Word Net Lemmatizer
lemma = WordNetLemmatizer()

#### using the object, convert each word to the root form
res = [lemma.lemmatize(x) for x in ans if(x != ' ')]
```

```python
res
```

#### OBSERVATIONS:

1.  Here all the words in the list have been converted into their root form with their proper meaning.

#### Step 9: Convert all the filterd words in list into the text form

```python
#### Convert all the filtered words in list into the text form
res = ' '.join(res)
```

```python
res
```

#### OBSERVATIONS:

1.  All the filtered words in the list have been converted into the text form.
