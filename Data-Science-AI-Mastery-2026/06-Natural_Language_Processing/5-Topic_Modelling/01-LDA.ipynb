#### Latent Dirichlet Allocation(LDA)

1.  Latent Dirichlet Allocation is an unsupervised topic modeling algorithm that helps in extracting and finding out all the hidden topics from the collection of documents.

Ex:--  Suppose LDA considers document 1 as the input

        In that document 1, there are three different topics, So LDA finds out and assigns the probability scores for each topic

        Even within the particular topic, there may be several sub-topics, SO LDA may assign several sub-topics with the probability scores.

        This goes on and on unless it reaches the end of the document page.


Text :-- "Virat Kohli scored a century in the cricket match"

    Based on the above text document, LDA assigns

            (a.)    Sports -------------->   70 %

            (b.)    Politics ------------>   20 %

            (c.)    Technology ---------->   10 %


#### Mathematical Intuition :----

1.  LDA assumes each text document as the input

    For each document, LDA 

        (a.)    selects a topic distribution

        (b.)    selects a topic

        (c.)    selects the word from the topic

        (d.)    selects the sub words from the word.

2.  This process goes on unless and until it reaches the end of the document page. 

3.  So in this way, LDA extracts all the hidden topics from the collection of documents.

#### Steps used in this Algorithm:----

1.  Import all the necessary libraries

2.  Define the Sample Dataset

3.  Perform Text Vectorization (Bag of Words) i.e Tfidf Vectorizer

4.  Apply LDA Model

5.  Display the Topics

6.  Predict Topic for Each Document


#### Step 1: Import all the necessary libraries

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

import  nltk

from    sklearn.feature_extraction.text import TfidfVectorizer
from    sklearn.decomposition           import LatentDirichletAllocation
```

#### OBSERVATIONS:

1.  TfidfVectorizer           ------->  Converts the input text into the matrix of tfidf numerical scores.

2.  LatentDirichletAllocation ------->  discovers the hidden topics from the collection of the text documents.

#### Step 2: Define the Sample Dataset

```python
#### Define the sample dataset
documents = [
    "I love to eat pizza and pasta",
    "The new movie was fantastic and thrilling",
    "Python is great for data science and machine learning",
    "I enjoy watching movies and eating popcorn",
    "Data scientists use Python for analysis and modeling",
    "The pizza at that restaurant was delicious",
]
```

```python
documents
```

#### OBSERVATIONS:

1.  Here the sample text dataset is defined.

#### Step 3: Perform Text Vectorization (Bag of Words) i.e Tfidf Vectorizer

```python
#### perform the Tfidf vectorizer on the sample text

from sklearn.feature_extraction.text import TfidfVectorizer

#### Create the object of TfidfVectorizer
tfidf = TfidfVectorizer()

#### Apply the object of TfidfVectorizer on to the text, to convert it into the numerical form
X_tfidf = tfidf.fit_transform(documents)
```

```python
X_tfidf
```

#### OBSERVATIONS:

1.  Here the input document text is converted into the sparse matrix of tfidf so as to only capture the non zero values and their positions and ignore all the zero values.

2.  This is done so as to prevent the memory wastage.

```python
#### Then convert the sparse matrix into the dense matrix
X = X_tfidf.toarray()
```

```python
X
```

#### OBSERVATIONS:

1.  Here the sparse matrix is converted into the numpy array for better view and visibility.

2.  The new input has become a complete numerical model that can easily be trained by the LDA model.

#### Step 4: Apply LDA Model

```python
from sklearn.decomposition import LatentDirichletAllocation

#### create an object for LatentDirichletAllocation
lda = LatentDirichletAllocation(n_components=2,random_state=42)
```

```python
#### Train the lda model
lda.fit(X)
```

#### OBSERVATIONS:

1.  The object of lda model has been made using the following parameters:---

    (a.)    n_components = 2 ------------> LDA discovers 2 hidden topics from the collection of the text documents.

    (b.)    random_state = 42 -----------> It ensures the reproducibility

2.  Then using the object of lda, the input numerical vector data has been trained.

#### Step 5: Display the Topics

```python
#### get the vocabularies in the document
vocabularies = tfidf.get_feature_names_out()
```

```python
vocabularies
```

#### OBSERVATIONS:

1.  It displays the list of all the vocabularies present in the text document.

```python
#### Discover the hidden topics
hidden_topics=lda.components_
```

```python
hidden_topics
```

#### OBSERVATIONS:

1.  The two hidden topics are discovered from the document using LDA algorithm.

```python
#### get the shape of the lda hidden topics
print("Shape of the LDA hidden topics is:", hidden_topics.shape)
```

#### OBSERVATIONS:

1.  There are a total of 2 hidden topics and each topic has 33 vocabularies in it.

```python
#### Get the top five words or each topic
for x,y in enumerate(hidden_topics):
    #### print the topic number
    print(f"Topic:{x+1}")
    #### print the top five words for each topic
    print([vocabularies[i] for i in y.argsort()[-5:]])
```

#### OBSERVATIONS:

1.  It returns the top five words for each topic.

#### Step 6: Predict Topic for Each Document

```python
#### List the documents
documents
```

#### OBSERVATIONS:

1.  The list of documents is obtained.

```python
#### get the predicted probabilities for two hidden topics in each of the documents
predicted_probabilities_topics = lda.transform(X)
```

```python
predicted_probabilities_topics
```

#### OBSERVATIONS:

1.  The predicted probabilities have been obtained for the two hidden topics in each of the documents.

```python
#### print the details of the predicted probability distributions
for i,doc in enumerate(documents):
    #### print each of the document
    print(f"\n Document {i+1} : {doc}")
    #### print the probability distribution for each topic in the document
    print(f"Predicted Probability for each topic is :{predicted_probabilities_topics[i]}")
    #### get the dominant topic
    print(f"Dominant Topic is:{predicted_probabilities_topics[i].argmax()+1}")
```

#### OBSERVATIONS:

1.  The above code explains about the details of the probability distributions for each document.

2.  It even computes the probability distributions for each topic in the document.

3.  It finds out the dominant topic from each of the document (the index position of the topic with the highest value)
