#### **Step 3: Perform the Pad Sequences on the inputs**

```python
#### Perform the padding on the input sequences
from tensorflow.keras.preprocessing.sequence import pad_sequences

#### define the maximum length for padding
max_length = 200
```

```python
X_train
```

```python
X_test
```

```python
print("Shape of the input training data before padding is:", X_train.shape)
print("Shape of the input testing data  before padding is:", X_test.shape)
```

```python
#### perform the padding operation on the input data to make it of the equal length
X_train = pad_sequences(X_train, maxlen=max_length)

X_test  = pad_sequences(X_test,  maxlen=max_length)
```

#### **OBSERVATIONS:**

1.  Initilally all the input sequences are of the variable sequences.

2.  But after padding all the input sequences are converted into the equal sequences.

3.  Each and every word id is converted into the dense dimensional vectors of the same length of 200.


4.  Padding is done so that

    (a.)  Model Training will become faster.

    (b.)  Batch Processing will not fail.


5.  If padding is not done then the Memory Layout Becomes Difficult


6.  Here padding has been applied on the input and all the inputs are of the equal sequences.


### **Types of Padding**

1.  Pre-padding -------->  Applying the zeros at the beginning of the word vector.

2.  Post-padding ------->  Applying the zeros after the ending of the word vectors.

#### **Step 4: Architecture of Simple RNN:**

Input Review

      ↓

Embedding

      ↓

SimpleRNN

      ↓

Dense

      ↓
      
Prediction

```python
#### Built the RNN Model
model = Sequential([
    #### define the embedding layer
    Embedding(input_dim = 10000, output_dim=32, input_length = max_length)                ,
    #### Apply the SimpleRNN Model on the top of the embedding layer
    SimpleRNN(units=32, activation='tanh')                                                ,
    #### Construct the output layer
    Dense(units=1, activation='sigmoid')
])
```

```python
model
```

#### **OBSERVATIONS:**

1.  The RNN Sequential model is built with the help of the following layers :--  

    (a.)  Embedding layer:

        (i) Here the 10000 most frequent words ra econverted into 32 dense dimensional vectors with 200 equal lenth of each.

        (ii) This dense dimensional vector sequence are now processed by SimpleRNN word by word.

    (b.) SimpleRNN layer:

        (i) Here the SimpleRNN has 32 hidden neurons in the layer. Each hidden neurons represents the hidden state.

        (ii) Each hidden state processes the input sequence by using the information of the past state and then predicting for the future state.

        (iii) In this manner the last hidden state summarizes all the sentence review.

        (iv) It also uses the 'tanh' activation function to keep the values bounded and helps in stable learning representations.

    (c.) Output layer:

        (i) Here it has one neuron in the output layer where sigmoid activation function is applied on it to determine whether the movie review is positive or negative.


```python
#### get the summary of the model
model.summary()
```

#### **Step 5: Compile the RNN Model**

```python
#### Compile the RNN Model
model.compile(
    optimizer = 'adam'                                  ,
    loss      = 'binary_crossentropy'                   ,
    metrics   = ['accuracy']
)
```

#### **OBSERVATIONS:**

1.  The RNN Model has been made ready for the training with the following parameters:--

    (a.)  optimizer = 'adam' . It is used to update the weights during the backpropogation.

    (b.)  loss = 'binary_crossentropy'. It is used to solve the binary classification problem.

    (c.)  metrics = ['accuracy']. It is used to learn the performance of the model.

#### **Step 6: Train the RNN Model**

```python
#### Train the RNN Model
history = model.fit(
    X_train                            ,
    Y_train                            ,
    epochs           =       5         ,
    batch_size       =       64        ,
    validation_split =       0.2       ,
    verbose          =        1
)
```

#### **Step 7: Evaluate the RNN Model**

```python
loss, accuracy = model.evaluate(X_test, Y_test, verbose=0)

print(f"Loss: {loss:.4f}")
print(f"Accuracy:{(accuracy)*100.0:.4f}")
```

#### **Step 8: Make Predictions on Custom Text**

```python
### Get the index of each and every word

word_index = imdb.get_word_index()

### define the function to encode the words into integers
def encoded_review(text):
  ### get the tokens from the text
  tokens = text.lower().split()

  encoded_text = [
        word_index.get(word, 2) + 3
        if word in word_index
        else 2
        for word in tokens
    ]

  encoded_text = pad_sequences(
        [encoded_text],
        maxlen=max_length
    )

  return encoded_text


### define the sample text
sample_text = "This movie was amazing and the actors were brilliant"


res = encoded_review(sample_text)
print(res)

### Make the predictions from the encoded data
predictions = model.predict(res)[0][0]
print(predictions)


if(predictions > 0.5):
  print("Positive Review")
else:
  print("Negative Review")
```
