#### **BackPropogation**

1.  Backpropogation refers to the technique used in the neural networks that is ued to compute the gradients of the loss function with respect to the weights and bias using the chain rule and then updating them using gradient descent to minimize the error.


#### **Simple Neural Network**

Input Layer (30 features)

        ↓

Hidden Layer (5 neurons, ReLU)

        ↓

Output Layer (1 neuron, Sigmoid)



#### **Step 1: Import all the necessary libraries**

#### **Step 3: Initialize the Parameters**

```python
np.random.seed(42)

input_dim = X_train.shape[1]    ##### specifies the number of neurons in the input layer
hidden_dim = 5                  #### specifies  the number of neurons in the hidden layer
output_dim = 1                  #### specifies the number of neurons in the output layer

#### update the weights and bias
W1 =  np.random.randn(input_dim,hidden_dim)*0.01
b1 =  np.zeros((1,hidden_dim))

W2 =  np.random.randn(hidden_dim,output_dim)*0.01
b2 =  np.zeros((1,output_dim))
```

#### **OBSERVATIONS:**

1.  As we know that there are three layers in this problem

    Input --------> hidden ------------> output

2.  So from the Input --------> hidden layer, we have W1 and b1 connecting the layers

3.  From hidden ----------> output layer, we have W2 and b2 connecting the layers

4.  We have first computed the values of Weights and the bias in both the layers

5.  Then the weights and the bias in both the layers have been updated to bring the predicted value equal to the actual value.

#### **Step 4: Define the Activation Functions**

```python
def sigmoid(z):
  return(1/(1+np.exp(-z)))

def relu(z):
  return(np.maximum(0,z))

def relu_derivative(z):
  return(z > 0).astype(float)
```

#### **Step 5: Perform Forward Propogation**

```python
#### perform the forward propogation
def forward(X):
  #### make some variables as global
  global Z1, A1, Z2, A2

  #### Perform forward propogation from the input to hidden layer
  Z1 = np.dot(X,W1) + b1
  A1 = relu(Z1)

  ### Perform forward propogation from the hidden to output layer
  Z2 = np.dot(A1,W2) + b2
  A2 = sigmoid(Z2)

  return(A2)
```

#### **Step 6: Compute the Loss Function**

```python
#### define the loss function
def loss_function(Y_true, Y_pred):
  #### define the number of samples
  m = Y_true.shape[0]
  #### compute the loss function
  loss = -(1/m)*np.sum(
      Y_true*np.log(Y_pred + 1e-8) + (1-Y_true)*np.log(1-Y_pred + 1e-8)
  )
  return(loss)
```

#### **Step 7: Perform the BackPropogation**

```python
#### perform the backpropogation
def backward(X, y, learning_rate=0.01):
    global W1, b1, W2, b2, Z1, A1, A2

    m = X.shape[0]

    # Output layer error
    dZ2 = A2 - y
    dW2 = np.dot(A1.T, dZ2) / m
    db2 = np.sum(dZ2, axis=0, keepdims=True) / m

    # Hidden layer error
    dA1 = np.dot(dZ2, W2.T)
    dZ1 = dA1 * relu_derivative(Z1)
    dW1 = np.dot(X.T, dZ1) / m
    db1 = np.sum(dZ1, axis=0, keepdims=True) / m

    # Update weights
    W1 -= learning_rate * dW1
    b1 -= learning_rate * db1
    W2 -= learning_rate * dW2
    b2 -= learning_rate * db2
```

#### **Step 8: Perform the Training loop**

```python
epochs = 1000
for i in range(epochs):
  #### perform the forward propogation
  Y_pred = forward(X_train)
  Y_pred = np.squeeze(Y_pred)
  #### compute the loss function
  loss = loss_function(Y_train, Y_pred)
  #### perform the backward propogation
  backward(X_train, Y_train)

  if (i % 100 == 0):
    print(f"Epoch {i}, Loss: {loss:.4f}")
```

#### **Step 9: Perform the prediction**

```python
#### perform the prediction
def predict(X):
  Y_pred = forward(X)
  return(Y_pred > 0.5).astype(int)
```

#### **Step 10: Get the accuracy of the model**

```python
Y_test_pred = predict(X_test)

accuracy = np.mean(Y_test_pred == Y_test)*100.0
print("Test Accuracy:", accuracy)
```
