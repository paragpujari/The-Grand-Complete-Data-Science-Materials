#### Step 5: Initialize the weights

```python
#### Initialize the weights

weights = np.array([0.1,0.1])
```

```python
weights
```

#### OBSERVATIONS:

1.  As there are two inputs namely Hours_Studied, Atendance, so two weights have been initialized.

#### Step 6: Initilize the bias

```python
#### Initialize the bias

bias = 0
```

```python
bias
```

#### OBSERVATIONS:

1. Bias has been added to the inputs so that if all the input values are all zero, then bias will add some weightage to the output.

#### Step 7: Create the Activation Function

```python
##### create the step activation function
def step_activation_function(x):
    #### if the input is above the threshold
    if(x >= 0):
        return(1)
    return(0)
```

```python
#### call the step activation function
step_activation_function(5)
```

#### OBSERVATION:

1.  Here the step activation function is created.

2.  A step activation function is one of the simplest activation function used in Artificial Neural Networks that determines whether the neuron should fire or not.

3.  It has two values:---

    (a.)    1 ------------>  If the input value is above the threshold.

    (b.)    0 ------------>  If the input is below the threshold.


#### Step 8: Set the learning rate

```python
#### determine the learning rate
learning_rate = 0.01
```

#### OBSERVATIONS:

1.  learning rate determines how much does the weights change.

#### Step 9: Train the Perceptron

```python
#### Train the perceptron
for epoch in range(100):
    for i in range(len(X)):
        #### Calculate the weighted sum
        z = np.dot(X[i],weights)+bias
        #### Apply the activation function on the weighted sum
        predicted = step_activation_function(z)
        #### get the error
        error = Y[i] - predicted
        #### update the weighted sum and the bias
        weights = weights + (learning_rate*error*X[i])
        bias    = bias    + (learning_rate*error)

print("Weights is:", weights)
print("Bias    is:", bias)
```

#### OBSERVATIONS:

1.  The Perceptron is trained for 100 iterations.

2.  For each iteration, it is looped through each student record

    (a.)  Weighted sum is calculated for each student record.

    (b.)  Then activation function is applied on each input record to get the predicted output for each student record.

    (c.)  The error is calculated for each student record by finding the diference between the actual - predicted value.

    (d.)  The weights and the bias are updated using the formulas given above in the code.

#### Step 10: Predict whether the student has passed or failed based on the test data

```python
#### define the test data
test_data = np.array([6, 75])

#### Calculate the weighted sum based on the test data
z = np.dot(test_data,weights) + bias
print(z)

#### Apply the step activation function of the weighted sum to get the predicted output
predicted = step_activation_function(z)
print("Predicted Output is:", predicted)

#### Student pass or fail
if(predicted == 1):
    print("Student will pass")
else:
    print("student will fail")
```

#### Step 11: Perform prediction on the series of the students marks

```python
#### define the students marks
test_students = np.array([
    [2, 35],   # Likely Fail
    [4, 55],   # Borderline
    [6, 75],   # Likely Pass
    [8, 90]    # Likely Pass
])

for x in test_students:
    #### Calculate the weighted sum
    z = np.dot(x,weights) + bias
    print(z)
    #### Apply the step activation function on the test data
    predictions = step_activation_function(z)
    print(predictions)

    print(
        f"Hours={x[0]}, Attendance={x[1]} -> "
        f"{'Pass' if predictions == 1 else 'Fail'}"
    )
  
```
