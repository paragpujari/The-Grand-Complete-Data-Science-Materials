**--------------------Basic Level Interview Questions-----------------**

#### **1. Write a PySpark program to create a SparkSession and display the Spark version**

```python
#### Create a new spark session
from pyspark.sql import SparkSession
session = SparkSession.builder.appName('sparksession').getOrCreate()
print(session)

#### display the spark version
print(session.version)
```

#### **OBSERVATIONS:**

1. SparkSession functionality is imported from pyspark.sql

2. Then we are building the spark session by using the following modules:--

    (a.) SparkSession ----> It is the functionality imported from pyspark.sql

    (b.) builder ----------> It is used to build the spark configurations.

    (c.) appName ----------> It is used to provide the name of the spark applications as 'SparkSession'

    (d.) getorCreate() ------> It is used to return the existing spark session or get a new spark session.

3. Then using the object of spark session, we are getting the version of pyspark.

4. Spark session is used as it helps the user to interact with the functionalities of pyspark.

5.  It is used

    (a.)  to execute the spark queries

    (b.)  helps the user to access the spark context

    (c.) helps the user to read the data from the dataframe

    (d.) loads the data

**11. Write a PySpark program to define a custom schema while reading a CSV file**

```python
#### Define the custom schema
from pyspark.sql.types import *

schema = StructType([
    StructField('employee_id', IntegerType(), True)              ,
    StructField('name'       , StringType() ,  True)             ,
    StructField('department' , StringType() ,  True)             ,
    StructField('salary'     , IntegerType(),  True)             ,
    StructField('city'       , StringType() ,  True)             ,
    StructField('job_title'  , StringType() ,  True)
])

#### read the csv file with the new schema
df = session.read.csv('/content/Employee.csv', header=True, schema = schema)

df.show()
```
