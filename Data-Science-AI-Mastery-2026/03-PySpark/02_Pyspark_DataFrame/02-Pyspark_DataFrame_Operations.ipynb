**1. Write a PySpark program to display all records from an employee DataFrame**

```python
from pyspark.sql import SparkSession

#### create a spark session
session = SparkSession.builder.appName('Session').getOrCreate()

#### Get the spark version
print(session.version)

df = session.read.csv('/content/Employee.csv', header=True, inferSchema=True)
df.show()
```

**24. Write a PySpark program to add a column Employee Grade based on salary using when()**

```python
df.withColumn(
    "Grade",
    when(col("salary")>80000,'A')
    .when((col('salary')>60000) & (col('salary') < 80000), 'B')
    .otherwise('C')
    .alias("Grade")
).show()
```

**25. Write a PySpark program to split the employee name into First Name and Last Name**

```python
from pyspark.sql.functions import split

df.withColumn("First Name",split(col("name")," ").getItem(0)).withColumn("Last Name",split(col('name')," ").getItem(1)).show()
```

**26. Write a PySpark program to concatenate employee name and department into a new column Employee_Info**

```python
from pyspark.sql.functions import  concat_ws

df.withColumn('Employee_Info',concat_ws('-',col('name'),col('department'))).show()
```

**27. Write a PySpark program to display employees whose salary is greater than the average salary**

```python
from pyspark.sql.functions import avg

avg_salary = df.select(avg(col('salary'))).collect()[0][0]

avg_salary

df.filter(col('salary')>avg_salary).orderBy(col('salary').asc()).show()
```

**28. Write a PySpark program to convert all employee names to uppercase**

```python
df.withColumn('Upper_Name',upper(col('name'))).show()
```

**29. Write a PySpark program to assign a sequential row number to each employee after sorting by salary in descending order**

```python
from pyspark.sql.window import Window
from pyspark.sql.functions import row_number
window = Window().orderBy(col('salary').desc())
row = row_number().over(window)

df.withColumn('Row_Number',row).show()
```

**30. Write a PySpark program to select employees whose salary is in the top 10% of all employees**

```python
from pyspark.sql.functions import percent_rank

window = Window().orderBy(col('salary'))
window_percent = percent_rank().over(window)

df.withColumn('Window_Percentage',window_percent).filter(col('Window_Percentage')>0.9).show()
```
