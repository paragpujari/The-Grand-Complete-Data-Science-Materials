#### Step 5: Divide the dataset into independent and dependent features

```python
#### independent features

X = df.drop(columns = 'MedHouseVal', axis=1)

print(X)
```

```python
#### dependent features

Y = df['MedHouseVal']

print(Y)
```

#### Step 6: Divide the independent and dependent features into training and testing data

```python
from  sklearn.model_selection   import  train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

#### Step 7: Perform Feature Scaling on the inputs

```python
from sklearn.preprocessing import StandardScaler

#### create an object for Standard Scaler

sc = StandardScaler()

#### using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)

X_train_scaled
```

```python
X_test_scaled = sc.transform(X_test)
```

#### Step 8: Train the Random Forest Regressor

```python
from    sklearn.ensemble    import  RandomForestRegressor

#### create an object for Random Forest Regressor

ranreg = RandomForestRegressor(
    n_estimators    =   100 ,
    random_state    =   42  ,
    max_depth       =   3   ,
    n_jobs          =   -1
)


#### using he object for Random Forest Regressor, train the model

ranreg.fit(X_train_scaled, Y_train)
```

#### OBSERVATIONS:

1.  The object for Random Forest Regressor has been created with the following parameters :----

    (a.)    n_estimators = 100 ==================> 100 decision trees to be used

    (b.)    random_state = 42  ==================> reproducibility

    (c.)    max_depth = 3      ==================> Maximum depth for each decision tree is 3

    (d.)    n_jobs    =  -1    ===================> printing the frequency of the output training data

2. The andom Forest Regressor model has been trained with the help of the training data.

#### Step 9: Predict the model

```python
Y_pred = ranreg.predict(X_test_scaled)
```

```python
Y_pred
```

#### Step 10: Evaluate the Regression Metrics

```python
from    sklearn.metrics             import   mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

print("==============================Mean Absolute Error========================================================")

mae = mean_absolute_error(Y_test, Y_pred)

print("Mean Absolute Error of the model is:", mae)


print("==============================Mean Squared Error========================================================")

mse = mean_squared_error(Y_test, Y_pred)

print("Mean Squared Error of the model is:", mse)


print("==============================Root Mean Squared Error========================================================")

rmse = root_mean_squared_error(Y_test, Y_pred)

print("Root Mean Squared Error of the model is:", rmse)


print("==============================R2 Score========================================================")

r2 = r2_score(Y_test, Y_pred)

print("R2 Score of the model is:", r2)
```

#### OBSERVATIONS:

1.  The R2 Score of the Random Forest Regressor Model using California Housing dataset is 0.53.

#### Step 13: Perform HyperParameter Tuning with GridSearchCV

```python
#### define the pipeline

from sklearn.pipeline   import  Pipeline

#### create the object for pipeline

pipeline = Pipeline([
    ("sc"           ,           StandardScaler())           ,
    ("ranreg"       ,           RandomForestRegressor())
])
```

```python
#### define the param grid

param_grid = {
    "ranreg__n_estimators"      :       [100,200]           ,
    "ranreg__max_depth"         :       [5,10,15,None]      ,
    "ranreg__min_samples_split" :       [2,5,10]            ,
    "ranreg__min_samples_leaf"  :       [1,2,4]             ,
    "ranreg__max_features"      :       ['sqrt','log2']
}
```

```python
#### perform the cross validation

cv = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
#### create the object for GridSearchCV

grid = GridSearchCV(
    pipeline                                                ,
    param_grid              =           param_grid          ,
     cv                     =              cv               ,
     scoring                =            'r2'               ,
     n_jobs                 =               -1  
)
```

```python
#### using the object of GridSearchCV, train the model

grid.fit(X_train, Y_train)
```

```python
#### Obtain the best parameter and estimator of the model

best_parameter = grid.best_params_

print("Best Parameter of the model is:", best_parameter)


best_estimator = grid.best_estimator_

print("Best Estimator of the model is:", best_estimator)
```

```python
#### predict the model

Y_pred_grid = best_estimator.predict(X_test)
```

```python
Y_pred_grid
```

```python
##### Evaluate the Regression Metrics

from    sklearn.metrics             import   mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score


print("==============================Mean Absolute Error========================================================")

maegrid = mean_absolute_error(Y_test, Y_pred_grid)

print("Mean Absolute Error of the model is:", maegrid)



print("==============================Mean Squared Error========================================================")

msegrid = mean_squared_error(Y_test, Y_pred_grid)

print("Mean Squared Error of the model is:", msegrid)



print("==============================Root Mean Squared Error========================================================")

rmsegrid = root_mean_squared_error(Y_test, Y_pred_grid)

print("Root Mean Squared Error of the model is:", rmsegrid)


print("==============================R2 Score========================================================")

r2grid = r2_score(Y_test, Y_pred_grid)

print("R2 Score of the model is:", r2grid)
```

#### OBSERVATIONS:

1.  The R2 Score of the Random Forest Regressor Model using California Housing dataset is 0.81.

2.  The R2 Score of the Random Forest Regressor has increased from 0.53 to 0.81 by performing Hyper Parameter Tuning using GridSerachCV.
