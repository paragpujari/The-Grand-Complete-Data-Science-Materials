#### Step 4: Perform Feature Engineering

```python
cols = ['mainroad',	'guestroom',	'basement',	'hotwaterheating' ,'airconditioning', 'prefarea']


#### convert the categorical features into the numerical column by mapping them

for x in cols:
    df[x] = df[x].map({"yes":1, "no":0})
```

```python
from sklearn.preprocessing  import LabelEncoder

#### create an object for Label Encoder

label = LabelEncoder()

#### transform the inputs for "furnishingstatus"

df['furnishingstatus'] = label.fit_transform(df['furnishingstatus'])
```

#### Step 5: Divide the dataset into independent and dependent features

```python
#### independent features

X = df.drop(columns='price',axis=1)

print(X)
```

```python
#### dependent features

Y = df['price']

print(Y)
```

#### Step 6: Divide the independent and dependent features into training and testing data

```python
from  sklearn.model_selection  import  train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

#### Step 7: Perform Hyper Parameter Tuning using GridSearchCV

```python
#### create the pipeline

from  sklearn.pipeline  import  Pipeline

#### create an object for Pipeline

pipeline = Pipeline([
    ("sc"         ,        StandardScaler())             ,
    ("ridge"      ,        Ridge())
]) 
```

```python
#### define the parameter grid

param_grid = {
    "ridge__alpha"     :         [0.01, 0.1,    1,  10, 100]
}
```

```python
#### perform the KFold Cross Validation

cv = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
#### create an object for GridSearchCV

grid = GridSearchCV(
    pipeline                ,
    param_grid              ,
    cv          =    cv     ,
    scoring     =    'r2'
)
```

```python
##### using the object of GridSearchCV, train the model

grid.fit(X_train, Y_train)
```

```python
##### get the best parameters and estimators of the model

best_parameters = grid.best_params_

print("Best Parameters of the model is:", best_parameters)
```

```python
best_estimator = grid.best_estimator_

print("Best Estimators of the model is:", best_estimator)
```

```python
#### predict the model

Y_pred_grid = best_estimator.predict(X_test)
```

```python
Y_pred_grid
```

#### Step 8: Calculate the Slope and Intercept of the Model

```python
print("Slope of the model is:", best_estimator.named_steps['ridge'].coef_[0])

print("Intercept of the model is:", best_estimator.named_steps['ridge'].intercept_)
```

#### Step 9: Perform the Regression Evaluation Metrics

```python
from  sklearn.metrics  import  mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

"================================Mean Absolute Error==========================================="

mae = mean_absolute_error(Y_test, Y_pred_grid)

print("Mean Absolute Error of the model is:", mae)
```

```python
"================================Mean Squared Error==========================================="

mse = mean_squared_error(Y_test, Y_pred_grid)

print("Mean Squared Error of the model is:", mse)
```

```python
"=============================Root Mean Squared Error==========================================="

rmse = root_mean_squared_error(Y_test, Y_pred_grid)

print("Root Mean Squared Error of the model is:", rmse)
```

```python
"=============================R2 Score========================================================="

r2 = r2_score(Y_test, Y_pred_grid)

print("R2 Score of the model is:", r2)
```

```python
"=============================R2 adjusted========================================================="

n = X_test.shape[0]   #### Number of rows

p = X_test.shape[1]   #### Number of independent features


adjusted_r2 = 1-((1-r2)*(n-1))/(n-p-1)


print("Adjusted R2 Score of the model is:", adjusted_r2)
```
