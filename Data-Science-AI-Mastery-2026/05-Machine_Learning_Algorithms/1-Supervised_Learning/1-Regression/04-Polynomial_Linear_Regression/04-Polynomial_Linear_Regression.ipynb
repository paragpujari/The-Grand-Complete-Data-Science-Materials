==================== (a.)  POLYNOMIAL REGRESSION ======================================

#### Step 12: Perform the polynomial degrees

```python
from  sklearn.preprocessing import PolynomialFeatures


##### create the object for Polynomial degrees

poly = PolynomialFeatures(degree=4)


##### using the object for Polynomial degrees, transform the inputs

X_train_poly = poly.fit_transform(X_train)

print(X_train_poly)
```

```python
X_test_poly = poly.transform(X_test)

print(X_test_poly)
```

#### OBSERVATIONS:

1.  Here all the input polynomial features have been scaled in the same scale range.

#### Step 13: Train the Polynomial Linear Regression

```python
from  sklearn.linear_model  import  LinearRegression

#### create an object for linear regression

lin_poly = LinearRegression()

#### using the object of linear regression, train the model

lin_poly.fit(X_train_poly, Y_train)
```

#### Step 14: Predict the Model

```python
#### Predict the model

Y_pred_poly = lin_poly.predict(X_test_poly)
```

```python
Y_pred_poly
```

#### Step 15: Find the slope and intercept of the model

```python
print("Slope of the model is:"    , lin_poly.coef_[0])
print("Intercept of the model is:", lin_poly.intercept_)
```

#### Step 16: Find the accuracy of the model

```python
#### Find the accuracy of the model

ac_poly = r2_score(Y_test, Y_pred_poly)

print("R2 Score of the model is:", ac_poly)
```

#### OBSERVATIONS:

1.  The R2 Score of the model using linear regression is 0.9024

2.  The R2 Score of the model using Polynomial linear regression is 0.9029

3.  The R2 Score (Polynomial linear) is almost same as the linear regression regression.

##### Step 17: Perform Hyper Parameter Tuning using Grid Search CV

```python
#### Create the pipeline

from  sklearn.linear_model  import  Ridge, Lasso, ElasticNet

from  sklearn.pipeline  import  Pipeline

pipeline = Pipeline([
    ("sc"            ,   StandardScaler())                        ,
    ("ridge"         ,   Ridge(alpha = 0.01))
])
```

```python
#### Create the parameter  grid

param_grid       =        {
    "ridge__alpha"        :       [0.01,    0.1,    1,  10, 100]
}
```

```python
#### perform the Startified K-Fold Cross Validation

from  sklearn.model_selection  import  KFold


cv = KFold(n_splits = 5, shuffle = True, random_state = 42)
```

```python
#### perform the Grid Search CV

grid = GridSearchCV(
    pipeline                                     ,
    param_grid   =    param_grid                 ,
    cv           =        cv                     ,
    scoring      =     'r2'                     
)
```

```python
#### Train the grid search cv

grid.fit(X_train_poly, Y_train)
```

```python
#### perform the best parameters and the estimators of the model

best_parameters = grid.best_params_

print("Best parameters of the model is:", best_parameters)



best_estimator = grid.best_estimator_

print("Best estimator of the model is:", best_estimator)
```

```python
#### predict the model

Y_pred_grid_poly = best_estimator.predict(X_test_poly)
```

```python
Y_pred_grid_poly
```

```python
#### Find the accuracy of the model


ac_poly = r2_score(Y_test, Y_pred_grid_poly)

print("R2 Score of the model is:", ac_poly)
```

#### OBSERVATIONS:

1.  The R2 Score of the model using linear regression is 0.9024

2.  The R2 Score of the model using Polynomial linear regression is 0.9029

3.  The R2 Score (Polynomial linear) is almost same as the linear regression regression.

4.  The R2 Score of the Polynomial Linear regression using Grid Search CV is 0.9056 which is slightly higher than linear and Polynomial Linear regression.
