#### Gradient Boosting Regressor

Gradient Boosting is an ensemble learning algorithm that combines multiple weak learners sequentially to reduce prediction errors.

#### Algorithms:---

(i)  Every weak learner runs simultaneously .

(ii) Each learner accepts the input and performs the predictions.

(iii) It reduces the prediction error caused by the previous learner and moves to the next.

(iv)  Then the strong predictive learned has the weighted average of all the outputs of the weak learner.

#### Step 7:  Perform Feature Scaling on the inputs

```python
from  sklearn.preprocessing import StandardScaler

#### create an object for Standard Scaler

sc = StandardScaler()

#### using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)
```

#### Step 8: Train the Gradient Boosting Regressor Model

```python
from sklearn.ensemble   import  GradientBoostingRegressor


#### create an object for Gradient Boosting Regressor

gbr = GradientBoostingRegressor(
    random_state = 42
)


#### using the object for GradientBoostingRegressor, train the model

gbr.fit(X_train_scaled, Y_train)
```

#### OBSERVATIONS:

1.  The object for Gradient Boosting Regressor has been initialized.

2.  The Gradient Boosting Regressor  model has been trained with the help of the training data.

#### Step 9: Predict the Model

```python
Y_pred = gbr.predict(X_test_scaled)
```

```python
Y_pred
```

#### Step 10: Evaluate the Regression Metrics

```python
from    sklearn.metrics             import   mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

print("==============================Mean Absolute Error========================================================")

mae = mean_absolute_error(Y_test, Y_pred)

print("Mean Absolute Error of the model is:", mae)


print("==============================Mean Squared Error========================================================")

mse = mean_squared_error(Y_test, Y_pred)

print("Mean Squared Error of the model is:", mse)


print("==============================Root Mean Squared Error========================================================")

rmse = root_mean_squared_error(Y_test, Y_pred)

print("Root Mean Squared Error of the model is:", rmse)


print("==============================R2 Score========================================================")

r2 = r2_score(Y_test, Y_pred)

print("R2 Score of the model is:", r2)
```

#### OBSERVATIONS:

1.  The R2 Score of the Gradient Boosting Regressor Model using California Housing dataset is 0.77.

#### Step 13: Perform Hyper Parameter Tuning with GridSearchCV

```python
#### define the pipeline

from sklearn.pipeline   import Pipeline

#### create an object for Pipeline

pipeline    =   Pipeline([
    ("sc"           ,       StandardScaler())           ,
    ("gbr"          ,       GradientBoostingRegressor())
])
```

```python
#### define the param grid

param_grid = {
    "gbr__n_estimators"         :   [15,20]                     ,
    "gbr__learning_rate"        :   [0.01,0.05,0.1]             ,
    "gbr__max_depth"            :   [3,5,7,9]                   ,
    "gbr__min_samples_split"    :   [2,5]                       ,
    "gbr__min_samples_leaf"     :   [1,2]                       , 
    "gbr__subsample"            :   [0.8,1.0]
}
```

```python
#### perform the cross validation

cv = KFold(n_splits = 3, shuffle = True, random_state = 42)
```

```python
#### create the object of GridSearchCV

grid = GridSearchCV(
    pipeline                                                    ,
    param_grid      =       param_grid                          ,
    cv              =         cv                                ,
    scoring         =        'r2'                               ,
    n_jobs          =         -1
)
```

```python
#### using the object of GridSearchCV, train the model

grid.fit(X_train, Y_train)
```

```python
#### Obtain the best parameter and the estimator of the model

best_parameter = grid.best_params_
print("Best Parameter of the model is:", best_parameter)


best_estimator = grid.best_estimator_
print("Best Estimator of the model is:", best_estimator)
```

```python
#### Predict the model

Y_pred_grid = best_estimator.predict(X_test)
```

```python
Y_pred_grid
```

```python
##### Evaluate the Regression Metrics

from    sklearn.metrics             import   mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score


print("==============================Mean Absolute Error========================================================")

maegrid = mean_absolute_error(Y_test, Y_pred_grid)

print("Mean Absolute Error of the model is:", maegrid)



print("==============================Mean Squared Error========================================================")

msegrid = mean_squared_error(Y_test, Y_pred_grid)

print("Mean Squared Error of the model is:", msegrid)



print("==============================Root Mean Squared Error========================================================")

rmsegrid = root_mean_squared_error(Y_test, Y_pred_grid)

print("Root Mean Squared Error of the model is:", rmsegrid)


print("==============================R2 Score========================================================")

r2grid = r2_score(Y_test, Y_pred_grid)

print("R2 Score of the model is:", r2grid)
```

#### OBSERVATIONS:

1.  The R2 Score of the Gradient Boosting Regressor Model using California Housing dataset is 0.78.

2.  The R2 Score of the Gradient Boosting Regressor has increased from 0.77 to 0.78 by performing Hyper Parameter Tuning using GridSerachCV.
