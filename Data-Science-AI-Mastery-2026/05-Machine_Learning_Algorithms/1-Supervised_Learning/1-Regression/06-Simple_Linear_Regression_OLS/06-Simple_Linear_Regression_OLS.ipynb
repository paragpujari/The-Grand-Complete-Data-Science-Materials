#### Step 5: Divide the dataset into independent and dependent features

```python
#### independent features

X = df.drop(columns='Height',axis=1)

print(X)
```

```python
#### dependent features

Y = df[['Height']]

print(Y)
```

#### Step 6: Divide the independent and dependent features into training and testing data

```python
from  sklearn.model_selection  import  train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

#### Step 7: Perform Feature Scaling on the inputs

```python
from  sklearn.preprocessing import  StandardScaler

#### create an object for Standard Scaler

sc = StandardScaler()

#### using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)

X_test_scaled  = sc.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

#### OBSERVATIONS:

1.  All the input features are in the same scale range

#### Step 8: Perform Linear Regression using Ordinary Least Squares Method

```python
import  statsmodels.api  as  sm

#### create an object for OLS and train the model

model = sm.OLS(X_train_scaled, Y_train).fit()


print(model)
```

```python
#### predict the model

Y_pred = model.predict(X_test_scaled)
```
