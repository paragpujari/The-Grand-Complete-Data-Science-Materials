```python
############################################################
#### Simple Linear Regression using Gradient Descent
############################################################
import numpy as np
import matplotlib.pyplot as plt

############################################################
#### Sample Dataset
############################################################
X = np.array([1, 2, 3, 4, 5]  , dtype=float)
Y = np.array([5, 7, 9, 11, 13], dtype=float)

############################################################
#### Initialize Parameters
############################################################
m = 0        # Slope
b = 0        # Intercept
learning_rate = 0.01
epochs = 1000
n = len(X)

############################################################
#### Gradient Descent Algorithm
############################################################

for i in range(epochs):
    #### Predicted values
    Y_pred = m * X + b

    ########################################################
    #### Calculate Gradients
    ########################################################

    dm = (-2/n) * np.sum(X * (Y - Y_pred))
    db = (-2/n) * np.sum(Y - Y_pred)

    ########################################################
    #### Update Parameters
    ########################################################

    m = m - learning_rate * dm
    b = b - learning_rate * db

    ########################################################
    #### Print Cost after every 100 epochs
    ########################################################

    if i % 100 == 0:
        cost = (1/n) * np.sum((Y - Y_pred) ** 2)
        print(f"Epoch {i} ---> Cost = {cost:.4f}")

############################################################
#### Final Parameters
############################################################

print("\nFinal Slope (m):", m)
print("Final Intercept (b):", b)

############################################################
#### Final Predictions
############################################################

Y_final_pred = m * X + b
print("\nPredicted Values:")
print(Y_final_pred)

############################################################
#### Plot the Regression Line
############################################################

plt.scatter(X, Y, label='Actual Data')
plt.plot(X, Y_final_pred, label='Regression Line')
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Simple Linear Regression using Gradient Descent")
plt.legend()
plt.show()
```
