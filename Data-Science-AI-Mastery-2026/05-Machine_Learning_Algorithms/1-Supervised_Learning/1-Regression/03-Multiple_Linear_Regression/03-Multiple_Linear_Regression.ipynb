#### Step 4: Perform Feature Engineering

```python
cols = ['mainroad',	'guestroom',	'basement',	'hotwaterheating' ,'airconditioning', 'prefarea']


#### convert the categorical features into the numerical column by mapping them

for x in cols:
    df[x] = df[x].map({"yes":1, "no":0})
```

```python
#### perform the label encoding on the furnishingstatus column

from  sklearn.preprocessing  import  LabelEncoder

#### create an object for Label Encoder

label = LabelEncoder()

#### using the object of label encoder, transform the input

df['furnishingstatus'] = label.fit_transform(df['furnishingstatus'])
```

#### Step 5: Divide the dataset into independent and dependent features

```python
#### independent features

X = df.drop(columns='price',axis=1)

print(X)
```

```python
#### dependent features

Y = df[['price']]

print(Y)
```

#### Step 6: Divide the independent and dependent features into training and testing data

```python
from  sklearn.model_selection  import  train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

#### Step 7: Perform Feature Scaling on the inputs

```python
from  sklearn.preprocessing  import  StandardScaler

#### create an object for Standard Scaler

sc = StandardScaler()

### using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)

X_test_scaled = sc.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

#### OBSERVATIONS:

1.  The input features have been scaled in the same input ranges.

#### Step 8: Train the Linear Model

```python
from  sklearn.linear_model  import  LinearRegression

#### create an object for linear regression

ridge = Ridge(alpha=0.1)

#### using the object for linear regression, train the model

ridge.fit(X_train_scaled, Y_train)
```

#### Step 9: Predict the Model

```python
Y_pred_ridge = ridge.predict(X_test_scaled)
```

```python
Y_pred_ridge
```

#### Step 10: Calculate the Slope and Intercept of the model

```python
print("Slope of the model is:", ridge.coef_[0])


print("Intercept of the model is:", ridge.intercept_[0])
```

#### Step 11: Evaluate the Regression Metrics

```python
from    sklearn.metrics          import  mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

mae = mean_absolute_error(Y_test, Y_pred_ridge)
print("Mean Absolute Error of the model is:", mae)
```

```python
mse = mean_squared_error(Y_test, Y_pred_ridge)
print("Mean Squared Error of the model is:", mse)
```

```python
rmse = root_mean_squared_error(Y_test, Y_pred_ridge)
print("Root Mean Squared Error of the model is:", rmse)
```

```python
r2ridge = r2_score(Y_test, Y_pred_ridge)
print("R2 Score of the model is:", r2ridge)
```

##### Step 12: Perform Hyper Parameter Tuning using GridSearchCV

```python
#### define the pipeline

from  sklearn.pipeline  import Pipeline

#### create an object for pipeline

pipeline = Pipeline([
    ("sc"      ,       StandardScaler())             ,
    ("ridge"   ,       Ridge(alpha=0.1))
])
```

```python
#### define the parameter grid

param_grid = {
    "ridge__alpha"  :  [0.01, 0.1, 1, 10, 100]
}
```

```python
#### perform the cross validation

from  sklearn.model_selection import KFold

cv = KFold(n_splits=5, shuffle=True, random_state=42)
```

```python
#### Create the object of GridSearchCV

grid = GridSearchCV(
    pipeline                            ,
    param_grid     =   param_grid       ,
    cv             =     cv             ,
    scoring        =     'r2'          
)
```

```python
#### Train the model using the object of GridSearch CV

grid.fit(X_train_scaled, Y_train)
```

#### OBSERVATIONS:

1. The GridSearchCV Model has been trained with the help of the training data.

```python
#### get the best parameters and the estimators for the model

best_parameters = grid.best_params_

print("Best Parameters of the model is:", best_parameters)
```

```python
best_estimators = grid.best_estimator_

print("Best Estimators of the model is:", best_estimators)
```

```python
#### using the best estimator, predict the model


Y_pred_grid = best_estimators.predict(X_test_scaled)
```

```python
Y_pred_grid
```

```python
#### Evaluate the Regression Metrics using GridSearchCV

from    sklearn.metrics          import  mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

maegrid = mean_absolute_error(Y_test, Y_pred_grid)
print("Mean Absolute Error of the model is:", maegrid)
```

```python
msegrid = mean_squared_error(Y_test, Y_pred_grid)
print("Mean Squared Error of the model is:", msegrid)
```

```python
rmsegrid = root_mean_squared_error(Y_test, Y_pred_grid)
print("Root Mean Squared Error of the model is:", rmsegrid)
```

```python
r2grid = r2_score(Y_test, Y_pred_grid)
print("R2 Score of the model is:", r2grid)
```
