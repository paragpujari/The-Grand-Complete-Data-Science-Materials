#### Step 4: Perform Feature Engineering and Data Cleaning

```python
#### Convert all the categorical features into numerical values

cols = ['mainroad',	'guestroom','basement',	'hotwaterheating',	'airconditioning',	'prefarea']

for x in cols:
    df[x] = df[x].map({"yes":1,"no":0})
```

```python
#### perform label encoding on the "furnishingstatus" column

from  sklearn.preprocessing import  LabelEncoder

#### create an object for LabelEncoder

label = LabelEncoder()

#### using the object for LabelEncoder, transform the inputs

df['furnishingstatus'] = label.fit_transform(df['furnishingstatus'])
```

#### Step 5: Divide the dataset into independent and dependent features

```python
#### independent features

X = df.drop(columns='price',axis=1)

print(X)
```

```python
#### dependent features

Y = df['price']

print(Y)
```

#### Step 6: Divide the independent and dependent features into training and testing data

```python
from  sklearn.model_selection   import  train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

#### Step 7: Perform Feature Scaling on the inputs

```python
from    sklearn.preprocessing   import  StandardScaler

#### create an object for Standard Scaler

sc = StandardScaler()

#### using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)
```

```python
X_train_scaled
```

```python
X_test_scaled = sc.transform(X_test)
```

```python
X_test_scaled
```

#### OBSERVATIONS:

1.  All the input features have been scaled in the same range between -1 to 1.

#### Step 8: Train the Regularization Techniques

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

print("================================Ridge Regression=========================================================")

ridge = Ridge(alpha=0.1)


ridge.fit(X_train_scaled,Y_train)


#### predict the model

Y_pred_ridge = ridge.predict(X_test_scaled)

print(Y_pred_ridge)


#### Calculate the Slope and Intercept of the model

print("Slope of the model is:", ridge.coef_[0])
print("Intercept of the model is:", ridge.intercept_)


#### Evaluate the Regression Metrics

from    sklearn.metrics          import  mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

mae = mean_absolute_error(Y_test, Y_pred_ridge)
print("Mean Absolute Error of the model is:", mae)


mse = mean_squared_error(Y_test, Y_pred_ridge)
print("Mean Squared Error of the model is:", mse)


rmse = root_mean_squared_error(Y_test, Y_pred_ridge)
print("Root Mean Squared Error of the model is:", rmse)


r2 = r2_score(Y_test, Y_pred_ridge)
print("R2 Score of the model is:", r2)
```

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

print("================================Lasso Regression=========================================================")

lasso = Lasso(alpha=1.0)


lasso.fit(X_train_scaled,Y_train)


#### predict the model

Y_pred_lasso = lasso.predict(X_test_scaled)

print(Y_pred_lasso)


#### Calculate the Slope and Intercept of the model

print("Slope of the model is:", lasso.coef_[0])
print("Intercept of the model is:", lasso.intercept_)


#### Evaluate the Regression Metrics

from    sklearn.metrics          import  mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

maelasso = mean_absolute_error(Y_test, Y_pred_lasso)
print("Mean Absolute Error of the model is:", maelasso)


mselasso = mean_squared_error(Y_test, Y_pred_lasso)
print("Mean Squared Error of the model is:", mselasso)


rmselasso = root_mean_squared_error(Y_test, Y_pred_lasso)
print("Root Mean Squared Error of the model is:", rmselasso)


r2lasso = r2_score(Y_test, Y_pred_lasso)
print("R2 Score of the model is:", r2lasso)
```

```python
from sklearn.linear_model import Ridge, Lasso, ElasticNet

print("================================Elastic Net Regression=========================================================")

elasticnet = ElasticNet(alpha=0.1,l1_ratio=0.5)


elasticnet.fit(X_train_scaled,Y_train)


#### predict the model

Y_pred_elasticnet = elasticnet.predict(X_test_scaled)

print(Y_pred_elasticnet)


#### Calculate the Slope and Intercept of the model

print("Slope of the model is:", elasticnet.coef_[0])
print("Intercept of the model is:", elasticnet.intercept_)


#### Evaluate the Regression Metrics

from    sklearn.metrics          import  mean_absolute_error, mean_squared_error, root_mean_squared_error, r2_score

maeelasticnet = mean_absolute_error(Y_test, Y_pred_elasticnet)
print("Mean Absolute Error of the model is:", maeelasticnet)


mseelasticnet = mean_squared_error(Y_test, Y_pred_elasticnet)
print("Mean Squared Error of the model is:", mseelasticnet)


rmseelasticnet = root_mean_squared_error(Y_test, Y_pred_elasticnet)
print("Root Mean Squared Error of the model is:", rmseelasticnet)


r2elasticnet = r2_score(Y_test, Y_pred_elasticnet)
print("R2 Score of the model is:", r2elasticnet)
```
