#### Voting Classifier

1.  Voting Classifier is an ensemble based machine learning technique that combines the prediction from multiple machine learning models and performs the final prediction based on voting.

#### Types of Voting Classifier

1.  Hard Voting ==============>  Hard Voting is a kind of Voting Classifier which uses the majority voting to get the final prediction.

    Ex ==========>

        (a.)    Random Forest       =============>    Output  =   1

        (b.)    Logistic Regression  ============>    Output  =   1

        (c.)    Support Vector Classifier =======>    Output  =   0

        So Hard Voting here performs the majority voting on the above three algorithms and the final output prediction = 1.

2.  Soft Voting ==============> Soft Voting is a kind of Voting Classifier that uses the prediction probabilities to get the final prediction.

    Ex ==========>

        (a.)    Random Forest       =============>    Output  =   0.8

        (b.)    Logistic Regression =============>    Output  =   0.9

        (c.)    Support Vector Classifier =======>    Output  =   0.7

                Average = (0.8 + 0.9 + 0.7)/3

        So Soft Voting performs the average of all the predicted probabilities of the above three algorithms to get the predicted output.

####  Step 11: Perform Hard Voting  Classifier


1.  Hard Voting Classifier is a kind of ensemble technique that uses majority voting to get the output of the final prediction.

    Ex ========>   Logistic Regression ---------------->   1 (Output)

                   Random Forest Classifier ----------->   0 (Output)

                   Support Vector Classifier ---------->   1 (Output)

2.  So on performing the majority voting on the above three algorithms, we get 

        (a.)    the count of the Output Category of (1)  as  2
        
        (b.)    the count of the Output Category of (0)  as  1

3.   As the count of 1 > count of 0 then the final output is 1.

```python
#### Create an object of Hard Voting Classifier

hard_vote = VotingClassifier(
    estimators = [
        ('log'  ,   LogisticRegression(max_iter=1000))                              ,
        ('rcf'  ,   RandomForestClassifier(n_estimators=100, random_state=42))      ,
        ('svc'  ,   SVC(probability=True,random_state=42))
    ], voting='hard'
)
```

#### Step 12: Train the Hard Voting Classifier model

```python
#### Train the model
hard_vote.fit(X_train_scaled, Y_train)
```

#### OBSERVATIONS:

1.  The object of the Hard Voting Classifier has been made with the three estimators 

    (a.)    Logistic Regression

    (b.)    Random Forest Classifier

    (c.)    Support Vector Classifier

2.  Then the model has been trained using the training data.

#### Step 13: Predict the model

```python
#### predict the model
hard_pred = hard_vote.predict(X_test_scaled)
```

```python
hard_pred
```

#### Step 14: Find the accuracy of the model using Hard Voting Classifier

```python
acc_hard = accuracy_score(Y_test, hard_pred)*100.0

print("Accuracy of the model using Hard Voting Classifier is:", acc_hard)
```

#### OBSERVATIONS:

1.  The accuracy of the Hard Voting Classifier model on breast cancer dataset is 98.24

#### OBSERVATIONS OF HARD VOTING CALSSIFIER:

1.  Hard Voting Classifier uses majority voting technique of finding out the maximum number of categories in the predicted output for all the models and then assigning it to the final output.

####  Step 15: Perform Soft Voting  Classifier


1.  Soft Voting Classifier is a kind of ensemble technique that uses the average of all the predicted probabilities to get the output of the final prediction.

    Ex ========>   Logistic Regression ---------------->   0.7 (Output)

                   Random Forest Classifier ----------->   0.8 (Output)

                   Support Vector Classifier ---------->   0.9 (Output)

2.  So on performing the average of all the predicted probabilities on the above three algorithms, we get the average of the predicted probabilities as the output.

```python
#### Create an object of Soft Voting Classifier

soft_vote = VotingClassifier(
    estimators = [
        ('log'  ,   LogisticRegression(max_iter=1000))                              ,
        ('rcf'  ,   RandomForestClassifier(n_estimators=100, random_state=42))      ,
        ('svc'  ,   SVC(probability=True,random_state=42))
    ], voting='soft'
)
```

#### Step 16: Train the Soft Voting Classifier model

```python
#### Train the model
soft_vote.fit(X_train_scaled, Y_train)
```

#### OBSERVATIONS:

1.  The object of the Soft Voting Classifier has been made with the three estimators 

    (a.)    Logistic Regression

    (b.)    Random Forest Classifier

    (c.)    Support Vector Classifier

2.  Then the model has been trained using the training data.

#### Step 17: Predict the model

```python
#### predict the model
soft_pred = soft_vote.predict(X_test_scaled)
```

```python
soft_pred
```

#### Step 18: Find the accuracy of the model using Soft Voting Classifier

```python
acc_soft = accuracy_score(Y_test, soft_pred)*100.0

print("Accuracy of the model using Soft Voting Classifier is:", acc_soft)
```

#### OBSERVATIONS:

1.  The accuracy of the Soft Voting Classifier model on breast cancer dataset is 97.36

#### OBSERVATIONS OF SOFT VOTING CALSSIFIER:

1.  Soft Voting Classifier uses the average of all the predicted probabilities to get the output of the final prediction. So on performing the average of all the predicted probabilities on the above three algorithms, we get the average of the predicted probabilities as the output.
