#### Step 7: Convert the input text into numerical vectors

```python
from  sklearn.feature_extraction.text  import  CountVectorizer


#### create an object for CountVectorizer

count = CountVectorizer(stop_words="english")


#### using the object for CountVectorizer, scale the inputs

X_train_scaled = count.fit_transform(X_train)

print(X_train_scaled)
```

```python
X_test_scaled = count.transform(X_test)

print(X_test_scaled)
```

```python
from  sklearn.naive_bayes import MultinomialNB

#### Create an object for Multinomial Naive Bayes

mnb = MultinomialNB()

#### using the object for Multinomial Naive Bayes

mnb.fit(X_train_scaled, Y_train)
```

#### Step 9: Predict the Model

```python
from        sklearn.metrics             import  accuracy_score, confusion_matrix, classification_report

print("===============================Accuracy score=======================================================")

ac = accuracy_score(Y_test, Y_pred)*100.0

print("Accuracy score of the model is:", ac)



print("===============================Confusion matrix=======================================================")

cm = confusion_matrix(Y_test, Y_pred)

print("Confusion matrix of the model is:", cm)



print("===============================Classification report=======================================================")

cr = classification_report(Y_test, Y_pred)

print("Classification report of the model is:", cr)
```

#### Step 11: Predict whether the email is spam or ham based on the sample data

```python
#### define the sample messages

sample_messages = [
    "Congratulations! You've won a free iPhone. Click here to claim.",
    "Hey, are we still meeting for lunch today?"
]

print(sample_messages)

#### perform the vectorization on the input text

scaled_inputs = count.transform(sample_messages)

print(scaled_inputs)


#### perform the prediction on the vectorized data

prediction = mnb.predict(scaled_inputs)

print(prediction)


print("==========================First Sentence===============================================")

if(prediction[0] == 1):
    print("The email message is Spam")
else:
    print("The email message is Ham")


print("==========================Second Sentence===============================================")

if(prediction[1] == 1):
    print("The email message is Spam")
else:
    print("The email message is Ham")
```

#### Step 12: Perform Hyper Parameter Tuning using GridSearchCV

```python
#### define the pipeline

from sklearn.pipeline  import  Pipeline

#### create the object for pipeline

pipeline = Pipeline([
    ("count"    ,   CountVectorizer())  ,
    ("mnb"      ,   MultinomialNB())
])
```

```python
#### define the param grid

param_grid = {
    "count__ngram_range"        :   [(1,1),(1,2)]              ,           #### unigram + bigram
    "count__max_df"             :   [0.75, 0.9, 1.0]           ,           #### remove most frequent words
    "count__min_df"             :   [1,2,5]                    ,           #### remove the rare words
    "count__stop_words"         :   [None,"english"]           ,           #### remove the words that have no  meaning
    "mnb__alpha"                :   [0.1, 0.5, 1.0, 2.0]                   #### smoothing parameter 
}
```

```python
#### define the cross validation

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
```

```python
#### create the object for GridSearchCV

grid = GridSearchCV(
    pipeline                            ,
    param_grid      =       param_grid  ,
    cv              =          cv       ,
    scoring         =       'accuracy'  ,
    n_jobs          =          -1       ,
    verbose         =           2 
)
```

```python
#### train the grid search cv model

grid.fit(X_train, Y_train)
```

```python
#### Obtain the best parameters and the estimator of the model

best_parameters = grid.best_params_

print("Best Parameter of the model is:", best_parameters)


best_estimator = grid.best_estimator_

print("Best Estimator of the model is:", best_estimator)
```

```python
#### predict the model

Y_pred_grid = best_estimator.predict(X_test)
```

```python
Y_pred_grid
```

```python
#### Evaluate the Classification Metrics


from        sklearn.metrics             import  accuracy_score, confusion_matrix, classification_report

print("===============================Accuracy score=======================================================")

acgrid = accuracy_score(Y_test, Y_pred_grid)*100.0

print("Accuracy score of the model is:", acgrid)



print("===============================Confusion matrix=======================================================")

cmgrid = confusion_matrix(Y_test, Y_pred_grid)

print("Confusion matrix of the model is:", cmgrid)



print("===============================Classification report=======================================================")

crgrid = classification_report(Y_test, Y_pred_grid)

print("Classification report of the model is:", crgrid)
```

#### OBSERVATIONS:

1.  The accuracy of email spam classification using Multinomial Naive Bayes Algorithm has increased from 98.45 to 98.64 using Hyper parameter tuning using Grid Search CV.
