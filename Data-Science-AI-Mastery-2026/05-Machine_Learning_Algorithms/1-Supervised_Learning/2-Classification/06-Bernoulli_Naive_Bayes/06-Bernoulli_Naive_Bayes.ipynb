#### Step 7: Perform Count Vectorization on the inputs

```python
from  sklearn.feature_extraction.text  import  CountVectorizer

###3 create an object for CountVectorizer

count = CountVectorizer(stop_words='english')

#### using the object of count vectorizer, transform the inputs

X_train_vectorized = count.fit_transform(X_train)

print(X_train_vectorized)
```

```python
X_test_vectorized = count.transform(X_test)

print(X_test_vectorized)
```

#### OBSERVATIONS:

1.  The input training and testing data which was in text form has been converted into the matrix of word count.

#### Step 8: Train the Bernoulli Naive Bayes Model

```python
from  sklearn.naive_bayes  import  BernoulliNB

#### create an object for Bernoulli Naive Bayes

bnb = BernoulliNB()

#### using the object of Bernoulli NB, train the model

bnb.fit(X_train_vectorized, Y_train)
```

#### OBSERVATIONS:

1.  The Bernoulli Naive Bayes Model has been trained with the training data.

#### Step 9: Predict the Model

```python
Y_pred = bnb.predict(X_test_vectorized)
```

```python
Y_pred
```

#### Step 10: Perform the Classification Evaluation Metrics

```python
from        sklearn.metrics             import  accuracy_score, confusion_matrix, classification_report

print("===============================Accuracy score=======================================================")

ac = accuracy_score(Y_test, Y_pred)*100.0

print("Accuracy score of the model is:", ac)



print("===============================Confusion matrix=======================================================")

cm = confusion_matrix(Y_test, Y_pred)

print("Confusion matrix of the model is:", cm)



print("===============================Classification report=======================================================")

cr = classification_report(Y_test, Y_pred)

print("Classification report of the model is:", cr)
```

#### OBSERVATIONS:

1.  The accuracy of the Bernoulli Naive Bayes model using spam classification dataset is 96.32

2.  The accuracy of the model is very high. So the model is working fine.

#### Step 11: Predict whether the email is spam or ham based on the sample data

```python
#### define the sample messages

sample_messages = [
    "Congratulations! You've won a free iPhone. Click here to claim.",
    "Hey, are we still meeting for lunch today?"
]

print(sample_messages)


#### perform vectorization on the inputs

vectorized_messages = count.transform(sample_messages)

print(vectorized_messages)


#### perform the prediction on the inputs

prediction = bnb.predict(vectorized_messages)

print(prediction)



print("==========================First Sentence===============================================")

if(prediction[0] == 1):
    print("The email message is Spam")
else:
    print("The email message is Ham")


print("==========================Second Sentence===============================================")

if(prediction[1] == 1):
    print("The email message is Spam")
else:
    print("The email message is Ham")
```

#### Step 12: Perform Hyper Parameter Tuning using GridSearchCV

```python
#### define the pipeline

from  sklearn.pipeline  import  Pipeline

#### create an object for pipeline

pipeline = Pipeline([
    ("count"    ,   CountVectorizer())      ,
    ("bnb"      ,   BernoulliNB())
])
```

```python
#### create the param grid

param_grid = {
     "count__ngram_range" : [(1,1), (1,2)]              ,
     "count__max_df"      : [0.75, 1.0]                 ,
     "count__min_df"      : [1, 2]                      ,
     "count__stop_words"  : [None, 'english']           ,
     "count__binary"      : [True]                      ,
     "bnb__alpha"         : [0.01, 0.1, 1.0]            ,
     "bnb__fit_prior"      : [True, False]
}
```

```python
#### perform the Cross Validation 

from  sklearn.model_selection  import  StratifiedKFold

cv = StratifiedKFold(n_splits = 5, shuffle = True, random_state=42)
```

```python
#### create the object for grid

grid = GridSearchCV(
    pipeline                                            ,
    param_grid       =       param_grid                 ,
    cv               =           cv                     ,
    scoring          =       'accuracy'                 ,
    n_jobs           =         -1
)
```

```python
#### Train the model

grid.fit(X_train, Y_train)
```

```python
#### Obtain the best parameters and the estimator for the model

best_parameter = grid.best_params_

print("Best Parameters for the model is:", best_parameter)


best_estimator = grid.best_estimator_

print("Best Estimator for the model is:", best_estimator)
```

```python
#### predict the model

Y_pred_grid = best_estimator.predict(X_test)
```

```python
Y_pred_grid
```

```python
#### Evaluate the Classification metrics

from        sklearn.metrics             import  accuracy_score, confusion_matrix, classification_report

print("===============================Accuracy score=======================================================")

acgrid = accuracy_score(Y_test, Y_pred_grid)*100.0

print("Accuracy score of the model is:", acgrid)



print("===============================Confusion matrix=======================================================")

cmgrid = confusion_matrix(Y_test, Y_pred_grid)

print("Confusion matrix of the model is:", cmgrid)



print("===============================Classification report=======================================================")

crgrid = classification_report(Y_test, Y_pred_grid)

print("Classification report of the model is:", crgrid)
```

#### OBSERVATIONS:

1.  The accuracy of the Berboulli Naive Bayes model for email spam classification has improved from 6.32 till 98.83.

2.  The accuracy of the model has increased. So it is working fine.
