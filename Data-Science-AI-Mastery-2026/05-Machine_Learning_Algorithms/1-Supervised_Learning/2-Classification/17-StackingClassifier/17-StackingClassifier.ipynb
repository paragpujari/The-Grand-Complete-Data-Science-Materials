#### Stacking Classifier

1.   A Stacking Classifier can be defined as the ensemble machine learning technique that performs the following operations:----

     (a.) The training data is passed to multiple base learners which are in a stack(one above the other).

     (b.) Then the stack model gets trained and predicted to give a predicted output.

     (c.) This predicted output is then passed to the next learner.

     (d.) This goes on and on unless and until we reach the final learner.

     (e.) The final learner which is the Meta Learner provides the ultimate prediction.

#### Step 12: Define all the base learners for Stacking

```python
#### Define all the base learners for stacking purpose

estimators = [
    ('log'      ,      LogisticRegression(max_iter=1000))                                                                                   ,
    ('rcf'      ,      RandomForestClassifier(n_estimators=100,random_state=42))                                                            ,
    ('xgb'      ,      XGBClassifier(n_estimators=100,max_depth=42,learning_rate=0.1,random_state=42,eval_metric='logloss'))
]
```

#### OBSERVATIONS:

1.  Here we have defined all the three models 

    (a.)    Logistic Regression

    (b.)    Random Forest Classifier

    (c.)    XGB Classifier

    and we have defined them for the stacking purpose.

#### Step 13: Define the Meta Learner

```python
#### define the meta model

meta_learner = LogisticRegression(max_iter=1000)
```

#### OBSERVATIONS:

1.  Here we have defined the meta model that provides the ultimate prediction.

2.  It accepts the output predictions of the base learners as the input and learns its predictions.

3.  Here we have used Logistic Regression as the meta learner. This is because :----

        (a.)    It is simple

        (b.)    It is fast

        (c.)    It is easy to use

        (d.)    less prone to overfitting

        (e.)    It provides the highest accuracies amongst all other accuracies of the models.

#### Step 14: Create the Stacking Classifier

```python
#### Create the object for the Stacking Classifier

stacking_classifier = StackingClassifier(
    estimators          =           estimators                  ,
    final_estimator     =           meta_learner                ,
    cv                  =               5
)
```

#### OBSERVATIONS:

1.  Here we have defined the object for the Stacking Classifier that contains the following parameters :------

    (a.)    estimators ==============> defined all the base learners for stacking purpose.

    (b.)    final_estimator =========>  defines the final base learner or the meta learner that makes the ultimate prediction.

    (c.)    cv = 5  ===============> provides unbiased predictions for training the meta learner.

#### Step 15: Train the Stacking Classifier

```python
#### using the object of Stacking Classifier, train the model

stacking_classifier.fit(
    X_train_scaled                                                          ,
    Y_train
)
```

#### OBSERVATIONS:

1.  Using the object of Stacking Classifier, the model is trained using the training data.

2.  The Stacking classifier model is trained along with the three models . They are :----
