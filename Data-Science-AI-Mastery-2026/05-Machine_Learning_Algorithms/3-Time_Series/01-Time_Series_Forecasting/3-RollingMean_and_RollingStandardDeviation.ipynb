#### Rolling Mean and Rolling Standard Deviation

It helps in understanding the

    (a.)    trend

    (b.)    variability

    (c.)    stationarity in the data time series.


#### Objective of the Problem Statement :-------------

1.  The main objective of the problem statement is to

    (a.) find out the rolling mean

    (b.) find out the rolling standard deviation

    (c.) detect the stability in the time series data.


#### Terminologies used :---------

1.  Rolling mean =========>  It helps in finding out the average of the rolling window values.

2.  Rolling std  =========>  It helps in finding out the spread and the fluctutaion of the time series data.

                            (a.)    Low std ============>  stable time series data.

                            (b.)    high std ===========>  highly fluctuating time series data.
                            

#### Step 1: Import all the necessary libraries

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns
```

#### Create the time series data

```python
#### Create the date data

dates = pd.date_range(
    start       =   '2025-01-01'                    ,
    periods     =       30                          ,
    freq        =   'D'
)

print(dates)

#### Create the sales data

sales = np.random.randint(50,100,30)
print(sales)


#### Combine the date data and the sales data in the dataframe

df = pd.DataFrame({
    'Date'      :       dates                       ,
    'Sales'     :       sales
})

#### set the date column as the index
df.set_index('Date',inplace=True)

print(df.head())
```

#### OBSERVATIONS:

1.  The above dataframe depicts the sales data distribution for the month of January starting from 1st January.

#### Step 3: Calculate the Rolling Mean

```python
#### Calculate the rolling mean

df['Rolling_Sales_Mean'] = df['Sales'].rolling(window=3).mean()
```

```python
df.head()
```

#### OBSERVATIONS:

1.  It considers the moving window of the size '3' where it considers the first 3 sales data, computes the mean, then next 3 observations and then finds out the mean, then next 3 sales data, finds out the mean.........and so on.

2.  This process will continue unless and until it reaches the final 3 sales data observations and finds out the mean.

3.  It is done to check the trend and stability in the time series data.

#### Step 4: Calculate the Rolling Standard Deviation

```python
#### Calculate the rolling standard deviation

df['Rolling_Sales_Std'] = df['Sales'].rolling(window=3).std()
```

```python
df.head()
```

```python
df
```

#### OBSERVATIONS:

1.  Here we are considering the moving window of the size 3 where we are considering the first three observations of the sales data and compute the std, then next three observations of the sales data and compute the std...... and son on.

2.  Here it will compute the standard deviation for all the sales data in moving window form.

3.  The main role of rolling std is to measure the spread and the fluctuation of the time series data.

        (a.)    Low the spread ===============>   more stable is the time series data

        (b.)    High the spread ==============>   highly fluctuating is the time series data.

#### Step 5: Plot the Time series data

```python
plt.figure(figsize=(10,6))

#### plot the time series data of the Rolling Mean
plt.plot(df.index, df['Rolling_Sales_Mean'], label = 'Rolling Mean')

#### plot the time series data of the Rolling Standard Deviation
plt.plot(df.index, df['Rolling_Sales_Std'], label = 'Rolling Standard Deviation')

#### Give the title of the graph
plt.title("Sales Data Dustribution for Rolling Mean and Rolling Standard Deviation")

#### Give the x axis of the graph
plt.xlabel('Date')

#### Give the y axis of the graph
plt.ylabel('Sales')

#### Give the legend of the graph
plt.legend()

#### Display the graph
plt.show()
```

#### OBSERVATIONS:

1.  The above graph explains the sales data distribution for the month of January for 30 days starting from 1st January 2025.

2.  Here the above graph shows the sales data distribution in terms of Rolling Mean and Rolling Standard Deviation.

3.  The value of Rolling mean is much higher than the Rolling Standard deviation as shown in the graph.

    (a.)    In the Rolling mean graph, we can see some trends and seasonality in the time series data.

4.  The value of the Rolling Standard deviation is very low

    (a.)    Lower the standard deviation ==============>  More stable is the time series data.
