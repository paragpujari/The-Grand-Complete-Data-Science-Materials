#### Moving Average

Moving Average is the most simplest step in the time series data analysis.

Here we consider the previous observations and compute the average of those observations while performing the time series analysis.

Ex :-----

For a three day moving average, we have values x1, x2, x3, so we take the average of the three values (x1+x2+x3)/3
while performing the time series analysis.

#### Objective of Moving Average:---

1.  The moving average is computed in the time series analysis so as :---------

    (a.)    To reduce the noise.

    (b.)    To smoothen the time series data.

    (c.)    To find the trends and patterns from the data.


#### Step 1: Import all the necessary libraries

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns
```

#### Step 2: Create the Time series data

```python
#### Create the date

dates = pd.date_range(
    start   =   '2025-01-01'            ,
    periods =       30                  ,
    freq    =       'D'
)


#### Create the sales
sales = [10,12,15,14,13,18,20,22,21,19,
         25,27,26,30,32,31,35,36,38,40,
         39,42,45,44,46,48,50,49,52,55]


#### Create the Dataframe from the dates and the sales data

df = pd.DataFrame({
    'Date'      :   dates                   ,
    'Sales'     :   sales
})

#### set the 'Date' as the index column 

df.set_index('Date',inplace=True)


print(df.head())
```

#### OBSERVATIONS:

1.  The Sales date dataframe has been created.

2.  It has two columns i.e date and sales.

3.  Out of date and sales column in the dataframe, date column has been made as an index.

#### Step 3: Compute the Moing Average

```python
df['MA_Sales']  =   df['Sales'].rolling(window=3).mean()
```

```python
df.head()
```

#### OBSERVATIONS:

1.  rolling(window=3).mean() ==============> It considers the moving window of the size 3, then it considers the first three observations, finds out the mean, then the next three obserations, finds out the mean ,....., and so on

    .   In this manner, it find out the mean average of all the observations.

#### Step 4: Plot the Original vs Moving Average

```python
plt.figure(figsize=(10,6))

#### plot the original sales data
plt.plot(df.index, df['Sales'], label = 'Original Sales data')

#### plot the moving average data
plt.plot(df.index, df['MA_Sales'], label = 'Moving Average Sales data')


#### Give the title of the graph
plt.title('Sales data distribution')

#### Give the xaxis of the graph
plt.xlabel('Date')

#### Give the yaxis of the graph
plt.ylabel('Sales')

#### Give the legend of the graph
plt.legend()

#### display the graph
plt.show()
```

#### OBSERVATIONS:

1.  The above graph depicts the sales data distribution for the month of January 2025.

2.  With the increase in the dates, both the original sales data and moving average sales data increases exponentially.

3.  But there are some trends or seasonality in the original sales data for the month of January 2025.
