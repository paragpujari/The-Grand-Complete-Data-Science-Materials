#### Step 1: Import all the necessary libraries

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns
```

#### Step 2: Create a Date Range

```python
#### create a date series

dates = pd.date_range(
    start       =   '2025-01-01'                ,
    periods     =        30                     ,
    freq        =        'D'
)
```

```python
print(dates)
```

#### OBSERVATIONS:

1.  It creates a series of all the date range of 30 dates starting from 1st January 2025 in a daily frequency manner.

2.  This date series will start from 1st January 2025 and will continue till 30 days.

```python
#### Get the type of the date series

print("Type of the date series is:", type(dates))
```

#### OBSERVATIONS:

1.  dates is a pandas date time index object designed for the time series data.

#### Step 3: Convert the time series index object to dataframe

```python
dates_df = pd.DataFrame(dates,columns=['Date'],dtype='datetime64[ns]')
```

```python
dates_df
```

#### OBSERVATIONS:

1.  The dates which is a time series index object is passed as an input into the dataframe.

2.  Then  the dates in time series index object is converted into the dataframe and stored in dates_df.

3.  This dates_df is a dataframe that stores the dates ranging from 1st January 2025 till 30th January 2025.

#### Step 4: Create the sales data

```python
#### Create the sales data

sales = np.random.randint(50,100,30)
```

```python
sales
```

#### OBSERVATIONS:

1.  30 random integers have been generated between 50 and 100 and stored in the sales data.

#### Step 5: Check the data type of dates and sales

```python
print("Type of dates is:", type(dates))
```

```python
print("Type of sales is:", type(sales))
```

#### Step 6: Create the DataFrame that stores the date and sales

```python
df = pd.DataFrame({
    "Date"  :   dates                       ,
    "Sales" :   sales
})
```

```python
df
```

#### OBSERVATIONS:

1.  The above dataframe represents the sales data for the first 30 days of the year 2025.

#### Step 7: Get the information about the columns in the dataset

```python
df.info()
```

#### OBSERVATIONS:

1.  The Date column is in date time format.

2.  The sales column is in integer format.

#### Step 8: Set the Date column as index

```python
df.set_index('Date', inplace=True)
```

```python
#### display the dataframe

df.head()
```

#### OBSERVATIONS:

1.  The date column has been set as an index.

#### Step 9: Plot the Time series sales data

```python
plt.figure(figsize=(10,6))

#### plot the sales data in a daily manner
plt.plot(df.index, df['Sales'])

##### give the title of the graph
plt.title("Daily Sales data")

#### give the x axis of the graph
plt.xlabel("Date")

#### gie the y axis of the graph
plt.ylabel("Sales")

#### display the graph
plt.show()
```

#### OBSERATIONS:

1.  The above graph represents the sales data in a daily manner ranging from 1st January 2025 till 30th January 2025.

2.  There are some trends or seasonality in the aboe graph.

3.  For some of the dates, the sales data is high and for some dates , the sales data is very low.
