#### Linear Discriminant Analysis (LDA)

Linear Discriminant Analysis is a superised machine learning technique that uses the target variable.

The main aim of Linear Discriminant Analysis is to  :--------

    (a.)    Maximize the distances between the clsses.

    (b.)    Minimize the spread within the classses.



The LDA is used when we want to improve the classification.


#### Problem Statement :------

(i.)    The main aim of this project is to apply LDA on the dataset to reduce the number of dimensions  and then convert it into only two features so as to capture the maximum distance between the classes and minimize the variance between the classes.

(ii.)   Then apply Logistic regression on the reduced 2d input feature dataset so as to visualize the results.

#### Step 8: Perform Feature Scaling on the inputs

```python
from sklearn.preprocessing import StandardScaler

#### Create an object for Standard Scaler

sc = StandardScaler()

#### using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)

X_test_scaled  = sc.transform(X_test)
```

#### Step 9: Apply LDA on the input scaled data

Important Rule in LDA

Number of Components ≤ (Number of Classes - 1)

Here:

Classes = 3
Maximum LDA Components = 2

```python
from    sklearn.discriminant_analysis import  LinearDiscriminantAnalysis


#### create an object for LinearDiscriminantAnalysis

lda = LinearDiscriminantAnalysis(n_components = 2)

#### using the object for LinearDiscriminantAnalysis, transform the input data

X_train_scaled_lda = lda.fit_transform(X_train_scaled, Y_train)
```

```python
X_train_scaled_lda
```

```python
X_test_scaled_lda = lda.transform(X_test_scaled)
```

```python
X_test_scaled_lda
```

#### OBSERVATIONS:

1.  By applying LDA on the scaled input, both the training and testing input data has been converted from 13 input features into 2 input features.

2.  This in turn performs the dimensionality reduction.

3.  By doing the dimensionality reduction, the LDA captures the maximum distances between the classes and minimizes the variance in the classes.

```python
#### Shape after reduction

print("Shape of the input training data before applying LDA:", X_train_scaled.shape)

print("Shape of the input training data after applying LDA:", X_train_scaled_lda.shape)
```

#### OBSERVATIONS:

1.  After applying LDA operation on the input scaled data, we see that 

    (a.) the number of input features before applying LDA = 13

    (b.) the number of input features after applying LDA got reduced from 13 to 2.
