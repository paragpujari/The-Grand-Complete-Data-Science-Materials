```python
## Arithmethic Operation

a   =   10
b   =    5

add     = a+b  #addiiton
sub     = a-b  #substraction 
mul     = a*b  #multiplication
div     = a/b  #division
floor   = a//b ## floor division
mod     = a%b  #modulus operation

exp     = a**b ## Exponentiation


print(add)
print(sub)
print(mul)
print(div)
print(floor)
print(mod)
print(exp)
```

```python
#### Comparision Operator

a   =   10
b   =    5

print(a < b)
print(a > b)
print(a <= b)
print(a >= b)
print(a == b)
print(a != b)
```

```python
#### Logical Operator

a   =   10
b   =    5

print((a > b) and (a != b))
print((a <= b) or (a != b))
print((a <= b) and not(a != b))
```

```python
## Simple Calculator

# Simple calculator
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))


add = num1 + num2
sub = num1 - num2
mul = num1 * num2
div = num1 / num2
floor = num1 // num2
exp = num1 ** num2
mod = num1  % num2


print("Addition value is:", add)
print("Substraction value is:", sub)
print("Muliplication value is:", mul)
print("Division value is:", div)
print("Floor value is:", floor)
print("Exponential value is:", exp)
print("Modulus value is:", mod)
```
