#### 1. Print Numbers from 1 to 10

```python
for i in range(1,11):
    if((i%2)==0):
        print(i)
```

#### 3. Sum of First N Natural Numbers

```python
sum = 0
n = int(input("Enter the numbers:"))

for x in range(1,(n+1)):
    sum = sum + x

print("Sum of N natural numbers is:", sum)
```

#### 4. Factorial of a Number

```python
fact = 1
n = int(input("Enter the numbers:"))

for x in range(1, (n+1)):
    fact = fact * x

print("Factorial of the number is:", fact)
```

#### 5. Multiplication Table

```python
n = int(input("Enter the number:"))

for x in range(1,11):
    print(n , "*", x, "=", (n * x))
```

### 6. Reverse a String Using for Loop

```python
name = input("Enter your name:")

print(name)

rev = " "

for x in name:
    rev = x + rev

print(rev)
```

### 7. Count Vowels in a String

```python
text = input("Enter your text:")

#### convert your text in lower case

text = text.lower()
print(text)

nv = 0
for x in text:
    if(x in "aeiou"):
        nv = nv + 1

print("No of vowels in the text is:", nv)
```

### 8. Find Largest Element in a List

```python
numbers = [10, 45, 23, 67, 89, 34]

### initialize the first element to the largest

largest = numbers[0]

for x in  numbers:
    if(x > largest):
        largest = x

print("Largest element in the array is:", largest)
```

#### 9. Print Pattern (Star Triangle)

```python
n = 5
for i in range(1,(n+1)):
    for j in range(1,i):
        print("*", end = " ")
    print("\n")
```

### 10. Check Prime Number

```python
num = int(input("Enter the number"))

is_prime = True

for i in range(2,(num+1)):
    if((num%i)==0):
        is_prime = False
    break

if(is_prime):
    print("Prime Number is:", num)
else:
    print("Not a Prime Number is:", num)
```

### 11. Find Duplicates in a List

```python
numbers = [1,1,1,2,2,3,3,3,4,4,5,6,7,7,7,8,8,9,10]

freq = {}

for x in numbers:
    if(x in freq):
        freq[x] = freq[x] + 1
    else:
        freq[x] = 1

print(freq)

for x,y in freq.items():
    if(y > 1):
        print(x)
```

### 12. Check Palindrome Number

```python
num = int(input("Enter the number:"))

rev = 0

x = num

while(x != 0):
    d   = x % 10
    rev = (rev * 10) + d
    x   =  x // 10

if(rev == num):
    print(num,"is Palindrome")
else:
    print(num, "is not Palindrome")
```

### 13. Armstrong Number

```python
num = int(input("Enter the number:"))

sum = 0

x = num

while(x != 0):
    d   = x % 10
    sum = sum + (d ** 3)
    x   = x // 10

if(sum == num):
    print(num,"is armstrong")
else:
    print(num,"is not an armstrong")
```
