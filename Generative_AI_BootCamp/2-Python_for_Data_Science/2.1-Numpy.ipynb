### Core NumPy Concepts

#### 1. NumPy Arrays

A NumPy array is a homogeneous multidimensional array.

```python
import numpy  as  np
```

### Creating Arrays

####  1D Array

```python
x = np.array([10, 20, 30, 40, 50])

print(x)

print("No of dimensions used in the array is:", x.ndim)

print("Type of the array is:", x.dtype)
```

#### 2D Array

```python
x = np.array([
    [1,2,3],
    [4,5,6]
])

print(x)

print("No of dimensions used in the array is:", x.ndim)

print("Type of the array is:", x.dtype)
```

```python
### Array Attribute

print("==================Original array ========================================================")
print(x)

print("==================No of dimensions used in the array=====================================")
print(x.ndim)

print("========================Data Type of every element in the array===========================")
print(x.dtype)

print("====================Shape of the array====================================================")
print(x.shape)

print("======================Size of the array====================================================")
print(x.size)
```

### 3. Special Arrays

##### Zeros -------------> It fills the array with zeros with sepcific sets of rows and columns.

```python
a = np.zeros((3,3), dtype='int32')

print(a)
```

### OBSERVATIONS:

1. Here we are creating the numpy array and filling zeros in the array with 3 rows and 3 columns. Each element in the numpy array is of int type.

##### Ones -------------> It fills the array with ones with sepcific sets of rows and columns.

```python
a = np.ones((3,3), dtype='int32')

print(a)
```

### OBSERVATIONS:

1. The numpy array is created of the size 3 * 3. Each element is filled with 1. It is of integer type.

### Identity Matrix -----> Ones at the diagonal position and the rest is zero.

```python
a = np.eye(3, dtype='int32')

print(a)
```

### OBSERVATIONS:

1.  Here a identity matrix is created with 1 at the diagonal position and the rest position is filled with zeros.

### 8. Reshaping

```python
a = np.arange(1,10)

print("=========================Original array ===============================================")

print(a)

### Convert the numpy array into 3 * 3

print("=========================Reshaped array ===============================================")

a1 = a.reshape(3,3)

print(a1)
```

### 9. Broadcasting

Here the numpy array automatically matches the dimensions.

```python
a = np.arange(1,10)

print("=========================Original array ===============================================")

print(a)

print("==========================Broadcasted array ===========================================")

a = a + 10

print(a)
```

### OBSERVATIONS:

1. 10 is automatically matched with all the dimensions of the array and added to every element of the array.

### NumPy Coding Practice

Important coding problems.

```python
### Problem 2

#### Find row mean.

print("=========================Original array ===============================================")
print(a)

print("=========================Reshaped array ===============================================")
x = a.reshape(5,5)
print(x)

### Find the row mean

print("===========================Row mean ===================================================")

ans = np.mean(x, axis=1)

print(ans)
```

```python
### Problem 3

#### Find column sum.

print("=========================Original array ===============================================")
print(a)

print("=========================Reshaped array ===============================================")
x = a.reshape(5,5)
print(x)

print("=========================Column sum ====================================================")
ans = np.sum(x,axis=0)
print(ans)
```

1. When axis = 1, then the operation is performed along the rows.

### Real World Project

Student Performance Analysis

### Goal

1. Perform all the numpy calculations and based on it get the performance of the student.


### Steps used in this Algorithm :---


1.  Import Libraries

2.  Create Dataset

3.  Calculate Average Score along the rows

4.  Find Top Student

5.  Calculate Subject Average along the columns

6.  Standard Deviation

7.  Rank Students

### Step 1: Import Libraries

```python
import numpy as np
```

### Step 2:  Create Dataset

```python
scores = np.array([
[85,78,92],
[70,65,60],
[90,88,95],
[60,55,58],
[75,80,72]
])
```

```python
scores
```

### Step 3: Calculate Average Score along the rows

```python
average = np.mean(scores,axis=1)
```

```python
average
```

### Step 4: Find Top Student

```python
### We need to get the student that has the maximum average score

### Find out the average scores of all the students along the rows

x = np.mean(scores,axis=1)
print(x)

### Find out the index having the maximum average score

ans = np.argmax(x)
print(ans)
```

### OBSERVATIONS:

1. Index 2 (91) is the index having the maximum average score.

### Step 5: Calculate Subject Average along the columns

```python
avg_scores_cols = np.mean(scores,axis=0)

print(avg_scores_cols)
```

### Step 6: Standard Deviation

```python
std = np.std(scores)

print("Standard deviation of the scores is:", std)
```

### Step 7: Rank Students

```python
### We need to rank the performance of the students according to the average score in each row

avg_scores_mean = np.mean(scores,axis=1)

print(avg_scores_mean)

### provide the rank by assigning the indices to every element in the array in ascending order

averagerank = np.argsort(avg_scores_mean)

print(averagerank)
```

```python
### provide the rank by assigning the indices to every element in the array in descending order

averagerankdesc = np.argsort(avg_scores_mean)[::-1]

print(averagerankdesc)
```

### 3. Find Duplicate Values

```python
import numpy as np

arr = np.array([1,2,3,2,4,5,1])

unique, counts = np.unique(arr, return_counts=True)

print(unique,counts)

### Get all the duplicate elements from the array
duplicates = unique[counts > 1]

print(duplicates)
```

### 7. Normalize Dataset (Min-Max Scaling)

```python
import numpy as np

data = np.array([10,20,30,40,50])

normalized_data = (data - np.min(data))/(np.max(data) - np.min(data))

print(normalized_data)
```

### 8. Standardize Data (Z-score)

```python
import numpy as np

data = np.array([10,20,30,40,50])


standarddata = (data - np.mean(data))/(np.std(data))

print(standarddata)
```
