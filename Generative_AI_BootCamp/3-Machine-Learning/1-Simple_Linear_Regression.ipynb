###  Write a code based on the Simple Linear Regression that predicts the score of the student based on the no of hours studied.


#### Step 1: Import all the necessary libraries

```python
import numpy             as np
import pandas            as pd
import matplotlib.pyplot as plt
import seaborn           as sns

from   sklearn.model_selection import train_test_split
from   sklearn.linear_model    import LinearRegression
from   sklearn.metrics         import accuracy_score, confusion_matrix, classification_report
```

#### Step 6:    Divide the dataset into independent and dependent features

```python
#### Independent features

X = df[['Hours']]

print(X)
```

```python
#### Dependent features

Y = df['Scores']

print(Y)
```

#### OBSERVATIONS:

1.  The above dataset is divided into the independent and dependent features.

#### Step 7: Divide the independent and dependent features into the training and testing data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
X_train
```

```python
X_test
```

```python
print("Shape of the input training data is:", X_train.shape)

print("Shape of the input testing data is:", X_test.shape)
```

```python
Y_train
```

```python
Y_test
```

```python
print("Shape of the output training data is:", Y_train.shape)

print("Shape of the output testing data is:", Y_test.shape)
```

#### Step 8:    Train the linear model

```python
from sklearn.linear_model import LinearRegression

#### create an object for Linear Regression

model = LinearRegression()

#### using the object of linear regression, train the model
model.fit(X_train, Y_train)
```

#### OBSERVATIONS:

1.  The linear regression model has been trained with the help of the training data.

#### Step 9: Predict the Model

```python
Y_pred = model.predict(X_test)
```

```python
Y_pred
```

#### OBSERVATIONS:

1.  The output has been predicted with the help of the input test data.

#### Step 10: Get the model parameters

```python
print("Slope     of the model is:", model.coef_[0])
print("Intercept of the model is:", model.intercept_)
```

#### Step 11: Compare the Actual vs Predicted Results and get the residual and store it in a dataframe

```python
Y_test.shape
```

```python
Y_pred.shape
```

```python
results = pd.DataFrame({
    'Actual Results'     : Y_test           ,
    'Predicted Results'  : Y_pred           ,
    'Residual'           : (Y_test - Y_pred)
})
```

```python
results
```

#### OBSERVATIONS:

1. The above dataframe contains the difference between the actual and the predicted output.

#### Step 12: Get the Regression Metrics of the Model

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error
```

```python
mae = mean_absolute_error(Y_test, Y_pred)

print("Mean Absolute Error of the model is:", mae)
```

```python
mse = mean_squared_error(Y_test, Y_pred)

print("Mean Squared Error of the model is:", mse)
```

```python
r2 = r2_score(Y_test, Y_pred)

print("R2 Score of the model is:", r2)
```

```python
rmse = root_mean_squared_error(Y_test, Y_pred)

print("Root Mean Squared Error of the model is:", rmse)
```

#### OBSERVATIONS:

1.  The model is completely fine as the r2 score of the model is 0.97 which is very high.

#### Step 13: Get the best line of the fit for the linear regression model

```python
### perform the scatter plot for the Original data points of the training data
plt.scatter(X_train, Y_train, color = 'red', label = 'Actual Data Points')

### plot the best line of fit
plt.plot(X_train, model.predict(X_train), color = 'green', label = 'Predicted Data Points')

### Get the title of the graph
plt.title("Simple Linear Regression between Hours vs Score of the training data")

### Get the x label of the graph
plt.xlabel("Hours Studied")

### Get the y label of the graph
plt.ylabel("Score Obtained")


#### display the graph

plt.show()
```
