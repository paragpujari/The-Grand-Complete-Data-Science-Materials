#### Step 5:-- Divide the dataset into independent and dependent variables

```python
#### Independent variables

X = df[['area', 'bedrooms', 'bathrooms', 'stories']]

#### Dependent variables

Y = df['price']
```

```python
print(X)
```

```python
print(Y)
```

#### Step 6:-- Divide the independent and dependent variables into the training and testing data

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

```python
print("Shape of the output training data is:", Y_train.shape)

print("Shape of the output testing  data is:", Y_test.shape)
```

#### Step 7:-- Perform the Feature Scaling

```python
from sklearn.preprocessing import StandardScaler

#### create an object for Standard Scaler

sc = StandardScaler()

#### Using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)

X_test_scaled = sc.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

#### OBSERVATIONS:

1.  After performing the feature scaling on the inputs , all the inputs has been scaled in the same range of 0 to 1.

#### Step 8:--- Train the Linear model

```python
from sklearn.linear_model import LinearRegression


#### create an object for Linear Regression

lin = LinearRegression()


#### using the object for linear regression, train the model


lin.fit(X_train_scaled, Y_train)
```

#### OBSERVATIONS:

1.  The linear regression model has been trained with the help of the training data.

#### Step 9 :-- Predict the model

```python
Y_pred = lin.predict(X_test_scaled)
```

```python
Y_pred
```

#### OBSERVATIONS:

1.  Based on the input test data, we have obtained the predicted value of the House Price.

#### Step 10:-- Compare the Actual vs Predicted prices and store the residual in the dataframe

```python
res = pd.DataFrame({
    'Actual Results'     :    Y_test          ,
    'Predicted Results'  :    Y_pred          ,
    'Residual'           :    (Y_test - Y_pred) 
})
```

```python
res
```

#### Step 11:-- Perform the Regression Metrics

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error

mae = mean_absolute_error(Y_test, Y_pred)

print("Mean Absolute error of the model is:", mae)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error

mse = mean_squared_error(Y_test, Y_pred)

print("Mean squared error of the model is:", mse)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error

rmse = root_mean_squared_error(Y_test, Y_pred)

print("Root Mean squared error of the model is:", rmse)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score, root_mean_squared_error

r2 = r2_score(Y_test, Y_pred)

print("R2 Score of the model is:", r2)
```

#### OBSERVATIONS:

1.  As the R2 score of the model is 0.51, so the model is ok.
