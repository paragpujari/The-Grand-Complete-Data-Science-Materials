##### Step 1:   Import  all the  necessary  libraries

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

from    sklearn.model_selection    import  train_test_split
from    sklearn.preprocessing      import  StandardScaler, MinMaxScaler, Normalizer
from    sklearn.linear_model       import  LogisticRegression
from    sklearn.metrics            import  accuracy_score, confusion_matrix, classification_report
```

#### Step 2: Load the Dataset

```python
#### define the url

url = "https://raw.githubusercontent.com/jbrownlee/Datasets/master/pima-indians-diabetes.data.csv"

#### define all the column names


columns = ["Pregnancies", "Glucose", "BloodPressure", "SkinThickness",
           "Insulin", "BMI", "DPF", "Age", "Outcome"]


##### define the dataset from the url and the columns

df = pd.read_csv(url, names = columns)

print(df)
```

#### OBSERVATIONS:

1.  The above dataset explains about the diabetic situation for each and every patient.

#### Step 3: Perform the Data Exploration

```python
#### get the first five rows of the dataset

df.head()
```

##### Step 4: Perform the Data Cleaning

```python
#### Check whether there are any NULL Values in the dataset

df.isnull().sum()
```

#### OBSERVATIONS:

1.  There are no NULL records in the dataset.

```python
#### To check for the duplicate records in the dataset

df[df.duplicated()]
```

#### OBSERVATIONS:

1.  There are no duplicate records in the dataset.

```python
#### Replace all the zeros value in the column with the median value

col = ["Pregnancies", "Glucose" , "BloodPressure" , "SkinThickness" , "Insulin", "BMI", "DPF", "Age"]

for x in  col:
    #### Replace all the zeros with the median value of it
    df[x] = df[x].replace(0, df[x].median())


df
```

#### OBSERVATIONS:

1.  All the zeros in every column has been replaced with the median value of that column.

#### Step 5: Divide the dataset into independent and dependent variables

#### Step 6: Divide the independent and dependent variables into the training and testing data

```python
from  sklearn.model_selection  import  train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

#### Step 7: Perform without Feature Scaling

```python
from  sklearn.linear_model  import LogisticRegression

#### Create an object for Logistic Regression

log = LogisticRegression(max_iter=1000)


#### using the object of Logistic Regression, train the model

log.fit(X_train, Y_train)

#### predict the model

Y_pred = log.predict(X_test)

print(Y_pred)

#### Perform the Classification metrics on the inputs

from  sklearn.metrics  import  accuracy_score, confusion_matrix, classification_report

ac = accuracy_score(Y_test, Y_pred)*100.0

print("Accuracy score of the model is:", ac)


cm = confusion_matrix(Y_test, Y_pred)


print("Confusion Matrix of the model is:", cm)

cr = classification_report(Y_test, Y_pred)

print(cr)
```

#### OBSERVATIONS:

1.  The overall accuracy of the model without feature scaling is 70.77 which is fine.

#### Step 8: Perform the Feature scaling on the inputs

#### (a.)   StandardScaler

```python
from  sklearn.linear_model  import LogisticRegression

#### Create an object for Standard Scaler

sc = StandardScaler()

X_train_scaledsc = sc.fit_transform(X_train)

X_test_scaledsc  = sc.transform(X_test)

#### Create an object for Logistic Regression

logsc = LogisticRegression(max_iter=1000)


#### using the object of Logistic Regression, train the model

logsc.fit(X_train_scaledsc, Y_train)

#### predict the model

Y_pred_sc = logsc.predict(X_test_scaledsc)

print(Y_pred_sc)

#### Perform the Classification metrics on the inputs

from  sklearn.metrics  import  accuracy_score, confusion_matrix, classification_report

ac_sc = accuracy_score(Y_test, Y_pred_sc)*100.0

print("Accuracy score of the model is:", ac_sc)


cm_sc = confusion_matrix(Y_test, Y_pred_sc)


print("Confusion Matrix of the model is:", cm_sc)

cr_sc = classification_report(Y_test, Y_pred_sc)

print(cr_sc)
```

#### (b.)  MinMaxScaler

```python
from  sklearn.linear_model  import LogisticRegression

#### Create an object for Standard Scaler

minmax = MinMaxScaler()

X_train_scaledminmax = minmax.fit_transform(X_train)

X_test_scaledminmax  = minmax.transform(X_test)

#### Create an object for Logistic Regression

logminmax = LogisticRegression(max_iter=1000)


#### using the object of Logistic Regression, train the model

logminmax.fit(X_train_scaledminmax, Y_train)

#### predict the model

Y_pred_minmax = logminmax.predict(X_test_scaledminmax)

print(Y_pred_minmax)

#### Perform the Classification metrics on the inputs

from  sklearn.metrics  import  accuracy_score, confusion_matrix, classification_report

ac_minmax = accuracy_score(Y_test, Y_pred_minmax)*100.0

print("Accuracy score of the model is:", ac_minmax)


cm_minmax = confusion_matrix(Y_test, Y_pred_minmax)


print("Confusion Matrix of the model is:", cm_minmax)

cr_minmax = classification_report(Y_test, Y_pred_minmax)

print(cr_minmax)
```

#### (c.) Normalizer

```python
from  sklearn.preprocessing  import StandardScaler, MinMaxScaler, Normalizer

#### Perform the Standard Scaler technique

norm = Normalizer()

X_train_norm = norm.fit_transform(X_train)

X_test_norm  = norm.transform(X_test)


#### Train the model

from  sklearn.linear_model  import LogisticRegression

#### Create an object for Logistic Regression

log_norm = LogisticRegression(max_iter=1000)

#### using the object of Logistic Regression, train the model

log_norm.fit(X_train_norm, Y_train)


#### using the object of logistic regression, predict the model

Y_pred_norm = log_norm.predict(X_test_norm)

print(Y_pred_norm)

##### Evaluate the Classification Metrics

from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

ac_norm = accuracy_score(Y_test, Y_pred_norm)*100.0

print("Accuracy score of the model is", ac_norm)


cm_norm = confusion_matrix(Y_test, Y_pred_norm)

print("Confusion Matrix of the model is", cm_norm)


cr_norm = classification_report(Y_test, Y_pred_norm)

print("Classification Report of the model is", cr_norm)
```
