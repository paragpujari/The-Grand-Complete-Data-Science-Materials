#### OBSERVATIONS:

1. The above dataset has more number of 0s than the 1s . So it is an imbalanced dataset.

```python
#### Perform the resampling on the imbalanced dataset to make it balanced

from  sklearn.utils  import resample

df_majority = df[df['Purchased'] == 0]

df_minority = df[df['Purchased'] == 1]
```

```python
#### perform the upsamling on the minority class

df_minority_upsample = resample(
    df_minority                                  ,
    replace= True                                ,
    n_samples= len(df_majority)                  ,
    random_state=42
)
```

```python
df_minority_upsample
```

```python
df_new = pd.concat([df_majority, df_minority_upsample])
```

```python
df_new
```

```python
df = df_new
```

```python
df
```

```python
#### Again get the count of all the categories in 'Purchased' Column

df['Purchased'].value_counts()
```

#### OBSERVATIONS:

1. After perfroming the upsampling on the minority class, the category 1 has been increased to 257. The total count of the category 0 = count of the category 1. So the dataset is a balanced.

#### Step 4: Perform the Feature Encoding


##### (a.)  Label Encoder

```python
#### perform the label encoding on the categorical columns

from sklearn.preprocessing import LabelEncoder

##### create an object for Label Encoder

lab = LabelEncoder()

df['Gender'] = lab.fit_transform(df['Gender'])
```

##### (a.)  One Hot Encoder

```python
df_one = pd.get_dummies(df, columns= ['Gender'], drop_first=True)
```

```python
df_one
```

```python
#### perform map operation on it to map the categorical values into the numerical values
```

```python

df_one['Gender_1'] = df_one['Gender_1'].map({True:1, False :0})
```

```python
df_one = df_one.rename(columns={'Gender_1':'Gender'})
```

```python
df = df_one
```

```python
df
```

```python
#### Drop the unnecessary columns from the dataset

df.drop(columns = 'User ID', axis=1, inplace=True)
```

```python
df
```

#### Step 5: Divide the dataset into independent and dependent features

```python
X = df.drop(columns = 'Purchased', axis=1)
```

```python
Y = df['Purchased']
```

```python
print(X)
```

```python
print(Y)
```

#### Step 6: Divide the independent and dependent features into the training and testing data

```python
from sklearn.model_selection  import  train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

```python
X_train
```

```python
X_test
```

```python
print("Shape of the input training data is:", X_train.shape)

print("Shape of the input testing  data is:", X_test.shape)
```

```python
Y_train
```

```python
Y_test
```

```python
print("Shape of the output training data is:", Y_train.shape)

print("Shape of the output testing  data is:", Y_test.shape)
```

#### Step 7: Perform the Feature Scaling on the inputs

```python
from  sklearn.preprocessing   import  StandardScaler

#### create an object for Standard Scaler

sc = StandardScaler()

#### using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)

X_test_scaled  = sc.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

#### Step 8: Train the Logistic Regression Model

```python
from  sklearn.linear_model  import LogisticRegression

#### create an object for Logistic Regression

log = LogisticRegression(max_iter=1000)

#### using the object of Logistic Regression, train the model

log.fit(X_train_scaled, Y_train)
```

#### Step 9: Predict the Model

```python
Y_pred_log = log.predict(X_test_scaled)
```

```python
Y_pred_log
```

#### Step 10: Evalaute the Classification Metrics

```python
from  sklearn.metrics  import  accuracy_score, confusion_matrix, classification_report

#### Accuracy of the model

ac = accuracy_score(Y_test, Y_pred_log)*100.0

print("Accuracy score of the model is:", ac)
```

```python
#### Confusion Matrix of the model

cm = confusion_matrix(Y_test, Y_pred_log)

print("Confusion  Matrix of the model is:", cm)
```

```python
#### Classification Report of the model

cr = classification_report(Y_test, Y_pred_log)

print("Classification  Report of the model is:", cr)
```

#### Step 11: Save and dump the model

```python
import joblib

joblib.dump(sc,"Standard_encoding.pickle")
```
