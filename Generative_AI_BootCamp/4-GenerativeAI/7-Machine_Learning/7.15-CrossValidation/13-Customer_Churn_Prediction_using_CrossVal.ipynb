#### Step 6:  Divide the independent and dependent variables into the training and testing data

```python
from  sklearn.model_selection  import train_test_split, KFold, cross_val_score

#### Divide the dataset into the training and testing data

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

#### Step 7: Perform Feature Scaling on the inputs

```python
from sklearn.preprocessing  import  StandardScaler

#### Craete an object for Standard Scaler

sc = StandardScaler()

#### Using the object for Standard Scaler, transform the inputs

X_train_scaled = sc.fit_transform(X_train)

X_test_scaled  = sc.transform(X_test)
```

```python
X_train_scaled
```

```python
X_test_scaled
```

#### OBSERVATIONS:

1.  All the inputs have been scaled in one range.

#### Step 8: Train the LogisticRegressionModel and also apply cross validation on it

```python
from  sklearn.linear_model  import LogisticRegression

#### create an object for Logistic Regression

log = LogisticRegression(max_iter=1000)

#### Apply cross validation on the logistic regression model to improve its accuracy

### use kfold to divide the dataset

kfold = KFold(n_splits=5, shuffle=True, random_state=42)

#### then compute the cross validation score

cvscore = cross_val_score(log, X_train_scaled, Y_train, cv = kfold, scoring= 'accuracy')


print("Cross Validation Score of the model is:", cvscore)

print("Mean Accuracy is:", np.mean(cvscore))
```

```python
#### Train the Model

log.fit(X_train_scaled, Y_train)
```

#### OBSERVATIONS:

1.  The dataset is divided into the K number of folds.

2.  Each fold has a training and validation accuracy.

3.  Each fold computes a accuracy .

4.  Then the mean or the average of all the accuracies is computed

5.  The cross validation is performed so as to:-----

    (a.)  To reduce the overfitting.

    (b.)  To improve the performance of the model.

    (c.)  To efficiently use the  datasets.


6.  The Logistic Regression model is trained with the help of the training data

#### Step 9: Predict the Model

```python
Y_pred = log.predict(X_test_scaled)
```

```python
Y_pred
```

#### Step 10: Evaluate the Classification Metrics

```python
from  sklearn.metrics  import accuracy_score, confusion_matrix, classification_report


print('==================Accuracy Score =====================================')

ac = accuracy_score(Y_test, Y_pred)

print("Accuracy score of the model is:", (ac * 100.0))
```

```python
print('==================Confusion Matrix=====================================')

cm = confusion_matrix(Y_test, Y_pred)

print("Confusion Matrix of the model is:", (cm))
```

```python
print("===================Classification Report================================")

cr = classification_report(Y_test, Y_pred)

print("Classification Report of the Model is:", cr)
```
