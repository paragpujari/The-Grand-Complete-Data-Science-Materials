#### Typecasting

It refers to the process of changing the type of the data.

```python
var = 2.12

#### Convert the float type to integer

print("====================Variable before Type Casting =====================================")

print("Data Variable before typecasting is:", var)

print("====================Variable after Type Casting =====================================")

x = int(var)

print("Data Variable after typecasting is:", x)
```

```python
var = 2

#### Convert the integer type to float

print("====================Variable before Type Casting =====================================")

print("Data Variable before typecasting is:", var)

print("Data Type before typecasting is:", type(var))

print("====================Variable after Type Casting =====================================")

x = str(var)

print("Data Variable after typecasting is:", x)
print("Data Variable after typecasting is:", type(x))
```
