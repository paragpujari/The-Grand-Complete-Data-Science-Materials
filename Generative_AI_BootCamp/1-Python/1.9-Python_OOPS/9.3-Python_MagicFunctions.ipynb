### MAGIC FUNCTIONS

It refers to the list of methods associated with the entity. It starts with __ end ends with __.

```python
# list of all the methods associated with int

print(dir(int))
```

```python
a = 10

a.__add__(10)
```

```python
a.__sub__(10)
```

```python
a.__mul__(10)
```

```python
# list of all the methods associated with the string

print(dir(str))
```

```python
## Write a program to calculate the area and perimeter of the rectangle

class rectangle:
    def __init__(self):
        self.length = 120
        self.breadth = 80

    def print_details(self):
        print("Length of the rectangle is:", self.length)
        print("Breadth of the rectangle is:", self.breadth)

    def print_area_rec(self):
        return(self.length * self.breadth)

    def print_peri_rec(self):
        return(2*(self.length + self.breadth))

obj = rectangle()
obj.print_details()
area = obj.print_area_rec()
peri = obj.print_peri_rec()

print("Area of the rectangle is:", area)

print("Perimeter of the rectangle is:", peri)
```

```python
class person:
    def __init__(self,n,a,s):
        self.name = n
        self.age  = a
        self.salary = s

    def __str__(self):
        return "This is my new object that will return something"

    def print_details(self):
        print(self.name)
        print(self.age)
        print(self.salary)


p = person('parag',30,52345)
p.print_details()

print(p)
```
