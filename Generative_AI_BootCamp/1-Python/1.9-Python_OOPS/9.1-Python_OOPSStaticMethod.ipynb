### Static Method:

Is is the method that is directly called by the class.

```python
class Employee:
    def enter_employee_details(self,n,a,s):
        self.name = n
        self.age  = a
        self.salary = s

    def print_employee_details(self):
        print("Name is:", self.name)
        print("Age  is:", self.age)
        print("Salary is:", self.salary)

    @staticmethod
    def print_mentor(mentor):
        print(mentor)

    @staticmethod
    def print_student_list(students):
        print(students)

### create the object for the class
obj = Employee()
obj.enter_employee_details("Parag",30,52656)
obj.print_employee_details()
### call the statisc method using the class
Employee.print_mentor("Akhilesh")
Employee.print_student_list("Parag,Nishant, Subham")
```
