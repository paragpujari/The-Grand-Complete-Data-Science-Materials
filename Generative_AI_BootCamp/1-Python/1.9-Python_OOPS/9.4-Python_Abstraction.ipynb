### Abstraction

It is a process of representing all the extrenal features without representing the internal details.

```python
### import all the abstract functions

from  abc  import ABC, abstractmethod


### define the abstract class

class Shape(ABC):
    ### Define the abstrcat method
    def print_area(self):
        pass

### Define the non abtract class
class Rectangle(Shape):
    def enter_rec_details(self,l,b):
        self.length = l
        self.breadth = b

    def print_rec_details(self):
        print("Length  of the rectangle is:", self.length)
        print("Breadth of the rectangle is:", self.breadth)

    def calc_rec_details(self):
        print("Area      of the rectangle is:", (self.length * self.breadth))
        print("Perimeter of the rectangle is:", (2*(self.length + self.breadth)))

class Square(Shape):
    def enter_square_details(self,s):
        self.side = s

    def print_square_details(self):
        print("Side  of the square is:", self.side)

    def calc_square_details(self):
        print("Area      of the square is:", (self.side * self.side))
        print("Perimeter of the square is:", (4*(self.side)))


### Create the object for the class
rec = Rectangle()
rec.enter_rec_details(120,80)
rec.print_rec_details()
rec.calc_rec_details()

print("====================================================================================")

squ = Square()
squ.enter_square_details(60)
squ.print_square_details()
squ.calc_square_details()
```

### OBSERVATIONS:

1. The abstract class and the abstrcat method is built that does not have any implementation or the object.

2. Then the non abstract class and the methods are made from the abstract class and the methods.
