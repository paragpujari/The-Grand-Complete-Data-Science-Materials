### OOPS

OOPS stands for Object Oriented Programming Language. Python is an object oriented programming language that uses objects and classes to write the code for better organization and reusuability.

### Advantages of OOPS:

(i) It provides better structure for code.

(ii) It provides better reusuability.

(iii) It is easier to maintain.

### Classes:

Class can be defined as the group of objects that do not have its own independnt existence.

It shares its same set of properties with the objects.

It can be defined as the blue print of objects.

```python
### Create a class
class Person:
    pass
```

### Objects

It is an independent entity that has its own independent existance.

```python
### Create a class
class Person:
    pass

### Create an object for the class
obj = Person()
```

### OBSERVATIONS:

1. The class 'Person' is created.

2. Then the object obj is created that is refering to the class 'Person'.

```python
## using the concepts of objects and class print the message "Hello World"

class Message:
    ### create the function to print the message "Hello World"
    def print_message(self):
        print('Hello World')

### create the object for the class
obj = Message()
obj.print_message()
```

### OBSERVATIONS:

1. The class 'Message' is created.

2. The object 'obj' is used to refer to the class 'Message'.

3. The object obj has a data self thet helps the class to refer to the object.

### 1. Perform the addition of two numbers?

```python
class Addition:
    def sum(self,x,y):
        self.a = x
        self.b = y
        print("Addition of two numbers is:", (self.a + self.b))

obj = Addition()
obj.sum(10,20)
```

### 2. Perform the multiplication of two numbers?

```python
class Multiply:
    def enter_two_numbers(self,x,y):
        self.x1 = x
        self.x2 = y

    def multiply(self):
        print("Multiplication of two numbers is:", (self.x1 * self.x2))

a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))
obj = Multiply()
obj.enter_two_numbers(a,b)
obj.multiply()
```

### 3. Calculate the increased salary of the person?

```python
class Person:
    def enter_details(self,x,y,z):
        self.name = x
        self.company = y
        self.salary = z

    def calculate_bonus(self,x):
        res = self.salary + (self.salary * (x/100))
        return(res)

obj = Person()
obj.enter_details("Parag","Motherson",52897)
ans = obj.calculate_bonus(10)
print("Increased salary of the person is:", ans)
```

### 4. Calculate the balance of the person?

```python
class Balance:
    def enter_balance(self,x):
        self.balance = x

    def print_balance(self):
        print("Balance Amount is:", self.balance)

    def deposit_money(self,x):
        self.balance = self.balance + x
        print("Deposited money is:", (self.balance))

    def withdraw_money(self,y):
        self.balance = self.balance - y
        print("Withdraw money is:", (self.balance))


obj = Balance()
obj.enter_balance(1000)
obj.print_balance()
obj.deposit_money(500)
obj.withdraw_money(300)
```
