### Python Iterators

An iterator is an object that contains countable number of values.

It can be iterated upon meaning we can traverse through all the values.


### Two methods used in Iterators:

(i)   iter() ---> It is used to generate an iterator.

(ii)  next() ---> It is used to generate the next element .

```python
# Return an iterator from a tuple, and print each value from the iterator object

fruits = ("apple","banana","cherry","guava","lichi")


x = iter(fruits)

print(x)
```

### OBSERVATIONS:

1. Using the iter function, we have obtained the iterator object.

```python
### print each value from the iterator object

print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```

### OBSERVATIONS:

1.  It helps in obtaining the value from the iterator object.

```python
## Lists are iterable and we can iterate through the lists using iterator

lists = [10,20,30,40,50]


### create an iterator object from the lists

x = iter(lists)

x
```

```python
print(next(x))
print(next(x))
print(next(x))
print(next(x))
print(next(x))
```

```python
# Strings are also iterable objects, containing a sequence of characters

x = "banana"

for i in x:
    print(i)
```

```python
### Create an iterator object from the string

res = iter(x)

res
```

```python
### extract the values from the iterator object

print(next(res))
print(next(res))
print(next(res))
print(next(res))
print(next(res))
print(next(res))
```
