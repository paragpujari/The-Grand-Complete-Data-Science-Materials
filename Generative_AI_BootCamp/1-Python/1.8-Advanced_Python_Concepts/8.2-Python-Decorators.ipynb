### Decorators

1.  Decorators let you add extra behavior to a function, without changing the function's code.

2.  A decorator is a function that takes another function as input and returns a new function

3.  Define the decorator first, then apply it with @decorator_name above the function

```python
### use the decorator to convert the text to upper case

def mydecorator(mytextdata):
    def inner():
        return(mytextdata().upper())
    return inner
    
### define the decorator
@mydecorator
def mytextdata():
    return('hello parag')
print(mytextdata())
```

### Multiple Decorator Calls

A decorator can be called multiple times. Just place the decorator above the function you want to decorate

```python
def mydeco(func):
    def inner():
        return(func().upper())
    return(inner)

@mydeco
def myfirstcase():
    return("hello parag")

@mydeco
def mysecondcase():
    return("I am sorry")

@mydeco
def mythirdcase():
    return("I live in noida")
    
print(myfirstcase())
print(mysecondcase())
print(mythirdcase())
```

### Arguments in the Decorated Function

Functions that requires arguments can also be decorated, just make sure you pass the arguments to the wrapper function

```python
# Here we are passing the arguments in the decorator

# define the decorator function to convert the text into upper case.

def myfunctionargdata(myfunctiondatacall):
    def inner(name):
        return(myfunctiondatacall(name).upper())
    return(inner)

@myfunctionargdata
def myfunctiondatacall(name):
    return('hello' + ' ' + name)

print(myfunctiondatacall('Parag'))
```
