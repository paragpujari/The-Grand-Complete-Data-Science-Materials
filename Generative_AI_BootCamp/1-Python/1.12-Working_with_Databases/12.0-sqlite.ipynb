```python
import sqlite3
```

```python
### sqlite ----> using sqlite help the python to get connected to the sqlite database

connection=sqlite3.connect('example.db')
```

### OBSERVATIONS:

```python
### Create a cursor

cursor = connection.cursor()
```

### OBSERVATIONS:

1. Here a cursor is created that is used to point to every SQL statement using the connection object.

```python
### Create a Table using the cursor

cursor.execute(''' 
Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text
    )
''')
```

```python
### Fetch the data from the table using cursor

cursor.execute(''' 
 SELECT * FROM employees
''')
```

```python
### Insert the data into the table using cursor

cursor.execute(''' 
Insert Into employees(name,age,department)
               values('Krish',32,'Data Scientist')
''')

cursor.execute(''' 
INSERT INTO employees (name, age, department)
VALUES ('Bob', 25, 'Engineering')
''')


cursor.execute(''' 
INSERT INTO employees (name, age, department)
VALUES ('Charlie', 35, 'Finance')
''')


### commit the changes

connection.commit()
```

```python
### Query the data from the table using cursor

cursor.execute(''' 
  SELECT * FROM employees
''')

### Fetches all the rows

rows = cursor.fetchall()

for x in rows:
    print(x)
```

### OBSERVATIONS:

1.  cursor is used to point to all the SQL statements using python.

2.  fetchall is used to point to all the rows

3.  Then the iteration is performed on all the rows to extract the records.

```python
## Update the data in the table  where name = "Krish"  -----> age = 34

cursor.execute(''' 
       UPDATE employees
       SET    age  =   34
       WHERE  name =  "Krish"
    ''')
```

```python
### Again query the data from the table using cursor

cursor.execute(''' 
  SELECT * FROM employees
''')

### Fetches all the rows

rows = cursor.fetchall()

for x in rows:
    print(x)
```

### OBSERVATIONS:

1. Wherever the name is Krish, their the age has been updated to 34.

```python
## Delete the data from the table whose name is 'Bob'

cursor.execute('''  
    DELETE from employees where name = 'Bob'
''')
```

```python
### Again query the data from the table using cursor

cursor.execute(''' 
  SELECT * FROM employees
''')

### Fetches all the rows

rows = cursor.fetchall()

for x in rows:
    print(x)
```

```python
### close the connection 

connection.close()
```

```python
### Working with the sales data

## Working Wwith Sales Data
# Connect to an SQLite database

### Create a connection object

connection = sqlite3.connect('sales_data.db')

### using the connection object, create a cursor
cursor = connection.cursor()

# Create a table for sales data
cursor.execute(''' 
CREATE TABLE IF NOT EXISTS sales (
    id INTEGER PRIMARY KEY,
    date TEXT NOT NULL,
    product TEXT NOT NULL,
    sales INTEGER,
    region TEXT
)
''')

# Insert data into the sales table
sales_data = [
    ('2023-01-01', 'Product1', 100, 'North'),
    ('2023-01-02', 'Product2', 200, 'South'),
    ('2023-01-03', 'Product1', 150, 'East'),
    ('2023-01-04', 'Product3', 250, 'West'),
    ('2023-01-05', 'Product2', 300, 'North')
]

### using the cursor, insert the records into the table
cursor.executemany(''' 
   INSERT INTO sales(date,product,sales,region) VALUES(?,?,?,?)
''',sales_data)
```

```python
### Query the data from the table using cursor

cursor.execute(''' 
  SELECT * FROM sales
''')

### Fetches all the rows

rows = cursor.fetchall()

for x in rows:
    print(x)
```

```python
### close the connection 

connection.close()
```
