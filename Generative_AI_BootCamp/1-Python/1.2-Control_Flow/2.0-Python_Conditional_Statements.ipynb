### 3:  else Statement

```python
### Check if the number if positive or negative

num = 10

if(num > 0):
    print("Positive:", num)
else:
    print("Negative:", num)
```

```python
### Check if the number if positive or negative

num = -10

if(num > 0):
    print("Positive:", num)
else:
    print("Negative:", num)
```

```python
### Check if the person is eligible to vote or not

age = 18

if(age > 18):
    print("Person is eligible to vote:")
else:
    print("Person is not eligible to vote:")
```

### 4:  elif Statement

```python
### Check if the number if positive or negative or zero

num = 10

if(num > 0):
    print("Positive:", num)
elif(num < 0):
    print("Negative:", num)
else:
    print("Zero:", num)
```

```python
### Check if the number if even, odd or zero

num = int(input("Enter the number:"))

print(num)

if((num%2)==0):
    print("Even number:", num)
elif(num == 0):
    print("Zero:", num)
else:
    print("Odd number:", num)
```

### 5.  Nested Conditional Statements

```python
### Check if the number if even positive or negative

num = int(input("Enter the number:"))

if((num%2)==0):
    if(num > 0):
        print("Number is even positive:", num)
    else:
        print("Number is even negative:", num)
else:
    print("Number is odd:", num)
```

```python
## Practical Examples

## Determine if a year is a leap year using nested condition statement

year = int(input("Enter the year:"))

if((year%4)==0):
    if((year%100)==0):
        if((year%400)==0):
            print(year,"is a leap year:")
        else:
            print(year,"is not a leap year:")
    else:
        print(year,"is a leap year:")
else:
    print(year,"is not a leap year:")
```

```python
## Simple Calculator program

num1 = int(input("Enter the first  number:"))
num2 = int(input("Enter the second number:"))
operation = input("Enter the opeartion : (+,-,*,/)")

if(operation == "+"):
    result = num1 + num2
elif(operation == "-"):
    result = num1 - num2
elif(operation == "*"):
    result = num1 * num2
else:
    result = num1 / num2

print("Result of the Arithmetic Operation is:", result)
```
