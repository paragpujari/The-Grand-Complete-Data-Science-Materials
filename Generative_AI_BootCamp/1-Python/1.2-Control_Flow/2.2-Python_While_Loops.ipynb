### Looping using while


1.  Check Palindrome Number

```python
num = int(input("Enter the number:"))

x = num

rev = 0

while(x != 0):
    d = x % 10
    rev = (rev*10)+d
    x = x // 10

if(rev == num):
    print(rev, "is a Palindrome number:")
else:
    print(rev, "is not a Palindrome number:")
```

2. Armstrong Number

```python
num = int(input("Enter the number:"))

x = num

sum = 0

while(x != 0):
    d   = x % 10
    sum = sum + (d ** 3)
    x   = x // 10

if(sum == num):
    print(num, "is an Armstrong number:")
else:
    print(num, "is not an Armstrong number:")
```

3. Count the Digits in the number

```python
num = int(input("Enter the number:"))

x = num

count = 0

while(x != 0):
    d = x % 10
    count = count + 1
    x = x // 10

print("Count of all the digits in the number is:", count)
```

4. Print the Fibonacci Series till N-terms

```python
num = int(input("Enter the terms:"))

a = 0
b = 1

count = 0

while(count <= num):
    print(a)
    c = a + b
    a = b
    b = c
    count = count + 1
```
