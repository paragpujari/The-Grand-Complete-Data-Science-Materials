### Data Visualization With Seaborn

Seaborn is a Python visualization library based on Matplotlib that provides a high-level interface for drawing attractive and informative statistical graphics. Seaborn helps in creating complex visualizations with just a few lines of code. In this lesson, we will cover the basics of Seaborn, including creating various types of plots and customizing them.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips_data = sns.load_dataset('tips')

print(tips_data)
```

```python
### perform the scatter plot for the total_bill vs tip

import seaborn as sns

sns.scatterplot(x='total_bill',y='tip',data=tips_data)
```

### OBSERVATIONS:

1.  With the increase in the number of total_bills, the tip increases.

```python
## Line Plot

sns.lineplot(x='size',y='total_bill',data=tips_data)
```

```python
## Categorical Plots
## BAr Plot

sns.barplot(x='day',y='total_bill',data=tips_data)
plt.title('Bar Plot of Total Bill By Day')
plt.show()
```

### OBSERVATIONS:

1.  The total bill per tip is very high in weekends than in the weekdays.

```python
### Box Plot

sns.boxplot(x='day',y='total_bill', data = tips_data)
```

### OBSERVATIONS:

1. The number of outliers data is maximum in thursday and minimum in the friday.

```python
## Violin Plot

sns.violinplot(x='day',y='total_bill', data = tips_data)
```

```python
### Histograms for total_bill

sns.histplot(tips_data['total_bill'],bins=10,kde=True)
```

```python
sns.kdeplot(tips_data['total_bill'],fill=True)
```

```python
# Pairplot
sns.pairplot(tips_data)
```

```python

corr = tips_data[['total_bill','tip','size']].corr()

corr
```

```python
sns.heatmap(corr, annot=True)
```

```python
import pandas as pd
sales_df=pd.read_csv('sales_data.csv')
sales_df.head()
```

```python
## Plot total sales by product

plt.figure(figsize=(8,5))

sns.barplot(x = "Product Category", y = "Total Revenue",data=sales_df,estimator=sum)
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales')
plt.show()
```

### OBSERVATIONS:

1. Electronics is the product that has the highest total sales.
