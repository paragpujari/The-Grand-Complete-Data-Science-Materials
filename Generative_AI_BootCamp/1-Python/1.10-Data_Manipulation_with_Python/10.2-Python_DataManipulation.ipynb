```python
## Renaming Columns

df.rename(columns={'Date':'Sales Date'},inplace=True)


df.head()
```

### OBSERVATIONS:

1. After renaming, the column 'Date' has been renamed to 'Sales Date'.

```python
## change datatypes
df['Value_new']=df['Value'].fillna(df['Value'].mean()).astype(int)
df.head()
```

```python
### Square the value present in 'Value' and store in a new column 'Value_news'

df['Value_news'] = df['Value'].apply(lambda x:x**2)

df.head()
```

### OBSERVATIONS:

1. We have made the new column 'Value_news' that stores the square of the 'Value' column.

```python
### Find out the mean value of the sales in a product wise manner

grouped_mean = df.groupby('Product')['Sales'].mean()
```

```python
print(grouped_mean)
```

### OBSERVATIONS:

1.  Here we have obtained the mean value of the sales per product.

```python
### Find out the mean value of the sales in a product wise manner, region wise

grouped_mean_sales_data = df.groupby(['Product','Region'])['Sales'].mean()
```

```python
print(grouped_mean_sales_data)
```

### OBSERVATIONS:

1.  Here we have obtained the mean values of the sales per product per region.

```python
## aggregate multiple functions
groudped_agg=df.groupby('Region')['Value'].agg(['mean','sum','count'])
groudped_agg
```

```python
### Merging and joining Dataframes
# Create sample DataFrames
df1 = pd.DataFrame({'Key': ['A', 'B', 'C'], 'Value1': [1, 2, 3]})
df2 = pd.DataFrame({'Key': ['A', 'B', 'D'], 'Value2': [4, 5, 6]})
```

```python
df1
```

```python
df2
```

```python
### perform the inner merge on the data frames

pd.merge(df1,df2,on='Key',how='inner')
```

```python
### perform the left merge on the data frames

pd.merge(df1,df2,on='Key',how='left')
```

```python
### perform the right merge on the data frames

pd.merge(df1,df2,on='Key',how='right')
```

```python
### perform the outer merge on the data frames

pd.merge(df1,df2,on='Key',how='outer')
```

### OBSERVATIONS:

1.   Here we have performed the left join, right join , inner join and the outer join on the two dataframes to get the output.
