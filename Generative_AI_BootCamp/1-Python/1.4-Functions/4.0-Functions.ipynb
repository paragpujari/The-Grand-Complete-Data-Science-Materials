### Introduction to Functions


#### Definition:

A function is a block of code that performs a specific task. Functions help in organizing code, reusing code, and improving readability.

```python
### To check whether the number is even or odd

def even_odd_number(num):
    if((num%2)==0):
        print("Even number:", num)
    else:
        print("Odd number:", num)


num = int(input("Enter the number:"))
even_odd_number(num)
```

```python
## function with multiple parameters

def add(a,b):
    return(a*b)


a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))

res = add(a,b)

print("Addition of two numbers is:", res)
```

```python
## Default Parameters

def greet(name):
    print(f"Hello {name}, Welcome to Generative AI Course:")

name = "Parag"
greet(name)
```

```python
### Positional arguments

def print_numbers(*numbers):
    for x in numbers:
        print(x)
        
print_numbers(1,2,3,4,5,6,7,8,"Krish")
```

```python
def print_details(**kwargs):
    for key, values in kwargs.items():
        print(f"{key}:{values}")

print_details(name="Krish",age="32",country="India")
```

```python
def print_details(**kwargs):
    for x,y in kwargs.items():
        print(f"{x}:{y}")

print_details(name="Krish",age="32",country="India")
```

```python
def print_details(*args,**kwargs):
    for x in args:
        print(x)

    for key,values in kwargs.items():
        print(f"{key}:{values}")

print_details(1,2,3,4,"Krish",name="Krish",age="32",country="India")
```

```python
### Return multiple parameters
def multiply(a,b):
    return a*b,a

multiply(2,3)
```
