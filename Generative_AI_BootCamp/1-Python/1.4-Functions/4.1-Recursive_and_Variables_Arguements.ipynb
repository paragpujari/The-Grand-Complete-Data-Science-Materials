### Recursive Functions:

1.  Factorial of a number

```python
def factorial(n):
    if(n == 1):
        return(1)
    else:
        return(n * factorial(n-1))

n = int(input("Enter the number:"))
f = factorial(n)
print("Factorial of the number is:", f)
```

2. Fibonacci of a number

```python
def fibonacci(n):
    if(n <= 1):
        return(n)
    else:
        return(fibonacci(n-1) + fibonacci(n-2))
        
n = int(input("Enter the number:"))
f = fibonacci(n)
print("Fibonacci of the number is:", f)
```

3. Sum of first N natural numbers

```python
def find_sum_natural_numbers(n):
    if(n <= 1):
        return(1)
    else:
        return(n + find_sum_natural_numbers(n-1))


n = int(input("Enter the number:"))
s = find_sum_natural_numbers(n)
print("Sum of n natural numbers:", s)
```

4. Check if a string is palindrome

```python
def is_Palindrome(str):
    x = str[::-1]
    #return(x)

    if(x[0] == x[-1]):
        return(True)
    else:
        return(False)


str = input("Enter the string:")

ans = is_Palindrome(str)
print(ans)
```

5. Count the number of digits in the number

```python
def count_digits(n):
    if(n == 0):
        return(0)
    else:
        return(1 + count_digits(n//10))

n = int(input("Enter the number:"))
ans = count_digits(n)
print("Number of digits in the number is:", ans)
```

6.  Power of the number

```python
def power_number(x,n):
    if(n == 0):
        return(1)
    else:
        return(x * power_number(x, n-1))


x = int(input("Enter the number:"))
n = int(input("Enter the term:"))
ans = power_number(x,n)
print(ans)
```

7. GCD of a number using recursion

```python
def gcd(a,b):
    if(b == 0):
        return(a)
    else:
        return(gcd(b,a%b))


a = int(input("Enter the first number:"))
b = int(input("Enter the second number:"))
x = gcd(a,b)
print(x)
```

8.  Binary Search using Recursion

```python
def binary_search(arr,start,end,target):
    if(start > end):
        return(-1)
    else:
        mid = (start+end)//2
    
    if(arr[mid] == target):
        return(mid)
    elif(arr[mid] > target):
        return(binary_search(arr,start,mid-1,target))
    else:
        return(binary_search(arr,mid+1,end,target))


arr = [1,3,5,7,9,11]
binary_search(arr,0,len(arr)-1,7)
```

### Variable Arguement Functions

```python
### print all the numbers in the tuple


def print_tuples(*args):
    for x in args:
        print(x)
        

tuples = (1,2,3,4,5)

print_tuples(tuples)
```

```python
### multiple all the numbers

def multiple(*args):
    pro = 1
    for x in args:
        pro = pro * x
    return(pro)

ans = multiple(2,3,4)
print(ans)
```

```python
### Print key-value pairs using **kwargs

def print_details(**kwargs):
    for key,value in kwargs.items():
        print(f"{key} : {value}")
        

print_details(name = "Parag", role = "Data Scientist")
```
