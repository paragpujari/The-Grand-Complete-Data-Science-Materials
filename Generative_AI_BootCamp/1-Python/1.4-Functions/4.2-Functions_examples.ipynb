### 1.  Check IF a String Is Palindrome

```python
def is_Palindrome(str):
    ### convert the string to lower case
    str = str.lower()

    rev = ""
    for x in str:
        rev = x + rev
    return(rev)



str = "madam"

res = is_Palindrome(str)

print(res)

if(res == str):
    print(res,"is Palindrome:")
else:
    print(res,"is not Palindrome:")
```

### 2. Calculate the factorials of a number using recursion

```python
def factorial(num):
    if(num <= 0):
        return(1)
    else:
        return(num*factorial(num-1))

num = int(input("Enter the number:"))

f = factorial(num)

print("Factorial of the number is:", f)
```

### 3. Temperature Conversion

```python
"""This function converts temperature between Celsius and Fahrenheit"""

def convert_temperature(temperature,unit):
    if(unit == 'C'):
        return((temperature * (9/5)) + 32)
    elif(unit == 'F'):
        return((temperature - 32) * (5/9))
    else:
        None


c = convert_temperature(25,'C')
print("Temperature in Celcius:", c)
```
