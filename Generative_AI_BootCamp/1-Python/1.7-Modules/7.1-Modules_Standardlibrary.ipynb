### Standard Library Overview


Python's Standard Library is a vast collection of modules and packages that come bundled with Python, providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly used modules and packages in the Python Standard Library.

```python
import array

### create an array of integer type

arr = array.array('i',[1,2,3,4,5])

print(arr)
```

```python
### find out the pi value

import math

print(math.pi)
```

```python
### find out the sqrt value

import math

x = math.sqrt(16)

x
```

```python
### Find out the random values between 1 to 10

import random

x = random.randint(1,10)

x
```

```python
### Find out the random choice

fruits = ['apple','banana','cherry']

x = random.choice(fruits)

x
```

```python
### Get the current working directory

import os

x = os.getcwd()

x
```

```python
### Make the directory

import os

os.mkdir('tes_dir')
```

```python
### Copy the contents of the source file to the destination

import shutil

shutil.copyfile('source.txt','destination.txt')
```

```python
## Data Serialization
import json
data={'name':'Krish','age':25}

json_str=json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
## csv

import csv

with open('example.csv',mode='w',newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['Krish',32])

with open('example.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
import re

pattern = r'\d+'

text = "There are 123 apples 456"

match = re.search(pattern,text)

print(match)
```

```python
## datetime
from datetime import datetime,timedelta

now=datetime.now()
print(now)

yesterday=now-timedelta(days=1)

print(yesterday)
```

```python
## time
import time
print(time.time())
time.sleep(2)
print(time.time())
```
