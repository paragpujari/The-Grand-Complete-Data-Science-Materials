### 3.  Naming Conventions

```python
#### 1. Variable should be descriptive

#### 2. It should have alphabet, letter, underscore or digit

#### 3. It should not start with digit

#### 4. It should be case sensitive

### valid variable names

name = "Parag"

Name = "Disha"

### Check if both the variables are same or not

print(name == Name)
```

### OBSERVATIONS:

1. name ---> Parag, Name ---> Disha. Evn though both the variables have the same name. But both are different because:---

   (a.) Python is case sensitive. It treats both the upper case and lower case names differently. So both are different.

### 4. Understanding Variable Types


Python is dynamically typed means the data type of the variable is only known during the runtime execution of the program.

```python
age = 30
height = 5.8
name = "Parag"
is_student = False


print("Name is:", name)

print("Data Type of name is:", type(name))
```

```python
print("Height is:", height)

print("Data Type of height is:", type(height))
```

```python
print("Age is:", age)

print("Data Type of age is:", type(age))
```

### 5.  Type Checking and Conversion

```python
x = '25'

print("Value of x before type conversion:", x)

print("Data Type of x before type conversion:", type(x))

print("=======================================================================")

ans = int(x)

print("Value of x after type conversion:", ans)

print("Data Type of x after type conversion:", type(ans))
```

```python
height  = 5.8

### integer form of height

x = int(height)

print(x)

### again convert the height from integer to float

res = float(x)

print(res)
```

### OBSERVATIONS:

1.  Type Conversion is applied on the height.

2.  Initially height was 5.8 . Type Conversion was applied on the height, the value got changed from 5.8 to 5. There was a loss of information.

3.  Then again the height was converted back from int to float (5 to 5.0)

4.  Type checking was done at first, then the type conversion was done.

### 6.    Dynamic Typing

```python
var = 10

print(var, type(var))

var = 4.567

print(var, type(var))

var = [1,2,3,4,5]

print(var, type(var))
```

### OBSERVATIONS:

1. var value was  declared as 10. Its data type is int.

2. var value was  declared as 4.567 . Its data type is float.

3. var value was  declared as [1,2,3,4,5]. Its data type is list.
