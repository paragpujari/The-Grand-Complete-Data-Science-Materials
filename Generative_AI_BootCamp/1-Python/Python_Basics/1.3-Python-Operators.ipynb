### Arithmethic Operation

```python
a = 10
b = 5

add_result = a + b
sub_result = a - b
mul_result = a * b
div_result = a / b
floo_resul = a // b
mod_result = a % b
exp_result = a ** b


print("Addition       is :", add_result)
print("Substraction   is :", sub_result)
print("Multiplication is :", mul_result)
print("Division       is :", div_result)
print("Floor          is :", floo_resul)  ### removes all the decimals
print("Modulus        is :", mod_result)  ### finds out the remainder
print("Exponential    is :", exp_result)
```

```python
## Simple Calculator

num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))

# Performing arithmetic operations

addition       = num1 + num2
subtraction    = num1 - num2
multiplication = num1 * num2
division       = num1 / num2
floor_division = num1 // num2
modulus        = num1 %  num2
exponentiation = num1 ** num2

# Displaying results
print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Division:", division)
print("Floor Division:", floor_division)
print("Modulus:", modulus)
print("Exponentiation:", exponentiation)
```
