### What is Python ?

Python is a programming language that can be used with the computer or any other technology to communicate to get out tasks done.

### Syntax:

A Syntax can be defined as the set of rules that involves the combination of all the possible symbols that are considered to be correctly structured in the program.


### Semantics:

Semantics can be defined as the meaning behind every text, symbol and code used in the program. It explains the meaning behind every line of the code used in the program.

It even specifies what the code is suppose to do when it runs.

### Rules used in Python:---

1. Python is case sensitive -----> It treats the small and capital letters seperately as two different terms.

```python
name = "Parag"

Name = "Pujari"


print(name)

print(Name)
```

```python
### To check whether the two names are same or different

print(name == Name)
```

### OBSERVATIONS:

1. Here when we compare name and Name, the output is False.
 
   This is because:--

     (a.)  Python is case-sensitive. It treats both the small and the capital names differently.

### Indentation

Indentation in Python is used to define the structure and hierarchy of the code. Unlike many other programming languages that use braces {} to delimit blocks of code, Python uses indentation to determine the grouping of statements. This means that all the statements within a block must be indented at the same level.

```python
age = 30

if(age>18):
    print("Eligible to vote:")
```

### OBSERVATIONS:
