### Lists

Lists helps in storing all the multiple values in one single variable.

Here all the elements in the lists can be of the same type or the different types.

Lists are always enclosed within closed brackets

```python
### empty lists

lists = []

print(lists)

print(type(lists))
```

```python
### lists with elements

lists = ['apples','banana','cherry','guava']

print(lists)

print(type(lists))
```

```python
### lists with elements

lists = [10,20,30,40]

print(lists)

print(type(lists))
```

### Properties of Lists

1. Lists are ordered

2. They are mutuable

3. It allows duplicates

4. It can have elements of same or different types

```python
### Lists are ordered

lists = [20,40,30,10,50]

print(lists)
```

3. After modifying the element of the list at index position 1 the value has changed from 40 to 70.

```python
### Lists allows duplicates

lists = [20,20,30,30,40,40,30,10,10,50]

print(lists)
```

### OBSERVATIONS:

### Python - Access Lists Items

#### Access Items

List items are indexed and you can access them by referring to the index number:

```python
### print the first element in the lists

thislist = ["apple", "banana", "cherry"]

print(thislist[0])
```

```python
### print the second element in the lists

thislist = ["apple", "banana", "cherry"]

print(thislist[1])
```

### Negative Indexing

Negative indexing means start from the end

-1 refers to the last item, -2 refers to the second last item etc.

```python
### print the last item in the lists

thislist = ["apple", "banana", "cherry"]

print(thislist[-1])
```

```python
### Print the second last item in the list

print(thislist[-2])
```

### Range of Indexes

You can specify a range of indexes by specifying where to start and where to end the range.

When specifying a range, the return value will be a new list with the specified items.

```python
# Return the third fourth and fifth item in the lists

thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]


print(thislist[2:5])
```

```python
# Return all the elements before kiwi


print(thislist[:4])
```

```python
# This example returns the items from "cherry" to the end

print(thislist[2:])
```

### Range of Negative Indexes

Specify negative indexes if you want to start the search from the end of the list:

```python
# This example returns the items from "orange" (-4) to, but NOT including "mango" (-1):

thislist = ["apple", "banana", "cherry", "orange", "kiwi", "melon", "mango"]


print(thislist[-4:-1])
```

### Python - Add List Items

###  Append Items


To add an item to the end of the list, use the append() method

```python
# append() = add element at the end of the list.

mylist = ['apple', 'banana', 'cherry']


### append "guava" at the end of the list

mylist.append("guava")

print(mylist)
```

### Extend List

To append elements from another list to the current list, use the extend() method

```python
thislist = ["apple", "banana", "cherry"]
tropical = ["mango", "pineapple", "papaya"]

### extend the elements of tropical into thislist

thislist.extend(tropical)

print(thislist)
```

### OBSERVATIONS:

1. All the elements of tropical has been extended into thislist.

extend() ------------> It adds the elements of another lists as an individual element.

append() ------------> It adds the elements of another list as an individual list.

#### Python - Join Lists

###  Join Two Lists

There are several ways to join, or concatenate, two or more lists in Python.

One of the easiest ways are by using the + operator.

```python
# Join two list:

list1 = ["a", "b", "c"]
list2 = [1, 2, 3]

lists = list1 + list2

print(lists)
```

```python
### extend the lists

list1.extend(list2)

print(list1)
```

```python
### append the list

list1.append(list2)

print(list1)
```

### OBSERVATIONS:

1. The main difference between the extend and append is that

   (a.)  extend -----> It adds each element into the list individually at the end of one list.

   (b.)  append -----> It considers the entire list as one element while inserting it into the other liat at the end.

### Reverse Order

What if you want to reverse the order of a list, regardless of the alphabet?

The reverse() method reverses the current sorting order of the elements

```python
# Reverse the order of the list items

thislist = ["banana", "Orange", "Kiwi", "cherry"]


thislist.reverse()

print(thislist)
```

### Sort List Alphanumerically


List objects have a sort() method that will sort the list alphanumerically, ascending, by default

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]


### Arrange all the elements of the list in ascending order
thislist.sort(reverse=False)

print(thislist)
```

### OBSERVATIONS:

1.   Python is case sensitive.

2.   Sorting is based on the ASCII Characters.

3.   Capital letters comes before the small letters.

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]

thislist.sort()

print(thislist)
```

```python
# Sort the list numerically

thislist = [100, 50, 65, 82, 23]

thislist.sort()

print(thislist)
```

### Sort Descending

To sort descending, use the keyword argument reverse = True

```python
thislist = ["orange", "mango", "kiwi", "pineapple", "banana"]

### Arrange all the elements in the list in descending order

thislist.sort(reverse=True)

print(thislist)
```

```python
# Sort the list numerically in descending order

thislist = [100, 50, 65, 82, 23]

thislist.sort(reverse=True)

print(thislist)
```

### Case Insensitive Sort

By default the sort() method is case sensitive, resulting in all capital letters being sorted before lower case letters

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]


thislist.sort()

print(thislist)
```

### OBSERVATIONS:

1.   Python is case sensitive.

2.   Sorting is based on the ASCII Characters.

3.   Capital letters comes before the small letters.

```python
thislist = ["banana", "Orange", "Kiwi", "cherry"]

### converts all the elements in the list to the lower case

thislist.sort(key = str.lower)

print(thislist)
```

### Using a While Loop

You can loop through the list items by using a while loop.

Use the len() function to determine the length of the list, then start at 0 and loop your way through the list items by referring to their indexes.

Remember to increase the index by 1 after each iteration.

```python
## Print all items, using a while loop to go through all the index numbers

thislist = ["apple", "banana", "cherry"]


l = len(thislist)

i = 0

while(i < l):
    print(i, thislist[i])
    i = i + 1
```

### Looping Using List Comprehension

List Comprehension offers the shortest syntax for looping through lists

```python
thislist = ["apple", "banana", "cherry"]

res = [x for x in thislist if(x != 0)]

print(res)
```

### List Comprehension

List comprehension offers a shorter syntax when you want to create a new list based on the values of an existing list.

### Example:

Based on a list of fruits, you want a new list, containing only the fruits with the letter "a" in the name.

Without list comprehension you will have to write a for statement with a conditional test inside

```python
fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

res = [x for x in fruits if(x != " ")]

print(res)
```

```python
# Only accept items that are not "apple"

fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

res = [x for x in fruits if(x != "apple")]

print(res)
```

```python
# You can use the range() function to create an iterable 

res = [x for x in range(11) if(x != 0)]

print(res)
```

```python
# Accept only numbers lower than 5

res = [x for x in range(5) if(x != 0)]

print(res)
```

```python
# Set the values in the new list to upper case

fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

res = [x.upper() for x in fruits if(x != " ")]

print(res)
```

```python
# Set all values in the new list to 'hello'

res = ["hello" for x in fruits if(x != "")]

res
```

```python
# Return "orange" instead of "banana"

fruits = ["apple", "banana", "cherry", "kiwi", "mango"]

res = ["orange" if x == "banana" else x for x in fruits]

print(res)
```
