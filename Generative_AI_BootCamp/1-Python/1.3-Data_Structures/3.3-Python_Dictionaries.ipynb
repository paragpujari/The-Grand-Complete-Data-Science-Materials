### Dictionaries

A Dictionary is an unordered collection of elements with the key-value pairs.

(i)   The key is immutable and cannot be changed.

(ii)  The value is mutable and it can be changed.


```python
### empty dictionary

empty_dict = dict({})

print(empty_dict)

print(type(empty_dict))
```

```python
### dictionary with elements

dicts = {"name":"Krish","age":32,"grade":24}

print(dicts)

print(type(dicts))
```

```python
# Single key is slways used

dicts = {"name":"Krish","age":32,"grade":24,"name":"Parag"}

print(dicts)

print(type(dicts))
```

### OBSERVATIONS:

1.  The name with the key value occuring at the last will be considered into account.

2.  Each key element will be considered at once.

```python
### accessing Dictionary Elements

student={"name":"Parag","age":30,"grade":'A'}

print(student["name"])

print(student["age"])

print(student["grade"])
```

```python
print(student.get("name"))

print(student.get("age"))

print(student.get("grade"))
```

```python
### Dictionaries are mutuable and we can either add or delete the elements from the dictionary

print(student)
```

```python
## update value for the key 'name'

student['name'] = 'nippon'

print(student)
```

```python
## update value for the key 'age'

student['age'] = 33

print(student)
```

```python
student['address'] = 'India'

print(student)
```

```python
## shallow copy

student_copy = student

print(student_copy)

print(student)
```

```python
### Modify the values of one entity

student['name'] = 'Parag'

print(student)

print(student_copy)
```

### OBSERVATIONS:

1. Here we are simply assigning the value of one dictionary to the other.

```python
student_copy = student

print(student)

print(student_copy)


### perform the shallow copy

student_copy = student.copy()

print(student)

print(student_copy)
```

```python
student['name'] = 'Parul'


print(student)

print(student_copy)
```

### OBSERVATIONS:

1. After applying shallow copy on the student, we are just copying the memory reference of the student to student_copy not the content.

```python
### Iterating Over Dictionaries
## You can use loops to iterate over dictionatries, keys,values,or items

## Iterating over keys

print(student)

for x in student.keys():
    print(x)
```

```python
## Iterating over values

for y in student.values():
    print(y)
```

```python
### Itearting over the items

for x,y in student.items():
    print(x,y)
```

```python
## Nested Disctionaries
students={
    "student1":{"name":"Krish","age":32},
    "student2":{"name":"Peter","age":35}
}
print(students)
```

```python
print(students["student1"]["name"])
```

```python
print(students["student1"]["age"])
```

```python
for x,y in students.items():
    print(x,y)
```

```python
### Dictionary Comprehension

res = {x for x in range(10) if(x%2)==0}

print(res)
```

```python
## Practical Examples

## Use a dictionary to count the frequency of elements in list

numbers=[1,2,2,3,3,3,4,4,4,4]
freq = {}

for x in numbers:
    if(x in freq):
        freq[x] = freq[x] + 1
    else:
        freq[x] = 1

print("count of the frequency of elements in list :", freq)
```
