### Tuples

1.  A tuple is a data structure that contains multiple items in a single variable.

2.  It is ordered.

3.  It is immutable.

4.  It contains the elements of different types.

5.  It can be indexed and contains all the duplicate elements.

```python
### empty tuple

empty_tuple = ()

print(empty_tuple)

print(type(empty_tuple))
```

```python
### Tuple with one element

tuples = (10,)

print(tuples)

print(type(tuples))
```

```python
### Tuples are ordered

tuples = (20,10,40,30,50)

print(tuples)
```

### OBSERVATIONS:

1.  All the elements of the tuple have a fixed index.

2.  The elements of the tuple will be printed in which they are saved and preserved.

```python
### Tuples are immutable

tuples = (20,10,40,30,50)

# tuples[1] = 70

print(tuples)
```

### OBSERVATIONS;

1. If we try to change any item inside the tuple, then it will throw an error. This is because

   (a.) The elements in the tuples are immutable.

```python
### It contains the elemets of same or different types

tuples = (20,10,30.98,True,(8+9j),'Parag')

print(tuples)
```

### OBSERVATIONS:

1. This tuple contains the elements of all the types like integer, float,boolean,complex,string.

```python
### Tuples also contains duplicates

tuples = (10,10,20,20,20,30,30,40,50)

print(tuples)
```

### Slicing

We can slice tuples, obtaining new tuples with the corresponding elements:


```python
mytuples = ("apple", "banana", "cherry", "orange", "kiwi", "melon", "mango")

print(mytuples)

# Slice from index 0 to index 2

print(mytuples[0:3])

# Slice from index 3 to index 4

print(mytuples[3:5])


# Return the third, fourth, and fifth item

print(mytuples[2:5])
```

```python
# This example returns the items from the beginning to, but NOT included, "kiwi"

print(mytuples[0:4])
```

```python
# This example returns the items from "cherry" and to the end
print(mytuples[2:])
```

```python
# This example returns the items from index -4 (included) to index -1 

print(mytuples[-4:])
```

```python
# This example returns the items from index -4 (included) to index -1 excluded

print(mytuples[-4:-1])
```

### Python - Update Tuples

Tuples are unchangeable, meaning that you cannot change, add, or remove items once the tuple is created.

But there are some workarounds.

Change Tuple Values Once a tuple is created, you cannot change its values. Tuples are unchangeable, or immutable as it also is called.

But there is a workaround. You can convert the tuple into a list, change the list, and convert the list back into a tuple

```python
x = ("apple", "banana", "cherry")

print(x)

y = x

print(y)

### Tuples are immutable as the elements in the tuple cannot be changed

# y[1] = "guava"

print(y)
```

### Add Items

Since tuples are immutable, they do not have a built-in append() method, but there are other ways to add items to a tuple.

Convert into a list: Just like the workaround for changing a tuple, you can convert it into a list, add your item(s), and convert it back into a tuple.

```python
### Tuple

print(x)

### Convert tuple to list

l = list(x)

print(l)

### Add the item into the list

l.append('guava')

print(l)

### convert the list back to tuple

tuples = tuple(l)

print(tuples)
```

```python
tups = tuples + ('mango',)
```

```python
tups
```

### Remove Items

Note: You cannot remove items in a tuple.

Tuples are unchangeable, so you cannot remove items from it, but you can use the same workaround as we used for changing and adding tuple items:

```python
### Tuple

x = ('apple', 'banana', 'cherry')

print(x)

### Convert tuple to list

l = list(x)

print(l)

### Remove the item into the list

l.remove('cherry')

print(l)

### convert the list back to tuple

tuples = tuple(l)

print(tuples)
```

```python
# The del keyword can delete the tuple completely

del tuples
```

### Unpacking a Tuple

When we create a tuple, we normally assign values to it. This is called "packing" a tuple

But, in Python, we are also allowed to extract the values back into variables. This is called "unpacking

```python
res = ('apple', 'banana', 'cherry', 'orange', 'guava')


print(res)
```

```python
### extracting the values from the tuple

a,b,c,d,e = res

print(a)
print(b)
print(c)
print(d)
print(e)
```

### Using Asterisk*


If the number of variables is less than the number of values, you can add an * to the variable name and the values will be assigned to the variable as a list:

```python
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

x,*y,z = fruits

print(x)

print(*y)

print(z)
```

```python
fruits = ("apple", "mango", "papaya", "pineapple", "cherry")

x,*y,z = fruits

print(x)
print(*y)
print(z)
```

### Python - Loop Tuples

### Loop Through a Tuple

You can loop through the tuple items by using a for loop.

```python

fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")

for x in fruits:
    print(x)
```

### Loop Through an Index

```python
l = len(fruits)

print(l)

for i in range(l):
    print(i, fruits[i])
```

```python
# Print all items, using a while loop to go through all the index numbers:

l = len(fruits)

i = 0

while(i < l):
    print(i, fruits[i])
    i = i + 1
```

### Python - Join Tuples

### Join Two Tuples

To join two or more tuples you can use the + operator:

```python
# 1. count() = returns the number of times a specified value appears in the tuple

thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)

x = thistuple.count(5)

print(x)
```
