### Sets:

1.  These are the in-built data types which contains the collection of all the unique elements.

2.  Sets are unordered and do not follow a specific order.

3.  It does not have any duplicate elements in it.

4.  They perform all the membership operations like union, intersection etc

```python
### empty sets

empty_set = {}

print(empty_set)

print(type(empty_set))
```

```python
### sets with elements

my_set = {1,2,3,4,5}

print(my_set)

print(type(my_set))
```

```python
### set with the lists of elements

lists = [1,2,3,4,5]

my_set = set(lists)

print(my_set)
```

```python
### empty list with set

my_set = set([])

print(my_set)
```

```python
print(type(my_set))
```

```python
## Basics Sets Operation
## Adding and Removing Elements

my_set = {1,2,3,4,5}

### add an element 6 into the set

my_set.add(6)

print(my_set)
```

### OBSERVATIONS:

1. The main difference between discard and remove is that 

   (a.)  discard ---------->   successfully even removes the element that is not present without throwing any error

   (b.)  remove  ---------->   throws an error if we try to remove an element not present in the set.

```python
### pop ----> removes the first element from the set

my_set = {1,2,3,4,5,6,7}

my_set.pop()

print(my_set)
```

### OBSERVATIONS:

1.  pop() --------->  By default, it will remove the first element from the set.

```python
## clear all the elements

my_set.clear()

print(my_set)
```

```python
## Set Memebership test
my_set={1,2,3,4,5}

print(3 in my_set)

print(10 not in my_set)
```

```python
## Mathematical Operation
set1={1,2,3,4,5,6}
set2={4,5,6,7,8,9}


### Union operation

x = set1.union(set2)

print(x)

### intersection operation

y = set1.intersection(set2)

print(y)

### set difference

res = x - y

print(res)


### intersection update

set1.intersection_update(set2)
print(set1)


## Symmetric Difference

res = set1.symmetric_difference(set2)
print(res)
```

```python
## Sets Methods
set1={1,2,3,4,5}
set2={3,4,5}

### Check the subset
print(set2.issubset(set1))
print(set1.issuperset(set2))
```

```python
lst=[1,2,2,3,4,4,5]

x = set(lst)

print(x)
```

```python
### Counting Unique words in text

text="In this tutorial we are discussing about sets"

x = text.split()

print(x)

res = set(x)

print(res)

### get the total words in the text

ans = len(res)

print("Count of all the unique words in the text:", ans)
```
