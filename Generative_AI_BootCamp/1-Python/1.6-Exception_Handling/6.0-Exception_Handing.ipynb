### Exception Handling

An exception is an error that occurs during the execution of code.

### try

It allows us to test the block of code containing the error.

### except

It helps in handling the error.

### else

It helps in executing the code if it does not have any errors.

### finally

It helps in executing the code regardless of the result of try-except block.

```python
## Division by zero exception


try:
    a = 10
    b = 0
    res = a/b
except(ZeroDivisionError) as e:
    print("Zero Division Exception has occured:")
else:
    print(res)
finally:
    print("Finally, this exception has occured:")
```

```python
## Name Error Exception

try:
    a = 10
    print(b)
except NameError as e:
    print("Name Error Exception has occured:")
else:
    print(b)
finally:
    print("Finally, this exception has occured:")
```

```python
### Index Out of Bound Exception

arr = [10,20,30,40,50]

try:
    print(arr[10])
except ZeroDivisionError as e:
    print("Zero Division Exception has occured:")
except NameError as e:
    print("Name Exception has occured:")
except  IndexError as e:
    print("Index Out of Bound Exception")
else:
    print("Some exception has occured")
finally:
    print("Finally, this exception has occured:")
```

```python
### Value Error Exception

s = "abc"

try:
    res = int(s)
except ValueError as e:
    print("Value Error Exception has occured:")
except ZeroDivisionError as e:
    print("Zero Division Exception has occured:")
except NameError as e:
    print("Name Exception has occured:")
except  IndexError as e:
    print("Index Out of Bound Exception")
else:
    print("Some exception has occured")
finally:
    print("Finally, this exception has occured:")
```

```python
### Raise the Value Error saying that "Ages cannot be negative"

try:
    age = -18
    if(age < 0):
        raise ValueError("Ages cannot be negative")
except ZeroDivisionError as e:
    print("Zero Division Exception has occured:")
except ValueError as e:
    print("Value Error Exception has occured:")
except IndexError as e:
    print("Index Out of Bound Exception has occured:")
except NameError as e:
    print("Name Exception has occured:")
else:
    print("Some other exception has occured:")
finally:
    print("Finally, this exception has occured:")
```

```python
# Raise Type Error saying that Only Integers are allowed.

try:
    s = "abc"
    if(type(s) is not int):
        raise TypeError("Only Integers are allowed:")
except ZeroDivisionError as e:
    print("Zero Division Exception has occured:")
except TypeError as e:
    print("Type Error Exception has occured:")
except NameError as e:
    print("Name Error Exception has occured:")
except IndexError as e:
    print("Index Error Exception has occured:")
else:
    print("Some other exception has occured:")
finally:
    print("Finally, this exception has occured:")
```
