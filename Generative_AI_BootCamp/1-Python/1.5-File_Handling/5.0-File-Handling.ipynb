### File Handling

It is one of the most important operation used in Python that is used for creating, reading and deleting the files.

```python
# Write the contents into the file

with open('text1.txt','w') as f:
    f.write('Hi, My name is Parag.\nI was working at Motherson Technology Services Limited.\nI am currently living in noida.')
## close the file
f.close()
```

```python
### Read the contents from the file

with open('text1.txt','r') as f:
    print(f.read())
### close the file
f.close()
```

```python
### Read the contents from the file

with open('test1.txt','r') as f:
    print(f.read())
### close the file
f.close()
```

```python
### Read only one line from the text

with open('test1.txt','r') as f:
    print(f.readline())
### close the file
f.close()
```

### OBSERVATIONS:

1.  readline() ------------> It reads only the first line from the text.

```python
# Read two lines from the file

with open('test1.txt','r') as f:
    ### read two lines from the files
    print(f.readline())
    print(f.readline())
### close the file
f.close()
```

```python
### Loop through the file line by line and print it

with open('test1.txt','r') as f:
    ### print the lines
    for i in f:
        print(i)
### close the file
f.close()
```

### Write to an Existing File

To write to an existing file, you must add a parameter to the open() function:

(i)      "a" - Append - will append to the end of the file

(ii)     "w" - Write - will overwrite any existing content

```python
### Add the new content at the end of the existing efile

with open('test1.txt','a') as f:
    ### Add the new content into the existing file
    f.write("\nI would like to solve the problems using the data.")
### close the file
f.close()
```

```python
### Read the content from the text

with open('test1.txt','r') as f:
    print(f.read())
### close the file
f.close()
```

```python
### Overwrite the entire file with the new content

with open('text1.txt','w') as f:
    ### write the new content into the file
    f.write("\nI would like to solve the problems using the data.")
### close the file
f.close()
```

```python
### Read the content from the text

with open('text1.txt','r') as f:
    print(f.read())
### close the file
f.close()
```

###  Delete a File

To delete a file, you must import the OS module, and run its os.remove() function

```python
import os

os.remove('tools1.txt')
```

### Create a New File

To create a new file in Python, use the open() method, with one of the following parameters

(i) x = Creates a new file but returns an error if the same file exists.

(ii) a = Creates a new file if that file does not exists.

(iii) w = Creates a new file if that file does not exists.

```python
### Create a new file 
with open('tools1.txt','x') as f:
    f.write("\n I am Parag.")
### close the file
f.close()
```

### Check if File exist:

To avoid getting an error, you might want to check if the file exists before you try to delete it

```python
import os

### Check for the file path

x = os.path.exists('text1.txt')

print(x)

### delete the file if exists

if(x):
    os.remove('text1.txt')
else:
    print("File is already removed.")
```
