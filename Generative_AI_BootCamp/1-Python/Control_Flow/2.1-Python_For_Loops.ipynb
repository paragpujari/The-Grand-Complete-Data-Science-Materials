### Looping using for


#### 1.  Print Numbers from 1 to 10

```python
for x in range(1,11):
    print(x)
```

### 2. Print Even Numbers from 1 to 20

```python
for x in range(1,21):
    if((x%2)==0):
        print(x)
```

### 3. Sum of First N Natural Numbers

```python
sum=0
num = int(input("Enter the number:"))

for x in range(1,num+1):
    sum = sum + x

print("Sum of N Natural numbers is:", sum)
```

### 4. Factorial of a Number

```python
fact=1
num = int(input("Enter the number:"))

for x in range(1,num+1):
    fact = fact * x

print("Factorial of the number is:", fact)
```

### 5. Multiplication Table

```python
num = int(input("Enter the number:"))

for x in range(1,11):
    print(num, "*",x,"=",(num*x))
```

### 6. Reverse a String Using for Loop

```python
name = input("Enter the name:")

print(name)

rev = " "

for x in name:
    rev = x + rev

print("Reversed name is:", rev)
```

### 7. Count Vowels in a String

```python
text = input("Enter the text:")
### convert the text to lower case
text = text.lower()

print(text)

nv = 0

for x in text:
    if(x in "aeiou"):
        nv = nv + 1

print("Number of vowels in the text is:", nv)
```

### 8. Find Largest Element in a List

```python
numbers = [10, 45, 23, 67, 89, 34]
### initialize the first element to the largest

largest = numbers[0]

for x in numbers:
    if(x > largest):
        largest = x

print("Largest element in the list is:", largest)
```

### 9. Print Pattern (Star Triangle)

```python
n = 7

for i in range(1,(n+1)):
    for j in range(1,i):
        print("*",end=" ")
    print("\n")
```

### 10. Check Prime Number

```python
num = int(input("Enter the number:"))
is_prime = True

if(num <= 1):
    is_prime = False

for x in range(2,num):
    if((num%x)==0):
        is_prime = False
    break

if(is_prime):
    print("Prime number:", num)
else:
    print("Not a prime number :", num)
```

### 11. Find Duplicates in a List

```python
numbers = [1, 2, 3, 4, 2, 5, 6, 3]

freq = {}

for x in numbers:
    if(x in freq):
        freq[x] = freq[x] + 1
    else:
        freq[x] = 1

print(freq)

res = dict(freq)

print(res)

for x,y in res.items():
    if(y > 1):
        print(x)
```
