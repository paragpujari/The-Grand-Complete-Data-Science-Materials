#### Step 5: Divide the dataset into independent and dependent features

```python
#### independent features

X = df[['YearsExperience']]

print(X)
```

```python
#### dependent features

Y = df['Salary']

print(Y)
```

#### Step 6: Divide the independent and dependent features into training and testing data

```python
from    sklearn.model_selection import  train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X,Y,test_size=0.2,random_state=42)
```

### Step 6: Train the Model using Linear Regression , Ridge, Lasso, Elastic Net

### (a.) Linear Regression

```python
from    sklearn.linear_model    import LinearRegression, Ridge, Lasso, ElasticNet

#### create an object for LinearRegression

lin = LinearRegression()

### using the object train the model

lin.fit(X_train, Y_train)

#### using the object, predict the model

Y_pred_lin = lin.predict(X_test)

print(Y_pred_lin)


#### get the score of the model

r2 = r2_score(Y_test, Y_pred_lin)

print("R2 Score of the model is:", r2)
```

#### OBSERVATIONS:

1.  As the R2 Score of the model is 0.9. It is very high. So the model is working perfectly fine.

### (b.)  Ridge Regression

```python
from    sklearn.linear_model    import LinearRegression, Ridge, Lasso, ElasticNet

#### create an object for Ridge Regression

ridge = Ridge(alpha=1.0)

### using the object train the model

ridge.fit(X_train, Y_train)

#### using the object, predict the model

Y_pred_ridge  = ridge.predict(X_test)

print(Y_pred_ridge)

#### get the score of the model

ridge_r2 = r2_score(Y_test, Y_pred_ridge)

print("R2 Score of the model is:", ridge_r2)
```

#### OBSERVATIONS:

1.  As the R2 Score of the model using Ridge Regression is 0.9. It is very high. So the model is working perfectly fine.

2.  Ridge Regression is used to prevent the overfitting.

### (c.)  Lasso Regression

```python
from    sklearn.linear_model    import LinearRegression, Ridge, Lasso, ElasticNet

#### create an object for Lasso Regression

lasso = Lasso(alpha=0.1)

### using the object train the model

lasso.fit(X_train, Y_train)

#### using the object, predict the model

Y_pred_lasso  = lasso.predict(X_test)

print(Y_pred_lasso)

#### get the score of the model

lasso_r2 = r2_score(Y_test, Y_pred_lasso)

print("R2 Score of the model is:", lasso_r2)
```

#### OBSERVATIONS:

1.  As the R2 Score of the model using Lasso Regression is 0.9. It is very high. So the model is working perfectly fine.

2.  Lasso Regression is used to perform the feature selection.

### (d.)  Elastic Net Regression

```python
from    sklearn.linear_model    import LinearRegression, Ridge, Lasso, ElasticNet

#### create an object for Lasso Regression

elasticnet = ElasticNet(alpha=1.0, l1_ratio = 0.5)

### using the object train the model

elasticnet.fit(X_train, Y_train)

#### using the object, predict the model

Y_pred_elasticnet  = elasticnet.predict(X_test)

print(Y_pred_elasticnet)

#### get the score of the model

lasso_elasticnet = r2_score(Y_test, Y_pred_elasticnet)

print("R2 Score of the model is:", lasso_elasticnet)
```

#### OBSERVATIONS:

1.  As the R2 Score of the model using Elastic Net Regression is 0.8962. It is very high. So the model is working perfectly fine.

2.  Elastic Net Regression is used to perform the feature selection as well as remove the overfitting.

3.  The r2 score for Lasso > r2 score for ridge > r2 score for elastic net. 

### Step 7: Perform Cross Validation

```python
from sklearn.model_selection import cross_val_score

### Train the model using Ridge Regression

model = Ridge(alpha=1.0)
scores = cross_val_score(model,X,Y,cv = 5, scoring='r2')

print("Cross Validation score is:", scores)

print("Average scores for the cross validation is:", np.mean(scores))
```
