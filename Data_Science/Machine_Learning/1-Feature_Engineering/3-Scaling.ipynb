#### Scaling

Scaling refers to the technique of converting all the input feature values in the same scale range.


##### Types of Scaling

#### 1.   Standardization

(a.)   It is used when the data is normalized.

(b.)   It is used when the mean is 0 and standard deviation is 1.

(c.)   Range = [0,1]

```python
from  sklearn.preprocessing     import  StandardScaler

print("==========================Original DataFrame=====================================================")

data = pd.DataFrame({
    'age': [10, 20, 30, 40, 100]
})

print(data)


print("==========================Scaled DataFrame=======================================================")
##### create an object for StandardScaler

sc  =   StandardScaler()

data['age_scaled'] = sc.fit_transform(data['age'])

print(data)
```

#### 2. Min Max Scaler

1.  It is used in neural networks

2.  Range = [-1,1]

```python
from  sklearn.preprocessing     import  MinMaxScaler

print("==========================Original DataFrame=====================================================")

data = pd.DataFrame({
    'age': [10, 20, 30, 40, 100]
})



print(data)


print("==========================Min Max DataFrame=======================================================")

minmax = MinMaxScaler()


data['age_minmax'] = minmax.fit_transform(data['age'])

print(data)
```
