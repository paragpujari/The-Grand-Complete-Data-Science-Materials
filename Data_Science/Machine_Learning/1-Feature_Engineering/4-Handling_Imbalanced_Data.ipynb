#### Handling Imbalanced Data

(a.)  Upsampling ------>

It refers to the technique of increasing the minority class samples so as to make it at par with the majority class samples.

```python
#### Load the Dataset

import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns


from    sklearn.utils       import  resample
```

```python
### Load the dataset

df = pd.read_csv("diabetes.csv")
```

```python
df
```

```python
#### Count the categories of Outcome '0' and Outcome '1' 

df['Outcome'].value_counts()
```

#### OBSERVATIONS:

1.  Here we see that the count of the category 0 is much higher than the count of category 1. So here we need to perform upsampling so as to increase the count of the category 1. 

2.  Thus the count of category 1 will be at par with the count of category 0

```python
#### perform upsampling

df_majority = df[df['Outcome'] == 0]

df_minority = df[df['Outcome'] == 1]
```

```python
df_majority
```

```python
df_minority
```

```python
#### perform the upsampling on the minority class

df_minority_upsample = resample(
                                 df_minority                                      ,
                                 replace        =     True                        ,
                                 n_samples      =     len(df_majority)            ,
                                 random_state   =       42  
                                )
```

```python
df_minority_upsample
```

```python
#### Join the resampled data

df_new = pd.concat([df_majority, df_minority_upsample])
```

```python
df_new['Outcome'].value_counts()
```

#### OBSERVATIONS:

1.  After performing the upsampling technique, the count of the minority class has been increased and has been at par with the majority class.

(b.)  Downsampling -------->

It refers to the technique of decreasing the majority data points so as to make it at par with the minority data points.

```python
from  sklearn.utils  import  resample



##### perform the Downsampling on the majority data points

### Load the dataset

df = pd.read_csv("diabetes.csv")

print(df)
```

```python
#### Count the values in each categories

df['Outcome'].value_counts()
```

```python
#### perform the downsampling on the majority data points

df_majority_downsample = resample(
    df_majority                                                ,
    replace           =             True                       ,
    n_samples         =             len(df_minority)           ,  
    random_state      =               42
)
```

```python
#### Join  the  resamples data

df_new = pd.concat([df_minority, df_majority_downsample])
```

```python
df_new
```

```python
#### Again count the values in each categories for the resmaples data

df_new['Outcome'].value_counts()
```

#### OBSERVATIONS:

1.  After performing the downsampling technique, the count of the majority class has been decreased and has been at par with the minority class.
