#### Encoding

Encoding referes to the technique of converting the catgeorical features into the numerical formats.


#### Type of Encoding

1.  Ordinal Encoding --------------> It can handle ordered data.

2.  One Hot Encoding ---------------> It can handle nominal data.

3.  Label Encoding   ---------------> It can handle ordinal data.

```python
import  numpy   as  np
import  pandas  as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

print("============================Original DataFrame===========================================")

data = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'blue']
})

print(data)


print("============================Label Encoding on the color =================================")


from  sklearn.preprocessing  import LabelEncoder

#### create an object for LabelEncoder

label = LabelEncoder()


data['color_label'] = label.fit_transform(data['color'])

print(data)
```

#### OBSERVATIONS:

1.  The Label Encoding has been applied on the color column and various color categories in the color column has been assigned with the numerical formats


    (a.)        blue  ------------------------>           0

    (b.)        green ----------------------->           1

    (c.)        red   ------------------------>           2

```python
#### Apply One Hot Encoding on the 'color' column of the dataframe

data_onehot = pd.get_dummies(data['color'])
```
