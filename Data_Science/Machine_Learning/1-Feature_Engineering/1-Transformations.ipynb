#### Transformations:--

It refers to the process of changing the shape of the input data or the distribution of the input data to make it more suitable , so that it can easily be trained by the machine learning models.

#### Importance of Transformations:---

1.  To change the shape of the input data so that it can more easily be trained by the machine learning models.

2.  To improve the speed and accuracy of the Machine Learning models.

3.  To reduce the effect of the outliers

4.  To normalize the data.

```python
import  numpy   as  np
import  pandas  as  pd
from    sklearn.preprocessing   import  PowerTransformer

#### Create a DataFrame
data = pd.DataFrame({
    'income': [20000, 30000, 40000, 100000, 500000]
})

print("=========================================Original Income==============================================================")

print(data)

print("=========================================Log form of Income===========================================================")

#### Apply log transformation of the dataframe

data['log_income'] = np.log1p(data['income'])

print(data)
```

#### OBSERVATIONS:

1.  The log transformation function has been applied on the original income column of the datframe and then we have obtained the log form of every value of the income column.

```python
#### Apply Power Transformation on the original column 'income'

from sklearn.preprocessing import PowerTransformer

print(data)

print("========================Power form of Income===============================================================")

#### create an object for PowerTransformer

power = PowerTransformer(method = 'yeo-johnson')

data['power_income'] = power.fit_transform(data[['income']])

print(data)
```

#### OBSERVATIONS:

1.  The power transformation has been applied on the income column and every value of the income column has been converted into the normalized form by applying the Power Transformation.
