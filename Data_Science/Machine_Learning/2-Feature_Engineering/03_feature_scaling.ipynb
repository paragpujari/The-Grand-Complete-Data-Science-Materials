#### Scaling

Scaling refers to the technique of converting all the input feature values in the same scale range.

#### Types of Scaling

1. Standardization


    (a.)    It is used when the data is normalized.

    (b.)    It is used when the mean is 0 and standard deviation is 1.

    (c.)    Range = [0,1]

```python
import pandas   as pd
from  sklearn.preprocessing     import  StandardScaler

print("==========================Original DataFrame=====================================================")

data = pd.DataFrame({
    'age': [10, 20, 30, 40, 100]
})

print(data)


print("==========================Scaled DataFrame=======================================================")

#### create an object for Standard Scaler

sc = StandardScaler()

data['age_scaled'] = sc.fit_transform(data[['age']])

print(data)
```

#### OBSERVATIONS:

1.  Here Standard Scaler has been applied on the age column.

2.  All the values in the age column has been converted in the same scale range using Standard Scaler.


2. MinMaxScaler


    (a.)    It is used in neural networks.

    (b.)     Range = [-1,1].

```python
import  pandas  as  pd

print("=====================Original DataFrame=============================================")

data = pd.DataFrame({
    'age': [10, 20, 30, 40, 100]
})


print(data)

print("=======================Min Max Scaled DataFrame====================================")

from sklearn.preprocessing  import  MinMaxScaler

#### create an object for Min Max Scaler
minmax = MinMaxScaler()

data['age_minmax'] = minmax.fit_transform(data[['age']])

print(data)
```
