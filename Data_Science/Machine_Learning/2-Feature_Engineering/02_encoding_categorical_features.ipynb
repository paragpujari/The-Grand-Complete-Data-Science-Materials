###  Encoding

Encoding refers to the technique of converting the catgeorical features into the numerical formats.

###  Type of Encoding

1.  Ordinal Encoding --------------> It can handle ordered data.

2.  One Hot Encoding ---------------> It can handle nominal data.

3.  Label Encoding ---------------> It can handle ordinal data.

### (a.)  Label Encoding

It refers to the process of converted the labels into the numerical format.

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

from    sklearn.preprocessing    import  LabelEncoder


print("============================Original DataFrame===========================================")

data = pd.DataFrame({
    'color': ['red', 'blue', 'green', 'blue']
})

print(data)



#### create an object for LabelEncoder

label = LabelEncoder()


print("===========================Label Encoded Form ==========================================")

data['label_color'] = label.fit_transform(data[['color']])

print(data)
```

#### OBSERVATIONS:

1.  The Label encoding has been applied on the color column of the dataframe and seperate column for labels has been formed in the dataset where


     (a.)        blue ---------------->   Label 0

     (b.)        green --------------->   Label 1

     (c.)        red   --------------->   Label 2


2. Here each value in the column of categorical type has been converted into the numerical format.

### (b.)  One-Hot Encoding

```python
#### Apply One Hot Encoding on the 'color' column of the dataframe


print("===============================Original Dataframe==============================================")

print(data['color'])

print("==============================One Hot Encoded form of Dataframe ===============================")

data_onehot = pd.get_dummies(data['color'])
```

```python
data_onehot
```

#### OBSERVATIONS:

1.  One Hot Encoding has been applied on the original dataframe.

2.  The values present in the color column has been converted into the seperate columns.

3.  The color value present at the particular index position, that color has been marked as True and the rest has been marked as False.
