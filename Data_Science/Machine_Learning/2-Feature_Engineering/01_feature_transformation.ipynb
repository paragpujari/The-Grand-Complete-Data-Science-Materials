#### Transformation:

It can be defined as the technique of changing the shape of the input data or the distribution of the input data to make it more suitable in such a manner that it can be more easily be trained using Machine Learning Models.

#### Importance of Transformations:---

1.  To change the shape of the input data so that it can be more easily be trained by the Machine Learning models.

2.  To provide stability to the model and improve the accuracy of the model.

3.  To reduce the effect of outliers.

4.  To normalize the outliers.

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

from    sklearn.preprocessing  import  PowerTransformer


#### Create a DataFrame
data = pd.DataFrame({
    'income': [20000, 30000, 40000, 100000, 500000]
})

print("=========================================Original Income==============================================================")

print(data)



print("======================================== Log Transformed Value of income =============================================")

data['log_income'] = np.log1p(data['income'])

print(data)
```

#### OBSERVATIONS:

1.  Here log function has been applied on the income column and we have obtained the income as well as the log value of income column.

```python
import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns

from    sklearn.preprocessing  import  PowerTransformer


#### Create a DataFrame
data = pd.DataFrame({
    'income': [20000, 30000, 40000, 100000, 500000]
})

print("=========================================Original Income==============================================================")

print(data)

print(data['income'].ndim)

print("========================================Power Transformation on the Income Column======================================")

#### create an object for Power Transformer

power = PowerTransformer(method = 'yeo-johnson')

data['power_income'] = power.fit_transform(data[['income']])

print(data)
```

#### OBSERVATIONS:

1.  The power transformation has been applied on the income column and every value of the income column has been converted into the normalized form using Power Transformations.
