#### Feature Extraction

Feature Extraction refers to the technique of deriving new features from the old and existing features so as to improve tha accuracy and the performance of the model.

#### Importance of Feature Extraction:----

(a.)  To remove the noise.

(b.)  To reduce the dimensionality of the model.

(c.)  To improve the performance and accuracy of the model.

(d.)  To derive new features from the old features so as to enhance the model further.

```python
from    sklearn.preprocessing   import  StandardScaler
from    sklearn.decomposition   import  PCA
import  pandas as pd
import numpy   as np

#### Create an original dataframe
print("====================================Original DataFrame================================================================")
data = pd.DataFrame({
    'x1': [1,2,3,4,5],
    'x2': [2,4,6,8,10]
})

print(data)

print("===================================Scaled Data=========================================================================")

#### performing the scaling on the original dataframe

#### Creating an object for Standard Scaler

sc = StandardScaler()


scaled_data = sc.fit_transform(data)

print(scaled_data)


print("================================Performing the Dimensionality Reduction on the Scaled Data=============================")

### Creating an object for PCA

pca = PCA(n_components=1)

scaled_pca_data = pca.fit_transform(scaled_data)

print(scaled_pca_data)
```

#### OBSERVATIONS:

1.  Scaling is done on the original dataframe so as to bring all the feature values in the same scale.

2.  Then dimensionality reduction has been applied on the scaled data so as to derive a new feature from the given two input features.

#### Text Feature Extraction

```python
from  sklearn.feature_extraction.text import TfidfVectorizer

#### create an object for TfiDFVectorizer

tfidf = TfidfVectorizer()

### define the corpus

corpus = ["I love ML", "ML is powerful"]

#### using tfidf vectorizer, trasnform and scale the data

X = tfidf.fit_transform(corpus)

print(X)


#### get all the feature names

print("Feature names are :", tfidf.get_feature_names_out())
```
