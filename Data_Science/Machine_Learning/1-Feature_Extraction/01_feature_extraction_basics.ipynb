### Feature Extraction

Feature Extraction refers to the technique of deriving new features from the old and existing features so as to improve tha accuracy and the performance of the model.

Importance of Feature Extraction:----

(a.) To remove the noise.

(b.) To reduce the dimensionality of the model.

(c.) To improve the performance and accuracy of the model.

(d.) To derive new features from the old features so as to enhance the model further.

```python
from    sklearn.preprocessing   import  StandardScaler
from    sklearn.decomposition   import  PCA
import  pandas as pd
import  numpy   as np

#### Create an original dataframe
print("====================================Original DataFrame================================================================")
data = pd.DataFrame({
    'x1': [1,2,3,4,5],
    'x2': [2,4,6,8,10]
})

print(data)

print("====================================Scaled Data=======================================================================")

sc = StandardScaler()

scaled_data = sc.fit_transform(data)

print(scaled_data)

print("===================================Extracted Data====================================================================")

pca = PCA(n_components=1)

extracted_data = pca.fit_transform(scaled_data)

print(extracted_data)
```

#### OBSERVATIONS:

1.  We have the original dataframe

2.  Then feature scaling has been performed on the original dataframe to bring down all the values in the same scale.

3.  PCA has been applied on the scaled data having two features to extract one new feature derived from it.

#### Text Feature Extraction

```python
from  sklearn.feature_extraction.text   import  TfidfVectorizer


#### create an object for TfidfVectorizer

tfidf = TfidfVectorizer()

### define the corpus

corpus = ["I love ML", "ML is powerful"]

print(corpus)


#### using tfidf vectorizer, transform and scale the inputs

X_tfidf = tfidf.fit_transform(corpus)

print(X_tfidf)
```

#### OBSERVATIONS:

1. An object for tfidf vectorizer has been created.

2. It has been applied on the corpus text and is converted into vectors along with the weight associated.

3. The vector having the highest weight is given the most importance.
