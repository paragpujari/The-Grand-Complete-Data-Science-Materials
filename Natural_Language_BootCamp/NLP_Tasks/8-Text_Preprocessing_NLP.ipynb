### Text Preprocessing

Text Preprocessing refers to the process of cleaning the text into clean and usuable form that can be easily used by the Machine Learning Models.


#### Steps used in Text Preprocessing  :-----

1.   Import  all the necessary libraries

2.   Download the necessary NLTK resources

3.   Define a sample text

4.   Convert text to lowercase

5.   Remove punctuation

6.   Perform the sentence tokenization

7.   Perform the word Tokenization

8.   Remove the stopwords

9.   Perform the Stemming process

10.   Perform the Lemmatization process

### Step 1:  Import  all the necessary libraries

```python
import   numpy              as    np
import   pandas             as    pd
import   matplotlib.pyplot  as    plt
import   seaborn            as    sns

import   nltk

from     nltk.tokenize     import  word_tokenize, sent_tokenize, RegexpTokenizer
from     nltk.stem         import  PorterStemmer, WordNetLemmatizer
```

### OBSERVATIONS:

1.  numpy    ---------------->    Computation of numerical  array

2.  pandas   ---------------->    Data Creation and Manipulation

3.  matplotlib -------------->    Data Visualization

4.  seaborn    -------------->    Data Correlation

5.  nltk      --------------->    Library for NLP used for performing the text preprocessing

6.  word_tokenize ----------->    convert the sentence into words

7.  sent_tokenize ----------->    convert the paragraphs into the sentences

8.  RegexpTokenizer --------->    preprocessing the regular expressions by removing the punctations and special symbols

9.  PorterStemmer ----------->    word to root word that may or may not have a meaning

10.  WordNetLemmatizer   ------>  word to root word that has a meaning

### Step 2:  Download the necessary NLTK resources

### Step 3:  Define a sample text

```python
text = """
Natural Language Processing (NLP) is a sub-field of Artificial Intelligence (AI) 
that focuses on the interaction between computers and humans through language. 
The ultimate objective of NLP is to read, decipher, understand, and make sense of human languages.
"""
```

```python
text
```

### OBSERVATIONS:

1. The text is printed.

### Step 4:  Convert text to lowercase

```python
### Convert the text to lower case

text = text.lower()
```

```python
text
```

### OBSERVATIONS:

1. The entire text is converted into the lower case.

### Step 5: Remove punctuation

```python
### create an object for Regular expression

reg = RegexpTokenizer(r'\w+')

### using the object of regular expression, normalize the text

ans = reg.tokenize(text)
```

```python
ans
```

```python
### Convert the text in lists to string

ans = " ".join(ans)
```

```python
ans
```

### OBSERVATIONS:

1.  We have the text as an input.

2.  Regular expression is applied on the text, it is regularized and all the punctuations and special symbols are removed from the text.

3. Now we have a cleaned text.

### Step 6:  Perform the sentence tokenization

```python
paragraph = """
Natural Language Processing (NLP) is a sub-field of Artificial Intelligence (AI) 
that focuses on the interaction between computers and humans through language. 
The ultimate objective of NLP is to read, decipher, understand, and make sense of human languages.
"""
```

```python
paragraph
```

```python
import re

ans = re.sub(r'[\n]','',paragraph)
```

```python
ans
```

```python
### perform the sentence tokenization

sentences = sent_tokenize(ans)
```

```python
sentences
```

### OBSERVATIONS:

1. Here we have first performed regular expression on the input text.

2. We have removed the special symbol '\n' from the text.

3. We have performed the sentence tokenization on the input text and broken the paragraphs into the sentences.

4. The sentence tokenization  accepts the input in the form of string.

5. The output is the sentences in the form of the lists.

### Step 7:  Perform the word Tokenization

```python
### perform the word tokenization

words = word_tokenize(ans)
```

```python
words
```

### OBSERVATIONS:

1. Here word tokenization is performed on the input text.

2. The input text is converted into the words in the form of ists.

### Step 8:  Remove the stopwords

```python
from nltk.corpus import stopwords

### define the english stopwords

english_stopwords = stopwords.words("english")
```

```python
english_stopwords
```

```python
### filter out all the english stopwords from the text

res = [x for x in words if(x not in english_stopwords)]
```

```python
res
```

```python
### convert the filtered text from list to strings

res = " ".join(res)
```

```python
res
```

```python
### create an object for Regular expression

reg = RegexpTokenizer(r'\w+')

### using the object of regular expression, normalize the text

ans = reg.tokenize(res)
```

```python
ans
```

### OBSERVATIONS:

1. All the english stopwords are defined.

2. These stopwords are removed from the text.

3. The filtered text in in lists.

4. It is converted from the lists to string.

5. Then the regular expression tokenization is applied on the filtered text string .

6. All the punctuations and special symbols are emoved from the text.

7. The filetered text will be free from the punctuations, special symbols and full stop.

### Step 9: Perform the Stemming process

```python
from nltk.stem  import PorterStemmer


#### create an object for Porter Stemmer

ps = PorterStemmer()

### perform the stemming on each and every word in the filtered text

ans_stem = [(x,ps.stem(x)) for x in ans]

print(ans_stem)
```

### OBSERVATIONS:

1. The object for Porter Stemmer is created.

2. The filtered text is looped word by word.

3. The stemmer accepts input word by word and reduces the word to the root word.

4.  The root word may or may not have a meaning.

    Ex ----->  (i)   has a meaning         ======>   ('Processing') ---------> ('Process')  has a valid root word

               (ii)  do not have a meaning ======>   ('Artificial') ---------> ('artifici')  do not have a valid root word

### Step 10:  Perform the Lemmatization process

```python
from nltk.stem import WordNetLemmatizer


### Create an object for WordNetLemmatizer

lemma = WordNetLemmatizer()

ans_lemma = [(x,lemma.lemmatize(x)) for x in ans]

print(ans_lemma)
```

### OBSERVATIONS:

1. The object for Porter Stemmer is created.

2. The filtered text is looped word by word.

3. The stemmer accepts input word by word and reduces the word to the root word.

4.  The root word has have a meaning.
