```python
### Calculate the range for the column 'OVERALL_LI'

education_districtwise.info()
```

```python
### To find out the range for the column 'OVERALL_LI' in education_districtwise


min_literacy_rate = education_districtwise['OVERALL_LI'].min()

print("Minimum literacy rate of Overall literacy is:", min_literacy_rate)
```

```python

max_literacy_rate = education_districtwise['OVERALL_LI'].max()

print("Maximum literacy rate of Overall literacy is:", max_literacy_rate)
```

```python
### Range of the column 'OVERALL_LI'

range = max_literacy_rate - min_literacy_rate

print("Difference between the minimum and maximum literacy rate is:", range)
```
