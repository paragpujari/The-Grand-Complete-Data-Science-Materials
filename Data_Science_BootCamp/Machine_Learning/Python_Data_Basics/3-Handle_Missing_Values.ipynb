#### Handling Missing Values in Employee Salary Dataset

```python
##### import all the necessary libraries

import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns
```

```python
#### Load the dataset

df = pd.read_csv("salary_data_with_nulls.csv")
```

```python
df
```

#### OBSERVATIONS:

1.  The above dataset is a employees salary dataset with the salary earned by the employees based on the years of experience.

2.  This dataset contains the NULL Values in both the YearsExperience column and in the Salary column.

```python
#### Replacing all the NULL Values in YearsExperience with the mean value of it

df['YearsExpnew'] = df['YearsExperience'].fillna(df['YearsExperience'].mean())
```

```python
df
```

```python
### Drop the column 'YearsExperience' having the NULL Values

df.drop(columns = 'YearsExperience', inplace=True)
```

```python
df['YearsExperience'] = df['YearsExpnew']
```

```python
df
```

```python
### Drop the column 'YearsExperience' having the NULL Values

df.drop(columns = 'YearsExpnew', inplace=True)
```

```python
df
```

```python
#### Replace all the NULL Values in Salary column with the median value of it

df['Salary_med'] = df['Salary'].fillna(df['Salary'].median())
```

```python
df
```

```python
#### Drop the column Salary having the NULL Values 

df.drop(columns = 'Salary', inplace=True,axis=1)
```

```python
df['Salary'] = df['Salary_med']
```

```python
#### Drop the column Salary having the NULL Values 

df.drop(columns = 'Salary_med', inplace=True,axis=1)
```

```python
df
```

#### OBSERVATIONS:

1.  All the NULL Values in the YearsExperience and Salary has been replaced by the 


     (a.)   mean value  of the YearsExperience

     (b.)   median value of the Salary

```python
df['Category'] = 'A'
```

```python
####make every second row as NULL

df.loc[::2,'Category'] = np.nan
```

```python
df
```

```python
#### Find the mode value of the Category column

mode_val = df['Category'].mode()[0]

print(mode_val)
```

```python
#### Reaplce all the NULL Value with the mode value of the Ctageory column

df['Category'] = df['Category'].fillna(df['Category'].mode()[0])
```

#### OBSERVATIONS:

1.  All the NULL Values in the category column has been filled with the mode value of it.
