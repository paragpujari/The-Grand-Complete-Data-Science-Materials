#### Perform the Standard Scaling on the dataset

```python
#### import  all the important features

import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns


from   sklearn.preprocessing  import  StandardScaler, MinMaxScaler
```

```python
#### Load the dataset

df = pd.read_csv("diabetes.csv")
```

```python
df
```

#### OBSERVATIONS:

1.  The above dataset explains about the diabetic situation for each and every patient and tells whether the patient is suffering from diabetes or not.

```python
#### Divide the dataset into independent and dependent features

X = df.drop(columns='Outcome', axis=1)
```

```python
X
```

```python
Y = df[['Outcome']]
```

```python
Y
```

#### Perform the Standard Scaling on the inputs

##### (a.)  Standard Scaler

```python
from  sklearn.preprocessing     import  StandardScaler

#### create an object for Standard Scaler

sc = StandardScaler()


#### using the object of standard scaler, transform the inputs

X_scaled = sc.fit_transform(X)
```

```python
X_scaled
```

#### OBSERVATIONS:

1.  After applying Standard Scaler on the inputs , all these inputs has been transformed into the scale of one range from 0 to 1.

#### (b.)  Min Max Scaler

```python
from  sklearn.preprocessing  import MinMaxScaler

#### create an object for min max scaler

minmax = MinMaxScaler()


#### using the object for min max scaler, transform the inputs

X_minmax = minmax.fit_transform(X)
```

```python
X_minmax
```

#### OBSERVATIONS:

1.  After applying Min max Scaler on the inputs , all these inputs has been transformed into the scale of one range from -1 to 1.
