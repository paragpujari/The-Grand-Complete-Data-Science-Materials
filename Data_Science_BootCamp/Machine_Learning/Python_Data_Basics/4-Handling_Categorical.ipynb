1.  Here one hot encoding has been applied on the 'Education' columns and the values present in the 'Education' column has been converted into seperate columns.

```python
from sklearn.preprocessing import OneHotEncoder

# Reload dataset
df = pd.read_csv("employees_data.csv")

# Initialize encoder
ohe = OneHotEncoder(sparse_output=False)

# Transform Department column
encoded = ohe.fit_transform(df[['Department']])

# Convert to DataFrame
encoded_df = pd.DataFrame(encoded, columns=ohe.get_feature_names_out())

# Merge with original data
df_final = pd.concat([df.drop('Department', axis=1), encoded_df], axis=1)

print(df_final.head())
```
