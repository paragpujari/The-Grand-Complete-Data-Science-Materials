#### Data Visualization

It helps in displaying the graphical representation for all the data present in the dataset.

```python
#### import all the necessary libraries

import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns
```

##### Data Visualization on

#### (a.)   Salary Data

```python
#### Load the first dataset

df1 = pd.read_csv("Salary_Data.csv")
```

```python
df1
```

#### OBSERVATIONS:

1.  The dataset explains about the salary earned by the employees based on the years of experience.

```python
#### Set the style of the graph

sns.set(style="whitegrid")
```

```python
#### perform the scatter plot between YearsExperience and Salary

plt.figure(figsize=(8,5))

sns.scatterplot(x="YearsExperience", y="Salary", data=df1)

plt.title("Scatter Plot between YearsExperience vs Salary")

plt.show()
```

#### OBSERVATIONS:

1.  There exists a linear relationship between YearsExperience vs Salary.

2.  With the increase in the yearsexperience, the salary earned by the employees also increases.

```python
#### perform the line plot between YearsExperience and Salary

plt.figure(figsize=(8,6))

sns.lineplot(x="YearsExperience",y="Salary",data=df1)

plt.title("Trend of Salary Growth")

plt.show()
```

#### OBSERVATIONS:

1.  There is a linear relationship between the YearsExperience and Salary.

2.  There is a continuous increasing trend of increasing growth of salary earned by the employees based on the YearsExperience.

```python
#### perform the histogram for the salaries earned by the employees

plt.figure(figsize=(8,6))

sns.histplot(df1['Salary'], kde=True)

plt.title("Salary Distribution")

plt.show()
```

#### OBSERVATIONS:

1.  The histogram divided the salary into the bins.

2.  Each bin explains about the frequency / count of all the data points under each it.

3.  The histogram explains about the count/distribution of all the data points that comes under each bin.

4.  As 60000 frequencey count is the highest, so it states that maximum number of employees are earning the salary of around 60000 rupees.

```python
#### perform the box plot for the Salary column

plt.figure(figsize=(8,6))

sns.boxplot(df1['Salary'])


plt.title("Salary Distribution ")

plt.show()
```

#### OBSERVATIONS:

1.  The above graph explains about the salary distribution of the employees with no outliers in it. So the dataset is completely fine.

```python
#### perform the count plot for the Salary_Category column

df1['Salary_Category'] = pd.cut(
    df1['Salary']                         ,
    bins=[0, 40000, 70000, 100000],
    labels=['Low', 'Medium', 'High']
)

print(df1)
```

```python

plt.figure(figsize=(8,6))

sns.countplot(x="Salary_Category", data = df1)

plt.title("Count of Salary Categories")

plt.show()
```

#### OBSERVATIONS:

1.  From the above graph, it is seen that maximum number of employees have the salary category as medium.

2.  It specifies that maximum number of employees are middle class employees and getting a salary of medium class range.

##### Data Visualization on

#### (b.)   Diabates Data

```python
#### import all the necessary libraries

import  numpy               as  np
import  pandas              as  pd
import  matplotlib.pyplot   as  plt
import  seaborn             as  sns
```

```python
#### Load the dataset

df2 = pd.read_csv("diabetes.csv")
```

```python
df2
```

#### OBSERVATIONS:

1.  The above dataset explains about the diabetic situation for each and every patient and even explains whether the patient is suffering from diabetes or not.

```python
#### set the style of the grid

sns.set(style="whitegrid")
```

```python
#### perform the scatter plot between 'Age' and 'Glucose'

plt.figure(figsize=(8,6))

sns.scatterplot(x='Age',y='Glucose', data=df2)

plt.title("Age vs Glucose")

plt.show()
```

#### OBSERVATIONS:

1.  From the above graph, it is seen that when the age is less, then the person is having a huge amount of Glucose concentration in their body.

2.  The glucose concentration of the person decreases when the age of the person increases.

```python
#### perform the line plot between 'Age' and 'Glucose'


plt.figure(figsize=(8,6))

sns.lineplot(x='Age', y='Glucose', data = df2)

plt.title("Trend of Glucose with Age")

plt.show()
```

#### OBSERVATIONS:

1.  The above graph explains about the seasonal trend in the glucose concentration between 50 to 70 age.

2.  It explains about the variation in the glucose concentration with the increase in the age.

```python
# 3. Distribution Plot

plt.figure(figsize=(8,6))


#### perform the histogram for all the glucose concentration

sns.histplot(df2['Glucose'], kde = True)


plt.title("Glucose Distribution")

plt.show()
```

#### OBSERVATIONS:

1.  The above graph depicts about the person having the high concentration of glucose is ranging between 75 till 150.

```python
#### perform the boxplot for the Glucose concentration


plt.figure(figsize=(8,6))

sns.boxplot(df2['Glucose'])

plt.title("Glucose Boxplot")

plt.show()
```

#### OBSERVATIONS:

1.  There is one outlier for the glucose concentration which is not coming under its range.

```python
# 5. Create Glucose Category (NEW)
df2['Glucose_Category'] = pd.cut(df2['Glucose'],
                               bins=[0, 100, 140, 200],
                               labels=['Normal', 'Prediabetes', 'Diabetes'])


```

```python
print(df2)
```

```python
#### perform the count plot of the Glucose_Catgeory column

plt.figure(figsize=(8,6))

sns.countplot(x="Glucose_Category", data=df2)

plt.title("Count of Glucose Categories")

plt.show()
```

#### OBSERVATIONS:

1.  From the above graph, it is seen that maximum number of patients are coming under Prediabetes  as Glucose_Category.
