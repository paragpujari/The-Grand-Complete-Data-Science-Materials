#### (b.)  Perform the Data Correlation and show its visualization

```python
corr_data = df.corr()

print(corr_data)
```

#### OBSERVATIONS:

1.  There is a very high correlation between YearsExperience and Salary.

```python
#### perform the visualization for the correlation

sns.heatmap(corr_data, annot=True,cmap='coolwarm')
```

1.  The  above graph explains about the graphical representation between the features used in the dataset.

2.  There is a very high correlation between YearsExperience and Salary marked blue in color
